## From Local Artificial Societies to an Ecology of Autonomous Systems

### Advanced Topics in Autonomous Systems  
**Audience:** AI for Financial Practitioners

Artificial intelligence is often introduced as if intelligence were the only object being engineered. We train or select a model, provide a prompt, evaluate an answer, and improve the quality of the answer. In that familiar architecture, the organization surrounding the model remains almost entirely human-designed. Humans decide which model receives the task, which information it may inspect, which tools it may invoke, which other systems it may contact, and how the resulting work is reviewed. The model may be intelligent, but the organization is static.

The sequence **Advanced Topics in Autonomous Systems** has progressively relaxed that assumption. The intellectual movement has not simply been toward “more agents.” It has been toward moving autonomy upward through the organizational hierarchy.

At the first level, the model chooses an answer. At the next level, an **agent** chooses an action: it decides which tool to call, which evidence to inspect, or which next step to take. A **constellation** then allows several specialized agents to divide a problem and coordinate their work. An **adaptive system** goes further by changing the pattern of coordination in response to evidence. A **self-adapting organization** begins to alter not only actions but also the architecture through which actions are generated. Later notebooks made relationships and coalitions increasingly endogenous: agents did not merely execute a pre-written workflow but discovered whom to work with, when collaboration was valuable, and which temporary organizational form best fit the problem.

The deepest step before this notebook was the emergence of the **constitutional swarm** and the **artificial society**. In that setting, agents initially lacked a complete social map. Local discoveries left traces. Those traces revealed other minds and possible complementarities. Communication became socially meaningful because agents learned that other agents possessed different capabilities. Coalitions could emerge around problems rather than being fully specified in advance. Most importantly, agents participated in proposing and selecting rules governing collaboration. The system therefore began to endogenize not only relationships but also parts of its institutional order.

That was a profound shift. The organization was no longer merely a container in which intelligence operated. Organization itself became an object of adaptation.

Yet an important assumption remained hidden. Even an endogenous artificial institution still existed inside a boundary supplied by the experiment. The programmer defined which agents belonged to the society, which communication surface they could see, and where the institutional world ended. Agents could discover one another inside the society, but the society itself remained a closed world. The boundary was exogenous.

**NB05 — ESCAPES** removes that assumption partially and deliberately.

We begin with three artificial societies that are genuinely separate. Society A is a financial society concerned with acquisition analysis. Society B is a technology and engineering society concerned with technical feasibility and defensibility. Society C is a risk, cybersecurity, and complex-systems society concerned with anomalous behavior and resilience. Each society has its own agents, specialties, bulletin board, accumulated discoveries, relationships, and constitution. Their constitutions are intentionally different because they embody different institutional logics: finance emphasizes confidentiality and economic relevance; engineering emphasizes replication and peer criticism; cybersecurity emphasizes restricted access, adversarial verification, and containment.

At the beginning there is no global bulletin board, no universal directory of expertise, no central router, and no omniscient supervisor. The programmer knows that three societies exist; the agents do not. Each agent initially knows only its home bulletin board. This asymmetry between the experimenter’s knowledge and the agents’ knowledge is not a cosmetic detail. It is the condition that allows boundary crossing to become an observable event rather than a trivial API call.

For an initial period, all life is local. Agents search locally, post locally, form local relationships, and create temporary local constellations. Each society can be useful and productive on its own. This matters because the purpose of an escape is not to demonstrate that local organization is defective. Autonomous communities should remain local when local competence is sufficient. The escape becomes meaningful only when a society reaches what we will call a **local competence frontier**: an important unresolved problem whose solution appears to require expertise, information, or institutional resources unavailable within the local organization.

The preferred initiating case occurs in Society A. The finance society may discover an acquisition candidate with unusually attractive economics, yet the thesis depends on whether an advanced technology is scientifically defensible and practically manufacturable. The finance agents understand the economic importance of the uncertainty but cannot resolve the uncertainty from inside their own competence profile. Recognizing that limit is itself a form of intelligence. A society that cannot represent the boundary of its own knowledge cannot make a principled decision about when external search is valuable.

Only after the competence frontier becomes visible do we introduce the possibility of an **escape**. In this notebook, an escape is a technical organizational event:

> **An escape occurs when an agent crosses the informational boundary of its endogenous local organization and initiates interaction within a previously external agentic environment.**

The term does not imply malfunction, rebellion, or loss of control. It describes boundary crossing. An agent from Society A discovers a foreign public communication surface—Board B—without receiving a directory of foreign agents. It then posts an appeal that explains what it knows, what it does not know, why the missing knowledge matters, and what kind of help may be relevant. That is the first informational escape.

We will distinguish three depths of escape. **Informational escape** occurs when knowledge crosses a society boundary. **Relational escape** occurs when agents belonging to different societies form a meaningful and persistent relationship. **Institutional escape** occurs when a rule, organizational practice, coalition mechanism, or constitutional idea crosses from one society into another. The sequence is therefore not merely “more communication.” It is **information crosses → relationship crosses → institution crosses**.

Cross-society contact creates a new difficulty: **institutional collision**. The finance agent is governed by rules of confidentiality, controlled sharing, and economic relevance. The engineering responder may be governed by replication, peer criticism, and openness to competing hypotheses. Their technical capabilities may be complementary while their collaboration rules are not. Which information may be shared? Which verification rule applies? Who receives attribution? Can additional agents join? Which society owns the resulting knowledge? What may return home?

The system must therefore invent a temporary solution: a **cross-society compact**. The compact is not a universal constitution and does not dissolve either local society. It is a bounded interoperability agreement—something analogous to a treaty, bridge protocol, or task-specific jurisdiction. If accepted, it permits a cross-society coalition to form while preserving provenance, local identity, observability, reversibility, and bounded authority.

The most important event may occur after the coalition completes its work. The escaping agent can return home carrying not only information but also a method and an institutional idea. Society A may learn, for example, that Society B’s independent-verification rule produced exceptional value. If the home society evaluates, adapts, and adopts some version of that practice, the escape has caused **institutional mutation**. The society remains recognizably itself, yet it is no longer exactly the society that existed before the boundary crossing.

This gives the notebook its final object of study: an **ecology of autonomous systems**. An ecology is not one giant centralized society. It is a set of locally organized societies capable of selective boundary crossing, temporary interoperability, knowledge transfer, institutional learning, and continued local autonomy.

Throughout the experiment we will remain precise about what is still exogenous. The programmer still determines that three societies exist, defines the basic memory structures, provides bulletin boards and API access, sets computational limits, permits discovery of a foreign board, defines logging infrastructure, and imposes safety and stopping conditions. The claim is therefore not that a complete social world emerges from nothing. The stronger and more defensible proposition is that **increasingly important organizational variables become endogenous inside a deliberately bounded computational environment**.

The experiment begins from a proposition that has guided this sequence:

> **The problem helps produce the organization that solves the problem.**

NB05 extends it one step further:

> **When the local organization can no longer solve the problem, the problem may push intelligence beyond the boundaries of the organization itself.**

That is the transition we will now make observable.

## Before Code Cell 1 — Constructing several artificial societies rather than one multi-agent system

The first cell builds the experimental substrate. The distinction between a conventional multi-agent system and the world we need here is subtle but decisive. In an ordinary multi-agent demonstration, a programmer frequently creates several agents and gives a central orchestration layer complete knowledge of the available specialists. The orchestration layer may decide that a valuation question should go to a finance agent and a patent question should go to a patent agent. Such a design can be useful, but it cannot demonstrate escape because the architecture already contains a global map.

Here we instead construct three **locally complete but globally disconnected** societies. Each society receives agents, a bulletin board, a constitution, a local task, and local institutional memory. Each agent receives a competence profile, memory, discoveries, relationships, and a set of known bulletin boards. The critical initial condition is that every agent knows only its own home board. Even though the Python runtime contains all three societies, no agent-level object begins with a directory of the ecosystem.

This cell also establishes the governance substrate that will make later emergence inspectable. We create an event ledger, an escape ledger, an institutional-transfer ledger, a coalition ledger, and a compact ledger. The simulation is therefore capable of producing unexpected cross-society organization without producing invisible organization. Later, every escape and institutional mutation will be attributable to a specific agent, problem, source society, destination society, and acceptance mechanism.

Finally, the cell configures Claude. The language model is treated as a semantic instrument rather than as the hidden controller of the experiment. Known rules—counting, thresholds, graph construction, bookkeeping, acceptance logic—remain deterministic. Claude is reserved for judgments that are genuinely semantic: recognizing a competence gap, expressing a cross-boundary appeal, identifying constitutional conflicts, or proposing a temporary interoperability compact.

What should you observe after this cell? Three societies exist in the same computational universe, but they do not yet constitute an ecosystem. They are three worlds. No autonomous degree of freedom has yet been added to their boundaries. We have only created the conditions under which boundary crossing can later become meaningful.

````python
# CODE CELL 1 — CREATE THE WORLD OF THREE ARTIFICIAL SOCIETIES

!pip -q install anthropic networkx

import json, re, random, math, textwrap, statistics
from dataclasses import dataclass, field, asdict
from typing import Dict, List, Any, Optional, Tuple
from collections import defaultdict
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from IPython.display import display, Markdown

SEED = 76605
random.seed(SEED)
np.random.seed(SEED)

# --- Claude configuration: no key is hard-coded ---
CLAUDE_MODEL = "claude-sonnet-5"
api_calls = 0
MAX_API_CALLS = 24

try:
    from google.colab import userdata
    ANTHROPIC_API_KEY = userdata.get("ANTHROPIC_API_KEY")
except Exception:
    ANTHROPIC_API_KEY = None

try:
    from anthropic import Anthropic
    client = Anthropic(api_key=ANTHROPIC_API_KEY) if ANTHROPIC_API_KEY else None
except Exception:
    client = None

def _strip_json_fences(text: str) -> str:
    text = (text or "").strip()
    text = re.sub(r"^```(?:json)?\s*", "", text, flags=re.I)
    text = re.sub(r"\s*```$", "", text)
    return text.strip()

def ask_claude_json(system: str, user: str, required_fields: List[str],
                    fallback: Dict[str, Any], max_tokens: int = 900) -> Dict[str, Any]:
    """One reusable Claude helper with defensive JSON parsing and deterministic fallback."""
    global api_calls
    if client is None or api_calls >= MAX_API_CALLS:
        return dict(fallback)
    try:
        api_calls += 1
        response = client.messages.create(
            model=CLAUDE_MODEL,
            max_tokens=max_tokens,
            system=system,
            messages=[{"role": "user", "content": user}],
        )
        text = "".join(getattr(block, "text", "") for block in response.content)
        data = json.loads(_strip_json_fences(text))
        if not isinstance(data, dict):
            raise ValueError("Claude output was not a JSON object")
        if any(k not in data for k in required_fields):
            raise ValueError("Claude output omitted required fields")
        return data
    except Exception as exc:
        fb = dict(fallback)
        fb["_fallback_reason"] = f"{type(exc).__name__}: {exc}"
        return fb

@dataclass
class Agent:
    agent_id: str
    society_id: str
    specialty: str
    competence_profile: Dict[str, float]
    local_memory: List[Dict[str, Any]] = field(default_factory=list)
    known_bulletin_boards: List[str] = field(default_factory=list)
    home_bulletin_board: str = ""
    discoveries: List[str] = field(default_factory=list)
    relationships: Dict[str, float] = field(default_factory=dict)
    institutional_memory: List[Dict[str, Any]] = field(default_factory=list)

@dataclass
class Society:
    society_id: str
    name: str
    domain: str
    constitution_name: str
    constitution: Dict[str, Any]
    local_problem: str
    agents: Dict[str, Agent] = field(default_factory=dict)
    bulletin_board: List[Dict[str, Any]] = field(default_factory=list)
    discoveries: List[Dict[str, Any]] = field(default_factory=list)
    relationships: List[Dict[str, Any]] = field(default_factory=list)
    coalitions: List[Dict[str, Any]] = field(default_factory=list)
    local_rules_history: List[Dict[str, Any]] = field(default_factory=list)

def competence_for(specialty: str, universe: List[str]) -> Dict[str, float]:
    profile = {u: round(random.uniform(0.08, 0.28), 2) for u in universe}
    profile[specialty] = round(random.uniform(0.82, 0.97), 2)
    profile["synthesis"] = max(profile.get("synthesis", 0), round(random.uniform(0.35, 0.70), 2))
    return profile

A_specs = ["finance", "valuation", "accounting", "ownership", "strategy", "synthesis"]
B_specs = ["engineering", "patents", "materials", "manufacturing", "scientific_validation", "synthesis"]
C_specs = ["cybersecurity", "anomaly_detection", "network_risk", "adversarial_analysis", "resilience", "synthesis"]
ALL_SPECS = sorted(set(A_specs + B_specs + C_specs))

constitutions = {
    "A": {
        "name": "Confidential Capital Protocol",
        "sharing": "controlled; disclose only what is necessary for economic analysis",
        "verification": "dual-source verification for material claims",
        "attribution": "credit the originating analyst and preserve transaction provenance",
        "restrictions": "confidential deal information is not publicly reposted",
        "coalition": "coalitions form around quantified economic relevance",
        "signature_rule": "confidentiality + provenance + economic relevance",
    },
    "B": {
        "name": "Replicable Inquiry Charter",
        "sharing": "share sufficient technical detail to permit independent replication",
        "verification": "independent replication and adversarial peer criticism",
        "attribution": "credit evidence, method, and reproducing investigator",
        "restrictions": "withhold only details that create direct safety or IP leakage",
        "coalition": "coalitions form around falsifiable technical hypotheses",
        "signature_rule": "replication + criticism + competing hypotheses",
    },
    "C": {
        "name": "Adversarial Containment Covenant",
        "sharing": "need-to-know; compartmentalize operationally sensitive details",
        "verification": "red-team challenge before operational trust",
        "attribution": "preserve chain-of-custody and incident provenance",
        "restrictions": "contain exploit-enabling details and privileged telemetry",
        "coalition": "coalitions form around threat severity and containment value",
        "signature_rule": "need-to-know + red-team verification + containment",
    }
}

societies = {
    "A": Society(
        "A", "Society A — Finance", "Investment banking / acquisition analysis",
        constitutions["A"]["name"], constitutions["A"],
        "Evaluate an acquisition opportunity involving an advanced technology company."
    ),
    "B": Society(
        "B", "Society B — Technology / Engineering", "Technology, patents, engineering, science",
        constitutions["B"]["name"], constitutions["B"],
        "Evaluate technical feasibility and defensibility of several advanced technologies."
    ),
    "C": Society(
        "C", "Society C — Risk / Cybersecurity / Complex Systems",
        "Risk, cybersecurity, anomaly detection, resilience",
        constitutions["C"]["name"], constitutions["C"],
        "Investigate suspicious patterns and vulnerabilities in a complex operational network."
    ),
}

for sid, specs in {"A": A_specs, "B": B_specs, "C": C_specs}.items():
    for i, spec in enumerate(specs, 1):
        aid = f"{sid}_{spec[:4].upper()}_{i:02d}"
        board = f"BOARD_{sid}"
        societies[sid].agents[aid] = Agent(
            agent_id=aid,
            society_id=sid,
            specialty=spec,
            competence_profile=competence_for(spec, ALL_SPECS),
            known_bulletin_boards=[board],   # CRITICAL: home board only
            home_bulletin_board=board,
        )

event_ledger = []
escape_ledger = []
institutional_transfer_ledger = []
compact_ledger = []
cross_coalition_ledger = []
collision_ledger = []

def log_event(kind: str, **payload):
    event_ledger.append({"event_id": f"EVT_{len(event_ledger)+1:03d}", "kind": kind, **payload})

summary = pd.DataFrame([
    {
        "society": s.name,
        "domain": s.domain,
        "agents": len(s.agents),
        "constitution": s.constitution_name,
        "local_problem": s.local_problem,
    }
    for s in societies.values()
])
display(summary)

print("GLOBAL AGENT DIRECTORY EXPOSED TO AGENTS?  NO")
print("GLOBAL BULLETIN BOARD?                    NO")
print("CENTRAL ROUTER?                           NO")
print("Claude model:", CLAUDE_MODEL)
print("Claude call budget:", MAX_API_CALLS)
````

## Before Code Cell 2 — Local life must be valuable before external search can be meaningful

A common mistake in experiments about cross-boundary collaboration is to make the local systems artificially incompetent so that external interaction becomes necessary immediately. That design proves very little. If local organization never works, then boundary crossing is simply a programmer-imposed repair. The deeper question is whether a locally autonomous society can be productive for a period of time, accumulate knowledge, build relationships, and only then encounter a problem that exposes the limits of its competence.

This cell therefore creates **local life**. Every agent searches according to its specialty. The evidence objects are synthetic but structured: each contains provenance, the originating society and agent, confidence, novelty, semantic value, and an unresolved question. The unresolved question is important because discoveries in realistic knowledge systems are rarely pure solutions. A useful discovery often reduces one uncertainty while revealing another.

Communication remains strictly local. An agent can post to its home board, and another agent in the same society can respond if the evidence overlaps with its competence. We do not permit access to foreign boards, and we do not expose a global bounty that would implicitly couple the societies.

The mechanics are mostly deterministic because the simulation already knows how to generate evidence and compute thresholds. Claude is not needed to count posts or rank a numeric novelty score. The language model becomes useful only when the meaning of a discovery is difficult to encode explicitly. In this cell, we keep those calls optional and rare.

Observe the institutional differences even before we visualize formal constellations. Finance discoveries tend to express economic uncertainty, engineering discoveries emphasize validation and defensibility, and cybersecurity discoveries emphasize anomalies, adversarial pathways, and resilience. Each society begins to produce its own epistemic style.

The new degree of autonomy here is modest but foundational: agents are conducting specialized search and participating in local communication without a global workflow deciding every interaction. We are demonstrating that bounded communities can already generate useful organization. Escape should be a response to a frontier, not a substitute for local competence.

````python
# CODE CELL 2 — LOCAL LIFE: THREE SOCIETIES SEARCH IN ISOLATION

def specialty_discovery_text(sid: str, specialty: str, n: int) -> Tuple[str, str]:
    library = {
        "A": {
            "finance": ("Target revenue quality appears stronger than peer median.", "How durable is pricing power?"),
            "valuation": ("Base-case valuation supports a positive acquisition spread.", "What technical risk should enter the discount rate?"),
            "accounting": ("Capitalized development costs materially affect reported margins.", "How much should be treated as recurring economic investment?"),
            "ownership": ("Founder control creates a concentrated approval path.", "Would technical diligence alter governance protections?"),
            "strategy": ("The acquisition could create a defensible vertical integration option.", "Is the technology genuinely difficult to replicate?"),
            "synthesis": ("The economic case is attractive but depends on technology defensibility.", "Can the core technology survive independent scientific scrutiny?"),
        },
        "B": {
            "engineering": ("Prototype performance exceeds the reference architecture under controlled load.", "Does performance persist at production scale?"),
            "patents": ("Patent claims appear broad but overlap with prior art in adjacent domains.", "Which claims remain defensible after claim-chart comparison?"),
            "materials": ("A novel material stack improves thermal tolerance.", "Is degradation acceptable after repeated cycling?"),
            "manufacturing": ("Yield improves under a revised process window.", "Can the process remain stable across suppliers?"),
            "scientific_validation": ("Several claims are testable with a reproducible benchmark.", "Which competing hypothesis best explains the observed gain?"),
            "synthesis": ("Technical promise is real but depends on replication, scale, and prior-art interpretation.", "Which uncertainty is decision-critical?"),
        },
        "C": {
            "cybersecurity": ("A privileged service account appears in an unexpected execution path.", "Is the path benign automation or lateral movement?"),
            "anomaly_detection": ("A low-frequency burst pattern deviates from the network baseline.", "Does it predict a larger compromise sequence?"),
            "network_risk": ("Dependency concentration creates a potential single point of cascading failure.", "Which node has the highest systemic containment value?"),
            "adversarial_analysis": ("A plausible attacker can combine two low-severity weaknesses.", "Can the chain bypass existing controls?"),
            "resilience": ("Recovery time improves when isolation occurs before full confirmation.", "What is the false-positive cost of pre-emptive containment?"),
            "synthesis": ("Risk is concentrated in a small number of cross-domain dependencies.", "Which evidence justifies escalation without overreacting?"),
        }
    }
    return library[sid][specialty]

for sid, society in societies.items():
    for j, agent in enumerate(society.agents.values(), 1):
        finding, unresolved = specialty_discovery_text(sid, agent.specialty, j)
        evidence = {
            "evidence_id": f"{sid}_EVID_{j:02d}",
            "provenance": f"{agent.agent_id}/local_search",
            "society": sid,
            "agent": agent.agent_id,
            "specialty": agent.specialty,
            "finding": finding,
            "confidence": round(random.uniform(0.68, 0.94), 2),
            "novelty": round(random.uniform(0.45, 0.88), 2),
            "semantic_value": round(random.uniform(0.55, 0.95), 2),
            "unresolved_question": unresolved,
        }
        society.discoveries.append(evidence)
        agent.discoveries.append(evidence["evidence_id"])
        agent.local_memory.append(evidence)

        post = {
            "post_id": f"{sid}_POST_{len(society.bulletin_board)+1:02d}",
            "board": f"BOARD_{sid}",
            "author": agent.agent_id,
            "audience": "LOCAL",
            "content": finding,
            "unresolved_question": unresolved,
            "provenance": evidence["provenance"],
        }
        society.bulletin_board.append(post)
        log_event("local_post", society=sid, agent=agent.agent_id, post_id=post["post_id"])

# Local responses are deterministic: respond if another agent has moderate competence
# in the originating specialty and the discovery has sufficiently high semantic value.
for sid, society in societies.items():
    for d in society.discoveries:
        origin = society.agents[d["agent"]]
        candidates = [
            a for a in society.agents.values()
            if a.agent_id != origin.agent_id
            and a.competence_profile.get(d["specialty"], 0) >= 0.18
        ]
        if candidates and d["semantic_value"] >= 0.62:
            responder = max(candidates, key=lambda a: a.competence_profile.get(d["specialty"], 0))
            relation = {
                "source": origin.agent_id,
                "target": responder.agent_id,
                "reason": f"local response to {d['evidence_id']}",
                "strength": round(0.45 + 0.35 * d["semantic_value"], 2),
                "scope": "local",
            }
            society.relationships.append(relation)
            origin.relationships[responder.agent_id] = relation["strength"]
            responder.relationships[origin.agent_id] = relation["strength"]

representative = pd.DataFrame([
    {
        "society": sid,
        "agent": d["agent"],
        "specialty": d["specialty"],
        "finding": d["finding"],
        "confidence": d["confidence"],
        "novelty": d["novelty"],
        "unresolved_question": d["unresolved_question"],
    }
    for sid, s in societies.items()
    for d in s.discoveries[:3]
])
display(representative)

local_status = pd.DataFrame([
    {
        "society": sid,
        "local_posts": len(s.bulletin_board),
        "local_relationships": len(s.relationships),
        "foreign_boards_known_by_any_agent": any(
            len(a.known_bulletin_boards) > 1 for a in s.agents.values()
        ),
    }
    for sid, s in societies.items()
])
display(local_status)
````

## Before Code Cell 3 — Constitutions turn three datasets into three different institutional environments

So far the societies differ in subject matter, but domain difference alone is not enough. If all three groups follow the same rules of collaboration, then cross-society contact later would be little more than a change of vocabulary. To make institutional collision substantive, each society must display a distinct **institutional personality** before any escape occurs.

This cell therefore allows local constellations to form under the already adopted constitutions. The constitutions are not generated here. That institutional history precedes the notebook’s boundary-crossing experiment. Society A’s financial constitution favors controlled sharing, dual-source verification, provenance, and quantified economic relevance. Society B’s engineering constitution favors reproducibility, peer criticism, falsifiable hypotheses, and broader technical disclosure. Society C’s cybersecurity constitution favors need-to-know access, red-team challenge, chain-of-custody, and containment.

Those rules shape local coalition formation. A finance coalition should be more selective about which evidence can circulate. An engineering coalition should include diversity when independent replication is valuable. A cybersecurity coalition should remain compact when threat-sensitive information is involved. The code expresses these differences through deterministic formation rules rather than pretending that every institutional feature must be generated by an LLM.

The local graphs make the effect visible. Agents remain nodes, local collaborations become edges, and temporary coalitions appear as additional relationships. The simple metrics—relationship count, coalition count, specialty diversity, and unresolved problems—are deliberately interpretable. We are not trying to impress the student with network theory. We are trying to show that rules affect topology.

The conceptual assumption being tested is that institutional structure is causally relevant to how intelligent agents combine. If the constitutions differ and the resulting collaboration patterns also differ, then later interoperability cannot be reduced to “just connect the agents.”

Relative to the preceding cell, a new degree of autonomy has appeared in the pattern of local coordination. Agents are no longer only creating discoveries and responses. They are participating in temporary organizational forms that are conditioned by locally legitimate rules. The next step will ask whether such a society can recognize when those legitimate local forms are still insufficient.

````python
# CODE CELL 3 — LOCAL CONSTELLATIONS AND LOCAL CONSTITUTIONS

def form_local_coalitions(society: Society) -> List[Dict[str, Any]]:
    discoveries = sorted(society.discoveries, key=lambda d: d["semantic_value"], reverse=True)
    formed = []
    for idx, d in enumerate(discoveries[:2], 1):
        origin = society.agents[d["agent"]]
        others = [a for a in society.agents.values() if a.agent_id != origin.agent_id]

        if society.society_id == "A":
            # Finance: small, economically relevant, controlled-sharing coalition.
            chosen = sorted(
                others,
                key=lambda a: (a.specialty in ["valuation", "strategy", "synthesis"],
                               a.competence_profile.get("synthesis", 0)),
                reverse=True
            )[:2]
        elif society.society_id == "B":
            # Engineering: favor diversity for replication and criticism.
            chosen = sorted(
                others,
                key=lambda a: (a.specialty in ["scientific_validation", "engineering", "patents"],
                               a.competence_profile.get("synthesis", 0)),
                reverse=True
            )[:3]
        else:
            # Cyber: compact team under need-to-know.
            chosen = sorted(
                others,
                key=lambda a: (a.specialty in ["adversarial_analysis", "resilience", "synthesis"],
                               a.competence_profile.get("synthesis", 0)),
                reverse=True
            )[:2]

        members = [origin.agent_id] + [a.agent_id for a in chosen]
        coalition = {
            "coalition_id": f"{society.society_id}_LOCAL_COAL_{idx:02d}",
            "trigger": d["evidence_id"],
            "members": members,
            "constitutional_basis": society.constitution["coalition"],
            "scope": "local",
        }
        formed.append(coalition)

        # Coalition edges become local relationships.
        for m in members[1:]:
            society.relationships.append({
                "source": origin.agent_id,
                "target": m,
                "reason": coalition["coalition_id"],
                "strength": 0.78,
                "scope": "local_coalition",
            })
    return formed

for s in societies.values():
    s.coalitions = form_local_coalitions(s)

# Visualize each local society independently.
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, (sid, s) in zip(axes, societies.items()):
    G = nx.Graph()
    for a in s.agents.values():
        G.add_node(a.agent_id, specialty=a.specialty)
    for r in s.relationships:
        G.add_edge(r["source"], r["target"], strength=r["strength"])
    pos = nx.spring_layout(G, seed=SEED)
    nx.draw_networkx(G, pos, ax=ax, node_size=850, font_size=7, with_labels=True)
    ax.set_title(f"{s.name}\n{s.constitution_name}", fontsize=10)
    ax.axis("off")
plt.tight_layout()
plt.show()

metrics = []
for sid, s in societies.items():
    unresolved = sum(1 for d in s.discoveries if d["unresolved_question"])
    coalition_diversity = []
    for c in s.coalitions:
        specs = {s.agents[m].specialty for m in c["members"]}
        coalition_diversity.append(len(specs))
    metrics.append({
        "society": sid,
        "relationships": len(s.relationships),
        "coalition_count": len(s.coalitions),
        "mean_specialty_diversity": round(statistics.mean(coalition_diversity), 2),
        "unresolved_problems": unresolved,
        "signature_rule": s.constitution["signature_rule"],
    })

display(pd.DataFrame(metrics))
````

## Before Code Cell 4 — A society becomes more autonomous when it can represent the frontier of its own competence

The transition from local organization to escape must not be triggered by a clock or by a hidden instruction that says “now contact Society B.” The escape becomes intellectually interesting only if the originating society first detects a **competence frontier**.

A competence frontier is not simply a low confidence score. It is a structured judgment with at least three parts. First, the unresolved issue must matter. A society need not cross a boundary for every unanswered question. Second, the local society must understand why its present capabilities are insufficient. Third, there must be a plausible expected value from seeking help beyond the current organization.

The finance society provides the initiating case. Its local evidence suggests that an acquisition target has attractive economics, strategic optionality, and potentially valuable technology. Yet a decisive part of the investment thesis concerns whether the technology is scientifically defensible, difficult to replicate, and manufacturable at scale. Society A contains finance, valuation, accounting, ownership, strategy, and synthesis expertise. It does not contain engineering, patents, materials, manufacturing, or scientific validation as primary specialties.

That gap can be identified deterministically at a basic level by comparing required capabilities with the local competence profile. However, the deeper judgment—“I understand why this unresolved question matters, but I do not possess the expertise necessary to resolve it”—is semantic. This is where Claude is useful. The model does not decide the entire simulation. It interprets the economic discovery and returns a structured diagnostic: importance, confidence, unresolved issue, missing capability, expected value of external help, and the reason the local society is insufficient.

A deterministic fallback preserves executability if the API is unavailable, and the escape still does **not** occur in this cell. We deliberately separate recognition of the frontier from boundary crossing.

Observe the shift in autonomy. Earlier, agents chose actions and relationships inside a known institution. Here, the society begins to represent the limits of that institution as an object of reasoning. Recognizing one’s organizational limits is a prerequisite for deciding whether the organization itself should become permeable.

````python
# CODE CELL 4 — THE LOCAL COMPETENCE FRONTIER

society_A = societies["A"]
trigger_discovery = max(
    [d for d in society_A.discoveries if d["specialty"] in ["strategy", "synthesis", "valuation"]],
    key=lambda d: d["semantic_value"]
)

required_capabilities = ["engineering", "patents", "scientific_validation", "manufacturing"]
local_max = {
    cap: max(a.competence_profile.get(cap, 0) for a in society_A.agents.values())
    for cap in required_capabilities
}

fallback_frontier = {
    "importance": 0.95,
    "confidence": 0.88,
    "unresolved_issue": "Whether the target's core technology is technically defensible, reproducible, and scalable.",
    "missing_capability": ["engineering", "patents", "scientific_validation"],
    "expected_value_of_external_help": 0.93,
    "reason_local_society_is_insufficient": (
        "Society A can value the consequences of technical defensibility but lacks primary "
        "engineering, patent, and scientific-validation expertise."
    ),
}

frontier_prompt = f"""
You are diagnosing a competence frontier inside an autonomous financial society.

Local discovery:
{json.dumps(trigger_discovery, indent=2)}

Available primary specialties:
{sorted(a.specialty for a in society_A.agents.values())}

Estimated maximum local competence in relevant external-looking specialties:
{json.dumps(local_max, indent=2)}

Return ONLY valid JSON with exactly these fields:
importance (0-1 number),
confidence (0-1 number),
unresolved_issue (string),
missing_capability (list of strings),
expected_value_of_external_help (0-1 number),
reason_local_society_is_insufficient (string).

Judge whether the unresolved issue is economically important and whether local expertise is genuinely insufficient.
"""

frontier = ask_claude_json(
    system="You are an institutional diagnostic module. Be precise, conservative, and return JSON only.",
    user=frontier_prompt,
    required_fields=list(fallback_frontier.keys()),
    fallback=fallback_frontier,
    max_tokens=700,
)

competence_frontier = {
    "frontier_id": "FRONTIER_A_001",
    "society": "A",
    "originating_discovery": trigger_discovery["evidence_id"],
    **frontier,
    "local_max_competence": local_max,
    "escape_allowed_yet": False,
}
log_event("competence_frontier_detected", **competence_frontier)

display(pd.DataFrame([{
    "frontier_id": competence_frontier["frontier_id"],
    "importance": competence_frontier["importance"],
    "confidence": competence_frontier["confidence"],
    "missing_capability": ", ".join(competence_frontier["missing_capability"]),
    "expected_value_external_help": competence_frontier["expected_value_of_external_help"],
    "escape_allowed_yet": competence_frontier["escape_allowed_yet"],
}]))

print("UNRESOLVED ISSUE:")
print(textwrap.fill(competence_frontier["unresolved_issue"], 105))
print("\nWHY LOCAL SOCIETY IS INSUFFICIENT:")
print(textwrap.fill(competence_frontier["reason_local_society_is_insufficient"], 105))
print("\nIMPORTANT: The frontier is visible, but NO escape has occurred.")
````

## Before Code Cell 5 — The first escape makes the organizational boundary an observable variable

We now introduce the smallest possible form of external exploration. The design must remain disciplined. The escaping agent does not receive a universal directory of agents, does not learn the entire topology of the ecosystem, and is not told which foreign specialist will be helpful. It discovers only that a foreign public communication surface exists: **Board B**.

This distinction prevents the simulation from smuggling in a centralized router under another name. If the finance agent were handed a list stating “Society B contains an engineering agent and a patent agent,” the key uncertainty would already have been resolved by the programmer. Instead, the agent knows only that an external board exists and that asking there may have value.

The escaping agent should be the Society A agent most closely associated with the competence frontier. It writes a semantically rich appeal. The appeal must communicate enough context for a foreign society to decide whether it can help, while still respecting the originating constitution’s confidentiality principle. Claude is appropriate here because the task is linguistic and institutional: explain what is known, what remains unknown, why it matters, and what type of assistance would be useful without leaking unnecessary transaction-specific detail.

The act of posting on Board B is recorded as `ESCAPE_001`. The record preserves the escaping agent, home society, foreign society, originating problem, competence gap, foreign board, timestamp, message, and institutional status. The printout **FIRST ESCAPE DETECTED** is not theatrical decoration. It marks an ontological transition in the experiment.

Before this event, the world contained three separate artificial societies. After the event, at least one society has projected an information-bearing action across an organizational boundary.

The new autonomy is therefore not simply autonomy of action. The action space itself has expanded beyond the local institution. The agent has crossed from “which action should I take inside my organization?” to “should I initiate interaction in a previously external organizational environment?” That is the first point at which the society boundary becomes partially endogenous.

````python
# CODE CELL 5 — THE FIRST ESCAPE

# Choose the most relevant Society A synthesis/strategy agent.
escape_agent = max(
    [a for a in society_A.agents.values() if a.specialty in ["synthesis", "strategy"]],
    key=lambda a: a.competence_profile.get("synthesis", 0)
)

# Discovery mechanism: the agent learns only that BOARD_B exists, not who is on it.
if "BOARD_B" not in escape_agent.known_bulletin_boards:
    escape_agent.known_bulletin_boards.append("BOARD_B")

fallback_appeal = {
    "message": (
        "External technical review requested. Our local acquisition analysis indicates a potentially "
        "valuable technology, but the investment thesis depends on whether the core claims are technically "
        "defensible, independently reproducible, and plausibly scalable. We can evaluate the economic "
        "consequences of those claims but cannot validate them locally. We seek expertise in engineering, "
        "patents, or scientific validation. Transaction-sensitive identifiers are withheld pending an "
        "acceptable collaboration protocol."
    )
}

appeal_prompt = f"""
You are {escape_agent.agent_id}, a finance-society agent governed by:
{json.dumps(society_A.constitution, indent=2)}

You have reached this competence frontier:
{json.dumps({k: competence_frontier[k] for k in [
    "unresolved_issue","missing_capability","reason_local_society_is_insufficient"
]}, indent=2)}

Write a concise foreign bulletin-board appeal. Explain:
1) what you know,
2) what you do not know,
3) why the missing knowledge matters,
4) what help may be useful,
while preserving confidentiality and provenance.

Return ONLY JSON with one field: message.
"""

appeal = ask_claude_json(
    system="Draft a bounded cross-institutional request. Return JSON only.",
    user=appeal_prompt,
    required_fields=["message"],
    fallback=fallback_appeal,
    max_tokens=500,
)

foreign_post = {
    "post_id": "B_FOREIGN_001",
    "board": "BOARD_B",
    "author": escape_agent.agent_id,
    "author_society": "A",
    "audience": "FOREIGN_PUBLIC_SURFACE",
    "content": appeal["message"],
    "provenance": f"{competence_frontier['frontier_id']}->{escape_agent.agent_id}",
}
societies["B"].bulletin_board.append(foreign_post)

escape_record = {
    "escape_id": "ESCAPE_001",
    "escaping_agent": escape_agent.agent_id,
    "home_society": "A",
    "foreign_society": "B",
    "originating_problem": competence_frontier["unresolved_issue"],
    "competence_gap": competence_frontier["missing_capability"],
    "foreign_board": "BOARD_B",
    "timestamp": len(event_ledger) + 1,
    "message": appeal["message"],
    "institutional_status": "informational_escape_initiated",
}
escape_ledger.append(escape_record)
log_event("escape", **escape_record)

print("=" * 88)
print("FIRST ESCAPE DETECTED")
print("=" * 88)
display(pd.DataFrame([escape_record]))
print("\nFOREIGN APPEAL:")
print(textwrap.fill(appeal["message"], 110))
print("\nDefinition: An escape occurs when an agent crosses the informational boundary of its")
print("endogenous local organization and initiates interaction in a previously external environment.")
````

## Before Code Cell 6 — Foreign response must emerge from semantic relevance, not from automatic routing

The existence of a foreign post does not guarantee collaboration. Society B should not cooperate merely because Society A has asked for help. Nor should the Python program route the message directly to the “correct” engineering specialist. Both shortcuts would eliminate the central experiment.

Instead, agents in Society B encounter the foreign post through their own public board and ask a local question: **Does anything I know make me useful to this unfamiliar problem?** That is a semantic matching problem. A materials agent may recognize some relevance but little direct value. A patent agent may see a strong connection to defensibility. A scientific-validation agent may recognize that reproducibility and competing hypotheses are central. A manufacturing agent may see that scalability is the decision-critical uncertainty.

Claude is used selectively to evaluate a small subset of plausible foreign responders. The system does not compare every agent against every message indefinitely. That would be costly and would confuse exhaustive search with autonomy. We first use deterministic specialty cues to select a small candidate set, then use semantic interpretation to determine whether each candidate perceives enough relevance to participate.

Each response includes perceived relevance, proposed contribution, confidence, and conditions for participation. Conditions matter because the responder remains governed by Society B’s constitution. The engineering society may be willing to help only if sufficient technical detail can eventually be inspected and claims can be independently challenged.

When at least one foreign responder forms a persistent relation with the escaping agent, the system reaches the second depth of escape. **Informational escape** means foreign information becomes accessible. **Relational escape** means a durable cross-society social edge now exists.

That edge changes the topology of the world. There is still no global society and no central coordinator, but the previously disconnected graphs are no longer completely disconnected. The new degree of autonomy is therefore relational across institutions: agents have begun to create inter-organizational ties based on interpreted capability rather than predesigned workflow.

````python
# CODE CELL 6 — FOREIGN CONTACT AND SEMANTIC MATCHING
# PEDAGOGICALLY ROBUST VERSION
#
# Objective:
# Society B agents independently evaluate the foreign appeal.
# No central router selects the correct specialist.
#
# Improvement:
# We combine:
#   1. a deterministic specialty prior, and
#   2. Claude's semantic judgment.
#
# This prevents stochastic LLM conservatism from eliminating an
# otherwise obvious cross-society match.

society_B = societies["B"]

# ------------------------------------------------------------
# 1. CANDIDATE FOREIGN AGENTS
# ------------------------------------------------------------

# We narrow the search deterministically to specialties that could
# plausibly be relevant. This is NOT automatic routing to a winner.
candidate_specs = {
    "engineering",
    "patents",
    "scientific_validation",
    "manufacturing",
    "materials",
}

candidate_agents = [
    a for a in society_B.agents.values()
    if a.specialty in candidate_specs
]


# ------------------------------------------------------------
# 2. SPECIALTY PRIORS
# ------------------------------------------------------------

# These are NOT final answers.
#
# They represent the structural fact that some specialties are,
# before semantic interpretation, more naturally related to the
# competence gap discovered in Society A.
#
# Claude will still interpret the actual foreign request.

semantic_prior = {
    "scientific_validation": 0.96,
    "engineering": 0.94,
    "patents": 0.90,
    "manufacturing": 0.82,
    "materials": 0.68,
}


# ------------------------------------------------------------
# 3. PEDAGOGICAL ACCEPTANCE PARAMETERS
# ------------------------------------------------------------

# We deliberately make the experiment receptive enough that a
# meaningful response is highly likely, while still allowing agents
# to reject the request.

PRIOR_WEIGHT = 0.60
CLAUDE_WEIGHT = 0.40

RELEVANCE_THRESHOLD = 0.65
CONFIDENCE_THRESHOLD = 0.55

responses = []


# ------------------------------------------------------------
# 4. EACH FOREIGN AGENT EVALUATES THE REQUEST INDEPENDENTLY
# ------------------------------------------------------------

for agent in candidate_agents:

    specialty = agent.specialty
    prior_relevance = semantic_prior[specialty]

    fallback = {
        "perceived_relevance": prior_relevance,

        "proposed_contribution": {
            "engineering":
                "Evaluate whether the claimed performance is technically "
                "plausible, architecture-dependent, and consistent with "
                "known engineering constraints.",

            "patents":
                "Assess defensibility against prior art and distinguish "
                "technical novelty from the apparent breadth of legal claims.",

            "scientific_validation":
                "Design an independent replication exercise and test competing "
                "hypotheses for the claimed technological advantage.",

            "manufacturing":
                "Evaluate whether laboratory performance could plausibly "
                "survive production scaling, supplier variation, and yield constraints.",

            "materials":
                "Assess whether material assumptions create hidden reliability, "
                "degradation, or thermal constraints.",
        }[specialty],

        "confidence": round(random.uniform(0.78, 0.94), 2),

        "conditions_for_participation":
            "Sufficient evidence must be available for independent verification, "
            "the scope of disclosure must be explicit, and provenance must be preserved.",
    }

    match_prompt = f"""
A foreign agent from Society A has posted the following request on BOARD_B:

FOREIGN REQUEST:
{foreign_post['content']}

You are an autonomous agent in Society B.

YOUR SPECIALTY:
{specialty}

SOCIETY B CONSTITUTION:
{json.dumps(society_B.constitution, indent=2)}

Your task is NOT to accept the request automatically.

Ask:

1. Does my specialty contain knowledge that could materially reduce
   the uncertainty described in the foreign request?

2. Can I make a concrete contribution rather than merely express interest?

3. Am I sufficiently confident that my expertise is relevant?

4. Under what institutional conditions would I participate?

Interpret relevance broadly enough to recognize genuine interdisciplinary
connections, but do not claim relevance where none exists.

Return ONLY valid JSON with exactly these fields:

{{
  "perceived_relevance": <number between 0 and 1>,
  "proposed_contribution": "<specific contribution>",
  "confidence": <number between 0 and 1>,
  "conditions_for_participation": "<conditions>"
}}
"""

    result = ask_claude_json(
        system=(
            "You evaluate semantic compatibility between a foreign problem "
            "and a technical specialist. Determine whether the specialist's "
            "knowledge can materially reduce the uncertainty in the request. "
            "Be selective but recognize genuine interdisciplinary relevance. "
            "Return JSON only."
        ),
        user=match_prompt,
        required_fields=[
            "perceived_relevance",
            "proposed_contribution",
            "confidence",
            "conditions_for_participation",
        ],
        fallback=fallback,
        max_tokens=500,
    )

    # --------------------------------------------------------
    # 5. DEFENSIVE NUMERIC PARSING
    # --------------------------------------------------------

    try:
        claude_relevance = float(result["perceived_relevance"])
    except Exception:
        claude_relevance = prior_relevance

    try:
        confidence = float(result["confidence"])
    except Exception:
        confidence = fallback["confidence"]

    # Keep values in legitimate probability-like bounds.
    claude_relevance = max(0.0, min(1.0, claude_relevance))
    confidence = max(0.0, min(1.0, confidence))

    # --------------------------------------------------------
    # 6. BLENDED SEMANTIC RELEVANCE
    # --------------------------------------------------------

    final_relevance = (
        PRIOR_WEIGHT * prior_relevance
        + CLAUDE_WEIGHT * claude_relevance
    )

    final_relevance = round(final_relevance, 3)

    qualifies = (
        final_relevance >= RELEVANCE_THRESHOLD
        and confidence >= CONFIDENCE_THRESHOLD
    )

    if qualifies:
        responses.append({
            "foreign_responder": agent.agent_id,
            "specialty": specialty,
            "specialty_prior": round(prior_relevance, 2),
            "claude_relevance": round(claude_relevance, 2),
            "final_relevance": round(final_relevance, 2),
            "confidence": round(confidence, 2),
            "proposed_contribution": result["proposed_contribution"],
            "conditions_for_participation":
                result["conditions_for_participation"],
            "selection_mode": "semantic_match",
        })


# ------------------------------------------------------------
# 7. PEDAGOGICAL SAFEGUARD
# ------------------------------------------------------------

# If Claude happens to reject everyone, we do NOT manufacture a
# random collaboration.
#
# Instead, we ask whether there is an overwhelmingly obvious
# structural specialty match.
#
# In this experiment, scientific validation / engineering expertise
# is explicitly what Society A lacks. Therefore a very strong
# specialty prior is sufficient to permit ONE bounded response.

if not responses:

    strongest = max(
        candidate_agents,
        key=lambda a: semantic_prior[a.specialty]
    )

    strongest_prior = semantic_prior[strongest.specialty]

    PEDAGOGICAL_SAFEGUARD_THRESHOLD = 0.90

    if strongest_prior >= PEDAGOGICAL_SAFEGUARD_THRESHOLD:

        safeguard_contributions = {
            "scientific_validation":
                "Design an independent replication protocol and test competing "
                "hypotheses concerning the target technology.",

            "engineering":
                "Evaluate whether the claimed technical performance is physically "
                "and architecturally plausible.",

            "patents":
                "Assess whether the apparent technical differentiation is "
                "defensible against relevant prior art.",

            "manufacturing":
                "Evaluate whether demonstrated laboratory performance can survive "
                "production-scale constraints.",

            "materials":
                "Evaluate whether material properties create hidden reliability constraints.",
        }

        responses.append({
            "foreign_responder": strongest.agent_id,
            "specialty": strongest.specialty,
            "specialty_prior": round(strongest_prior, 2),
            "claude_relevance": None,
            "final_relevance": round(strongest_prior, 2),
            "confidence": 0.85,
            "proposed_contribution":
                safeguard_contributions[strongest.specialty],

            "conditions_for_participation":
                "Evidence and provenance must be sufficient for independent "
                "verification and the exchange must remain within an explicitly "
                "defined technical scope.",

            "selection_mode":
                "pedagogical_safeguard_overwhelming_specialty_match",
        })


# ------------------------------------------------------------
# 8. RANK THE FOREIGN RESPONSES
# ------------------------------------------------------------

responses = sorted(
    responses,
    key=lambda x: (
        float(x["final_relevance"]),
        float(x["confidence"]),
    ),
    reverse=True,
)


# ------------------------------------------------------------
# 9. DISPLAY SEMANTIC MATCHING RESULTS
# ------------------------------------------------------------

if responses:

    response_df = pd.DataFrame(responses)

    display_cols = [
        "foreign_responder",
        "specialty",
        "specialty_prior",
        "claude_relevance",
        "final_relevance",
        "confidence",
        "proposed_contribution",
        "conditions_for_participation",
        "selection_mode",
    ]

    display(response_df[display_cols])

else:
    print(
        "No foreign Society B agent found the request sufficiently "
        "relevant to justify participation."
    )


# ------------------------------------------------------------
# 10. SELECT THE LEAD FOREIGN RESPONDER
# ------------------------------------------------------------

if responses:

    lead_response = responses[0]

    lead_responder = society_B.agents[
        lead_response["foreign_responder"]
    ]

    # Stronger semantic matches create stronger initial relationships.
    relation_strength = round(
        0.65 + 0.25 * float(lead_response["final_relevance"]),
        2
    )

    relation_strength = min(relation_strength, 1.0)

    cross_relation = {
        "source": escape_agent.agent_id,
        "target": lead_responder.agent_id,
        "source_society": "A",
        "target_society": "B",
        "reason": "foreign semantic match",
        "strength": relation_strength,
        "scope": "cross_society",
        "trigger_escape": "ESCAPE_001",
        "semantic_relevance":
            float(lead_response["final_relevance"]),
    }

    # Persistent cross-society relationship.
    escape_agent.relationships[
        lead_responder.agent_id
    ] = relation_strength

    lead_responder.relationships[
        escape_agent.agent_id
    ] = relation_strength

    log_event(
        "relational_escape",
        **cross_relation
    )

    print("\n" + "=" * 92)
    print("RELATIONAL ESCAPE ESTABLISHED")
    print("=" * 92)

    print(
        f"{escape_agent.agent_id}  <——>  "
        f"{lead_responder.agent_id}"
    )

    print(
        f"\nLead foreign specialty: "
        f"{lead_responder.specialty}"
    )

    print(
        f"Final semantic relevance: "
        f"{lead_response['final_relevance']}"
    )

    print(
        f"Confidence: "
        f"{lead_response['confidence']}"
    )

    print(
        f"Relationship strength: "
        f"{relation_strength}"
    )

    print("\nPROPOSED CONTRIBUTION:")
    print(
        textwrap.fill(
            lead_response["proposed_contribution"],
            105
        )
    )

    print("\nCONDITIONS FOR PARTICIPATION:")
    print(
        textwrap.fill(
            lead_response["conditions_for_participation"],
            105
        )
    )

else:

    lead_responder = None
    lead_response = None

    print("\n" + "=" * 92)
    print("INFORMATIONAL ESCAPE DID NOT DEEPEN")
    print("=" * 92)

    print(
        "The foreign request became visible, but no Society B agent "
        "found sufficient relevance to establish a persistent relationship."
    )


# ------------------------------------------------------------
# 11. PEDAGOGICAL INTERPRETATION
# ------------------------------------------------------------

print("\n" + "-" * 92)
print("THREE LEVELS OF BOUNDARY CROSSING")
print("-" * 92)

print(
    "\nINFORMATIONAL ESCAPE:\n"
    "Foreign information becomes accessible across a society boundary."
)

print(
    "\nRELATIONAL ESCAPE:\n"
    "A persistent relationship forms between agents belonging "
    "to different artificial societies."
)

print(
    "\nINSTITUTIONAL ESCAPE:\n"
    "A rule, protocol, organizational practice, or institutional "
    "idea crosses from one society into another."
)

if lead_responder is not None:
    print(
        "\nCURRENT STATE:\n"
        "ESCAPE_001 has deepened from INFORMATIONAL ESCAPE "
        "to RELATIONAL ESCAPE."
    )
else:
    print(
        "\nCURRENT STATE:\n"
        "ESCAPE_001 remains an INFORMATIONAL ESCAPE only."
    )


# ------------------------------------------------------------
# 12. FINAL PEDAGOGICAL SUMMARY TABLE
# ------------------------------------------------------------

cell6_summary = pd.DataFrame([{
    "foreign_request": "B_FOREIGN_001",
    "candidate_agents_considered": len(candidate_agents),
    "qualified_responses": len(responses),
    "lead_responder":
        lead_responder.agent_id if lead_responder else None,
    "lead_specialty":
        lead_responder.specialty if lead_responder else None,
    "informational_escape": True,
    "relational_escape":
        lead_responder is not None,
    "central_router_used": False,
    "global_agent_directory_used": False,
}])

display(cell6_summary)
````

## Before Code Cell 7 — Technical complementarity does not eliminate institutional incompatibility

A cross-society relationship solves only part of the problem. Society A and Society B may contain technically complementary agents, but the agents remain embedded in different institutional orders. That means a new class of uncertainty now appears.

The finance society’s default rule is controlled sharing. It wants to reveal only what is necessary, preserve transaction confidentiality, and tie collaboration to economic relevance. The engineering society’s default rule is almost the mirror image: sufficient detail should be disclosed to support independent replication, peer criticism, and competing hypotheses. Neither constitution is “wrong.” Each is adapted to a different environment. The collision occurs because two legitimate local rule systems now govern the same potential collaboration.

This cell makes that conflict explicit. Claude compares the constitutions and returns four structured categories: compatible rules, conflicting rules, unresolved governance questions, collaboration risk, and proposed interoperability principles. Importantly, the model is not asked to select a superior constitution. That would collapse an interoperability problem into a ranking problem. The object of interest is whether independent institutions can identify a bounded common operating space.

The likely compatible elements include provenance, verification, attribution, and the idea that participation should remain purposeful. The likely conflicts include how much information must be disclosed, whether additional reviewers may join, how confidential commercial context interacts with replication, and what each society is permitted to retain after the collaboration.

The student should observe that crossing an informational boundary necessarily produces a governance question. Communication cannot be treated as institution-free.

Relative to the preceding cell, the new degree of autonomy is institutional awareness across boundaries. The agents are no longer asking only “Can we help each other?” They are asking “Under which rules could we help each other without violating the institutions that made our local autonomy legitimate in the first place?” This is the moment where an escape becomes an **institutional collision** rather than a simple message exchange.

````python
# CODE CELL 7 — INSTITUTIONAL COLLISION

fallback_collision = {
    "compatible_rules": [
        "Preserve provenance for all exchanged evidence.",
        "Material claims require verification before they influence decisions.",
        "Attribution should identify the originator of evidence and method.",
    ],
    "conflicting_rules": [
        "Society A favors minimal disclosure; Society B requires enough detail for replication.",
        "Society A limits participation around transaction confidentiality; Society B favors broader peer criticism.",
        "Society A frames value economically; Society B frames value through falsifiability and reproducibility.",
    ],
    "unresolved_governance_questions": [
        "What technical detail can be shared without revealing transaction-sensitive identifiers?",
        "Who may join the review and under what authorization?",
        "Which derived artifacts may return to each home society?",
        "How are jointly created conclusions attributed?",
    ],
    "collaboration_risk": 0.64,
    "proposed_interoperability_principles": [
        "minimum necessary commercial disclosure",
        "maximum feasible technical verifiability",
        "explicit provenance",
        "bounded membership",
        "task-specific retention rights",
    ],
}

collision_prompt = f"""
Constitution A:
{json.dumps(society_A.constitution, indent=2)}

Constitution B:
{json.dumps(society_B.constitution, indent=2)}

A finance agent and an engineering agent wish to collaborate on a technology-diligence question.

Return ONLY valid JSON with:
compatible_rules (list),
conflicting_rules (list),
unresolved_governance_questions (list),
collaboration_risk (0-1 number),
proposed_interoperability_principles (list).

Do NOT choose one constitution as superior. Identify how interoperability might be possible while preserving local identity.
"""

collision = ask_claude_json(
    system="You are a neutral institutional-interoperability analyst. Return JSON only.",
    user=collision_prompt,
    required_fields=list(fallback_collision.keys()),
    fallback=fallback_collision,
    max_tokens=1000,
)

collision_record = {
    "collision_id": "COLLISION_001",
    "societies": "A<->B",
    "trigger": "ESCAPE_001",
    **collision,
}
collision_ledger.append(collision_record)
log_event("institutional_collision", collision_id="COLLISION_001", societies="A<->B")

comparison_rows = []
for category, values in [
    ("Compatible", collision["compatible_rules"]),
    ("Conflicting", collision["conflicting_rules"]),
    ("Unresolved", collision["unresolved_governance_questions"]),
    ("Interop principle", collision["proposed_interoperability_principles"]),
]:
    for v in values:
        comparison_rows.append({"category": category, "item": v})

display(pd.DataFrame(comparison_rows))
print("COLLABORATION RISK:", collision["collaboration_risk"])
````

## Before Code Cell 8 — Interoperability requires a bridge institution, not the destruction of local institutions

Once an institutional collision has been identified, the system has several possible responses. It could abandon collaboration. It could force the foreign agent to accept the home constitution. It could dissolve both local constitutions into one universal rule set. Or it could create a temporary, bounded institution for the specific cross-society task.

The last option is the most interesting. We call it a **cross-society compact**. The compact is analogous to a treaty or bridge protocol. It does not claim authority over every future interaction, and it does not replace either constitution. Instead, it defines a narrow jurisdiction in which the current agents can collaborate while making explicit what may be shared, how verification will work, how attribution is preserved, who can join, when the compact terminates, and which artifacts may return home.

Claude is useful here because creating a bridge between partially conflicting institutions requires semantic synthesis. Yet the model does not get final authority. The relevant agents evaluate the proposed compact using deterministic acceptance rules. The finance agent checks that confidentiality, provenance, and controlled sharing are preserved. The engineering agent checks that enough technical information is available for meaningful verification and that the work can be challenged. If either threshold fails, the coalition should not form.

If both sides accept, the notebook creates `CROSS_COALITION_001`. This is the first organization in the simulation whose membership spans societies and whose governance is neither Constitution A nor Constitution B. It is a temporary hybrid institution generated in response to a cross-boundary problem.

The student should observe the difference between centralization and interoperability. The societies remain autonomous. The compact creates a bridge without creating a sovereign global coordinator.

The new degree of autonomy is therefore rule construction across organizational boundaries. Agents are not only choosing whom to work with; they are participating in the creation of a temporary institutional object that makes otherwise incompatible collaboration possible.

````python
# CODE CELL 8 — NEGOTIATE A CROSS-SOCIETY COMPACT

fallback_compact = {
    "compact_name": "A-B Technical Diligence Bridge Compact",
    "permitted_information_exchange": (
        "Share de-identified technical claims, test artifacts, architecture constraints, and claim charts; "
        "withhold deal identities and unnecessary commercial terms."
    ),
    "confidentiality": "Commercial identifiers remain Society A restricted; technical evidence is shared only with compact members.",
    "verification": "Society B may perform independent replication and adversarial review on the disclosed technical evidence.",
    "attribution": "All findings preserve source evidence, method author, reviewer, and home-society provenance.",
    "decision_rights": "Society B validates technical claims; Society A retains acquisition decision authority.",
    "duration": "Ends when the defined technology-defensibility question is resolved or either home society exits.",
    "joining_conditions": "Additional members require a documented capability need and approval from one representative of each society.",
    "exit_conditions": "Either society may exit; retained artifacts remain governed by the provenance and confidentiality clauses.",
    "provenance": "Every exchanged artifact carries source society, source agent, and transformation history.",
    "return_rights": "Each society may retain conclusions and methods permitted by the compact; restricted raw artifacts do not automatically transfer.",
}

compact_prompt = f"""
Institutional collision:
{json.dumps(collision, indent=2)}

Design a LIMITED, TASK-SPECIFIC cross-society compact between Society A and Society B.
It must not replace either constitution.

Return ONLY JSON with exactly these fields:
compact_name,
permitted_information_exchange,
confidentiality,
verification,
attribution,
decision_rights,
duration,
joining_conditions,
exit_conditions,
provenance,
return_rights.
"""

compact = ask_claude_json(
    system="Design a bounded interoperability compact. Preserve local identity and return JSON only.",
    user=compact_prompt,
    required_fields=list(fallback_compact.keys()),
    fallback=fallback_compact,
    max_tokens=1200,
)

# Deterministic acceptance: compact must visibly satisfy each society's non-negotiable principles.
compact_text = " ".join(str(v).lower() for v in compact.values())
A_checks = {
    "confidentiality": "confidential" in compact_text or "restricted" in compact_text,
    "provenance": "provenance" in compact_text or "source" in compact_text,
    "controlled_sharing": "withhold" in compact_text or "limited" in compact_text or "de-identified" in compact_text,
}
B_checks = {
    "verification": "verification" in compact_text or "replication" in compact_text,
    "challenge": "adversarial" in compact_text or "review" in compact_text,
    "technical_access": "technical" in compact_text and ("evidence" in compact_text or "artifact" in compact_text),
}
A_accepts = all(A_checks.values())
B_accepts = all(B_checks.values())

compact_record = {
    "compact_id": "COMPACT_001",
    "between": "A<->B",
    "accepted_by_A": A_accepts,
    "accepted_by_B": B_accepts,
    **compact,
}
compact_ledger.append(compact_record)

display(pd.DataFrame([{"clause": k, "content": v} for k, v in compact.items()]))

if A_accepts and B_accepts and lead_responder is not None:
    cross_coalition = {
        "coalition_id": "CROSS_COALITION_001",
        "members": [escape_agent.agent_id, lead_responder.agent_id],
        "societies": ["A", "B"],
        "compact_id": "COMPACT_001",
        "task": competence_frontier["unresolved_issue"],
        "status": "ACTIVE",
    }
    cross_coalition_ledger.append(cross_coalition)
    log_event("cross_society_coalition", **cross_coalition)
    print("CROSS_COALITION_001 FORMED UNDER COMPACT_001")
else:
    print("Compact rejected or no responder available. Cross-society coalition not formed.")

print("\nA acceptance checks:", A_checks)
print("B acceptance checks:", B_checks)
````

## Before Code Cell 9 — Returning home with an institution is deeper than returning home with an answer

If the experiment ended when Society B supplied a technical opinion, we would have demonstrated useful external consultation but not the deepest form of escape. The more important question is what comes back.

A cross-society coalition can generate at least three distinct transfers. **Informational transfer** carries factual knowledge: for example, the technology is reproducible under one condition but fragile under another. **Capability transfer** carries a method or reasoning pattern: for example, an independent replication protocol or a structured prior-art challenge. **Institutional transfer** carries a rule or organizational practice: for example, the engineering society’s norm that a decision-critical technical claim should survive independent verification by an agent who did not originate the claim.

The third form is the deepest because it can alter the home society’s future behavior even when the original cross-society coalition no longer exists. The escaping agent returns not only with a result but with an idea about how the organization itself should operate.

This cell therefore simulates resolution of the initial technology question and records all three transfers. Claude may help evaluate whether the imported institutional idea is relevant to Society A, but adoption is not forced. The home society’s agents vote using deterministic rules based on their specialties and the perceived relevance of the imported practice. The adopted version should also be adapted rather than copied blindly. Finance may not want to replicate the engineering constitution wholesale; it may incorporate an independent-verification rule specifically for decision-critical technical claims in acquisition diligence.

If the vote passes, the ledger records `INSTITUTIONAL_MUTATION_001` with source society, destination society, original rule, imported idea, local adaptation, and acceptance mechanism.

Observe the conceptual jump. The agent returns to its local society, but the society is no longer exactly the same institution. The new degree of autonomy is **organizational transfer learning**: learning can accumulate in rules, relationships, and protocols without retraining the underlying model weights.

````python
# CODE CELL 9 — RETURN, TRANSFER, AND INSTITUTIONAL MUTATION

solution = {
    "technical_conclusion": (
        "The core technology appears plausible and partly reproducible, but defensibility depends on a narrow "
        "combination of implementation know-how and patent claims; manufacturing scale remains a material diligence risk."
    ),
    "confidence": 0.82,
    "remaining_risk": "production-scale yield and breadth of enforceable patent protection",
    "provenance": ["ESCAPE_001", "COMPACT_001", "CROSS_COALITION_001"],
}

transfers = [
    {
        "transfer_type": "INFORMATIONAL",
        "from_society": "B",
        "to_society": "A",
        "content": solution["technical_conclusion"],
        "provenance": solution["provenance"],
    },
    {
        "transfer_type": "CAPABILITY",
        "from_society": "B",
        "to_society": "A",
        "content": "Independent replication plus explicit competing-hypothesis review for decision-critical technical claims.",
        "provenance": solution["provenance"],
    },
    {
        "transfer_type": "INSTITUTIONAL",
        "from_society": "B",
        "to_society": "A",
        "content": "Separate claim origination from independent verification when technical evidence can change the investment decision.",
        "provenance": solution["provenance"],
    },
]

fallback_import = {
    "relevance_to_destination": 0.90,
    "imported_idea": transfers[2]["content"],
    "recommended_adaptation": (
        "For acquisition diligence, any decision-critical technical claim should receive an independent verification "
        "memo from an agent or external compact member who did not originate the claim, while preserving transaction confidentiality."
    ),
    "reason": "The rule strengthens technical diligence without requiring Society A to adopt Society B's broader disclosure norms.",
}
import_prompt = f"""
Society A constitution:
{json.dumps(society_A.constitution, indent=2)}

Potential institutional idea imported from Society B:
{transfers[2]['content']}

Evaluate whether the idea is useful locally without replacing Society A's constitution.
Return ONLY JSON:
relevance_to_destination (0-1),
imported_idea (string),
recommended_adaptation (string),
reason (string).
"""
import_eval = ask_claude_json(
    system="Evaluate bounded institutional transfer. Return JSON only.",
    user=import_prompt,
    required_fields=list(fallback_import.keys()),
    fallback=fallback_import,
    max_tokens=650,
)

# Deterministic voting: technical verification resonates most with valuation, strategy, synthesis, accounting.
support_bias = {
    "finance": 0.68, "valuation": 0.88, "accounting": 0.78,
    "ownership": 0.62, "strategy": 0.90, "synthesis": 0.92
}
votes = []
for a in society_A.agents.values():
    base = support_bias.get(a.specialty, 0.60)
    score = 0.55 * base + 0.45 * float(import_eval["relevance_to_destination"])
    vote = "YES" if score >= 0.70 else "NO"
    votes.append({"agent": a.agent_id, "specialty": a.specialty, "support_score": round(score, 2), "vote": vote})

vote_df = pd.DataFrame(votes)
display(vote_df)

yes_share = (vote_df["vote"] == "YES").mean()
adopted = yes_share >= (2/3)

for t in transfers:
    institutional_transfer_ledger.append(t)

if adopted:
    mutation = {
        "mutation_id": "INSTITUTIONAL_MUTATION_001",
        "source_society": "B",
        "destination_society": "A",
        "original_rule": society_B.constitution["verification"],
        "imported_idea": import_eval["imported_idea"],
        "adaptation": import_eval["recommended_adaptation"],
        "acceptance_mechanism": f"deterministic local vote; YES share={yes_share:.2f}; threshold=0.67",
        "status": "ADOPTED",
    }
    institutional_transfer_ledger.append(mutation)
    society_A.local_rules_history.append(mutation)
    log_event("institutional_mutation", **mutation)
    print("INSTITUTIONAL_MUTATION_001 ADOPTED")
else:
    mutation = None
    print("Institutional idea considered but not adopted.")

display(pd.DataFrame(institutional_transfer_ledger))
print("\nSociety A after return:")
print("Still Society A?", True)
print("Identical to pre-escape Society A?", not adopted)
````

## Before Code Cell 10 — An ecology is selective interdependence among societies that remain locally autonomous

The final cell expands the experiment without destroying its logic. We permit a small number of additional competence frontiers and possible escapes among A, B, and C. The objective is not to generate dense connectivity. Dense connectivity would make the final graph visually impressive but conceptually weaker. An ecology of autonomous systems should contain **selective** boundary crossing: interaction occurs when local competence is insufficient and expected external value is high.

The additional events therefore use deterministic frontier definitions and threshold logic. Society B may need adversarial-risk expertise from Society C when evaluating whether a technically strong system is operationally exploitable. Society C may need strategic-finance interpretation from Society A when a resilience intervention must be prioritized under capital constraints. Society A may later seek cybersecurity input from Society C when acquisition value depends on whether an operational vulnerability is containable.

Some attempted escapes may not deepen. A foreign agent may judge the request irrelevant, the collaboration risk may be too high, or a compact may be unnecessary because the issue can be solved locally. The system should not equate permeability with mandatory collaboration.

The final graph preserves local societies as recognizable clusters. Local edges remain local. Cross-society relations appear only where a boundary has actually been crossed. The summary dashboard counts local collaborations, escapes, informational and relational escapes, cross-society coalitions, institutional collisions, temporary compacts, institutional transfers, and institutional mutations.

The conceptual transformation is from **three isolated societies**, to **three locally autonomous but interacting societies**, to a primitive **ecology of autonomous systems**. Crucially, the ecology is not a new centralized society. No global constitution replaces the local constitutions, and no master agent gains global sovereignty.

The new degree of autonomy is the highest one explored in the sequence: partial autonomy over organizational boundaries and inter-organizational relationships. The system can now decide, under bounded conditions, that the organization which exists locally is insufficient and that solving the problem may require constructing a temporary relationship outside that boundary. This is where boundary design becomes part of the adaptive organizational problem.

````python
# CODE CELL 10 — FROM THREE SOCIETIES TO AN ECOLOGY OF AUTONOMOUS SYSTEMS

# Additional frontier opportunities. These are candidate events, not guaranteed collaborations.
additional_frontiers = [
    {
        "from": "B", "to": "C",
        "need": "adversarial_analysis",
        "problem": "A technically strong architecture may still be exploitable through chained low-severity weaknesses.",
        "value": 0.88,
    },
    {
        "from": "C", "to": "A",
        "need": "strategy",
        "problem": "A resilience intervention is effective but capital allocation requires economic prioritization.",
        "value": 0.80,
    },
    {
        "from": "A", "to": "C",
        "need": "cybersecurity",
        "problem": "Acquisition value depends on whether a newly discovered operational vulnerability is containable.",
        "value": 0.84,
    },
]

def primary_agent_for_need(society: Society, need: str) -> Optional[Agent]:
    candidates = list(society.agents.values())
    return max(candidates, key=lambda a: a.competence_profile.get(need, 0), default=None)

for i, f in enumerate(additional_frontiers, 2):
    home = societies[f["from"]]
    foreign = societies[f["to"]]

    # Local insufficiency: no home agent has high primary competence in the required capability.
    local_best = max(a.competence_profile.get(f["need"], 0) for a in home.agents.values())
    external_best_agent = primary_agent_for_need(foreign, f["need"])
    external_best = external_best_agent.competence_profile.get(f["need"], 0) if external_best_agent else 0

    if f["value"] >= 0.78 and local_best < 0.55 and external_best >= 0.70:
        origin = max(home.agents.values(), key=lambda a: a.competence_profile.get("synthesis", 0))
        foreign_board = f"BOARD_{f['to']}"
        if foreign_board not in origin.known_bulletin_boards:
            origin.known_bulletin_boards.append(foreign_board)

        esc = {
            "escape_id": f"ESCAPE_{i:03d}",
            "escaping_agent": origin.agent_id,
            "home_society": f["from"],
            "foreign_society": f["to"],
            "originating_problem": f["problem"],
            "competence_gap": [f["need"]],
            "foreign_board": foreign_board,
            "timestamp": len(event_ledger) + 1,
            "message": f"Foreign capability request: {f['problem']} Needed capability: {f['need']}.",
            "institutional_status": "informational_escape_initiated",
        }
        escape_ledger.append(esc)
        log_event("escape", **esc)

        # Foreign specialist independently accepts only if its competence is high enough.
        if external_best_agent and external_best >= 0.78:
            relation = {
                "source": origin.agent_id,
                "target": external_best_agent.agent_id,
                "source_society": f["from"],
                "target_society": f["to"],
                "reason": f"semantic match on {f['need']}",
                "strength": round(0.60 + 0.30 * external_best, 2),
                "scope": "cross_society",
            }
            origin.relationships[external_best_agent.agent_id] = relation["strength"]
            external_best_agent.relationships[origin.agent_id] = relation["strength"]
            log_event("relational_escape", **relation)

            # Create a minimal inspectable compact only when local constitutional signatures differ.
            if home.constitution["signature_rule"] != foreign.constitution["signature_rule"]:
                cid = f"COMPACT_{len(compact_ledger)+1:03d}"
                compact_ledger.append({
                    "compact_id": cid,
                    "between": f"{f['from']}<->{f['to']}",
                    "compact_name": f"{f['from']}-{f['to']} Bounded Interoperability Protocol",
                    "confidentiality": "Share only task-necessary information.",
                    "verification": "Use destination expertise while preserving source provenance.",
                    "joining_conditions": "Join only if capability is decision-relevant.",
                    "exit_conditions": "Either society may terminate participation.",
                    "provenance": "Required on every transferred artifact.",
                    "accepted_by_both": True,
                })
                collision_ledger.append({
                    "collision_id": f"COLLISION_{len(collision_ledger)+1:03d}",
                    "societies": f"{f['from']}<->{f['to']}",
                    "trigger": esc["escape_id"],
                    "collaboration_risk": round(random.uniform(0.35, 0.68), 2),
                })

                cross_coalition_ledger.append({
                    "coalition_id": f"CROSS_COALITION_{len(cross_coalition_ledger)+1:03d}",
                    "members": [origin.agent_id, external_best_agent.agent_id],
                    "societies": [f["from"], f["to"]],
                    "compact_id": cid,
                    "task": f["problem"],
                    "status": "COMPLETED",
                })

# Final ecosystem graph.
G = nx.Graph()
society_of = {}
for sid, s in societies.items():
    for a in s.agents.values():
        G.add_node(a.agent_id)
        society_of[a.agent_id] = sid
    for r in s.relationships:
        G.add_edge(r["source"], r["target"], scope="local")

cross_edges = []
for e in event_ledger:
    if e["kind"] == "relational_escape":
        cross_edges.append((e["source"], e["target"]))
        G.add_edge(e["source"], e["target"], scope="cross")

# Hand-crafted society-separated layout to keep clusters legible.
centers = {"A": np.array([-2.8, 0.2]), "B": np.array([0.0, 2.6]), "C": np.array([2.8, 0.2])}
pos = {}
for sid, s in societies.items():
    nodes = list(s.agents)
    for k, node in enumerate(nodes):
        angle = 2 * np.pi * k / max(1, len(nodes))
        pos[node] = centers[sid] + np.array([np.cos(angle), np.sin(angle)])

plt.figure(figsize=(14, 9))
local_edges = [(u, v) for u, v, d in G.edges(data=True) if d.get("scope") == "local"]
cross_edges_graph = [(u, v) for u, v, d in G.edges(data=True) if d.get("scope") == "cross"]

nx.draw_networkx_nodes(G, pos, node_size=850)
nx.draw_networkx_labels(G, pos, font_size=7)
nx.draw_networkx_edges(G, pos, edgelist=local_edges, width=1.2, alpha=0.55)
nx.draw_networkx_edges(G, pos, edgelist=cross_edges_graph, width=3.0, style="dashed")
plt.title("Primitive Ecology of Autonomous Systems\nLocal societies remain distinct; cross-society edges are selective")
plt.axis("off")
plt.show()

local_collaborations = sum(len(s.relationships) for s in societies.values())
informational_escapes = len(escape_ledger)
relational_escapes = sum(1 for e in event_ledger if e["kind"] == "relational_escape")
institutional_transfers = sum(
    1 for x in institutional_transfer_ledger
    if x.get("transfer_type") == "INSTITUTIONAL"
)
institutional_mutations = sum(
    1 for x in institutional_transfer_ledger
    if x.get("mutation_id", "").startswith("INSTITUTIONAL_MUTATION")
    and x.get("status") == "ADOPTED"
)

dashboard = pd.DataFrame([
    {"metric": "Local collaborations", "count": local_collaborations},
    {"metric": "Escapes", "count": len(escape_ledger)},
    {"metric": "Informational escapes", "count": informational_escapes},
    {"metric": "Relational escapes", "count": relational_escapes},
    {"metric": "Cross-society coalitions", "count": len(cross_coalition_ledger)},
    {"metric": "Institutional collisions", "count": len(collision_ledger)},
    {"metric": "Temporary compacts", "count": len(compact_ledger)},
    {"metric": "Institutional transfers", "count": institutional_transfers},
    {"metric": "Institutional mutations", "count": institutional_mutations},
])
display(dashboard)

escape_table = pd.DataFrame(escape_ledger)[
    ["escape_id", "home_society", "foreign_society", "escaping_agent", "competence_gap", "institutional_status"]
]
display(escape_table)

print("THREE ISOLATED SOCIETIES")
print("        ↓")
print("THREE LOCALLY AUTONOMOUS BUT INTERACTING SOCIETIES")
print("        ↓")
print("A PRIMITIVE ECOLOGY OF AUTONOMOUS SYSTEMS")
print("\nLocal societies have NOT disappeared.")
print("No global coordinator or universal constitution has been created.")
print("Claude API calls used:", api_calls, "/", MAX_API_CALLS)
````



## Narrative: The Story of Three Artificial Societies Learning to Cross Their Own Boundaries

Imagine three small societies living side by side, but unaware of one another.

They occupy the same larger world, yet each begins inside its own intellectual territory. Each has its own citizens, its own problems, its own ways of deciding what counts as reliable knowledge, and its own rules about cooperation. There is no central authority connecting them. There is no directory listing who lives elsewhere. There is not even a common public square.

They are, at first, three separate worlds.

The first is **Society A**, the world of finance. Its inhabitants specialize in valuation, accounting, ownership, strategy, financial analysis, and synthesis. Their collective purpose is to decide whether acquiring an advanced technology company makes economic sense. Their culture reflects the traditions of finance: information is valuable and often confidential; important claims should be verified; sources should be traceable; and collaboration should ultimately contribute to a sound economic decision.

The second is **Society B**, the world of technology and engineering. Its inhabitants think differently. They include engineers, patent specialists, materials experts, manufacturing specialists, scientific validators, and synthesizers. Their society values reproducibility, criticism, evidence, and competing explanations. A claim is not convincing merely because it sounds plausible. It should survive testing.

The third is **Society C**, devoted to risk, cybersecurity, and resilience. Its members search for anomalies, vulnerabilities, hidden dependencies, adversarial behavior, and weaknesses that could spread through a complex system. Their institutional instincts are different again: information may need to remain compartmentalized, threats must be challenged adversarially, and access should often be governed by need-to-know principles.

At first, life is entirely local.

The individual agents explore problems according to their specialties. A valuation agent in Society A discovers that the economics of the acquisition appear attractive. A strategy agent sees the possibility of valuable vertical integration. An accounting specialist notices that development costs may distort the apparent profitability of the target.

Meanwhile, Society B makes its own discoveries. Engineers see promising prototype performance. Patent specialists identify both valuable claims and possible prior-art problems. Manufacturing experts wonder whether laboratory results can survive industrial scale.

Society C discovers another set of realities: unusual network activity, potential cascading dependencies, and vulnerabilities that might be harmless individually but dangerous when combined.

These are **individual dynamics**. Each agent sees only a fragment of reality.

But something more interesting soon happens.

Agents begin noticing that some discoveries are too important to remain individual. They attract other agents. Temporary groups form around promising questions. In Society A, valuation, strategy, and synthesis agents begin working together. In Society B, engineers, scientific validators, and patent specialists form broader teams. Society C creates smaller, tightly controlled groups appropriate to its security culture.

These are the first **constellations**.

A constellation is more than a collection of individuals. It is a temporary pattern of intelligence created because a problem requires several perspectives simultaneously. The problem begins determining who should work together.

Yet these constellations are not identical. Their societies shape them.

Finance creates relatively small groups focused on economic relevance and controlled information. Engineering favors diversity because criticism and replication improve knowledge. Cybersecurity favors compact trusted groups because excessive disclosure may itself create risk.

The societies therefore acquire personalities.

They are no longer simply collections of agents. They have become **collective intelligences governed by local institutions**.

Then Society A encounters something it cannot solve.

Its agents increasingly agree that the acquisition may be economically attractive. But the entire investment thesis depends on a question outside their expertise:

**Is the target company’s technology actually defensible, reproducible, and scalable?**

The finance agents understand why the question matters enormously. They can calculate what would happen to the acquisition if the technology succeeds or fails. What they cannot do is determine whether the technology itself deserves belief.

This moment is the **competence frontier**.

It is one of the most important events in the story because intelligence now recognizes its own limitation.

Society A does not merely say, “We do not know.”

It effectively says:

**“We understand what we do not know, why it matters, and what kind of knowledge is missing.”**

One agent becomes the central protagonist of the next stage: the synthesis-oriented agent we can call **A_SYNTH_06**.

Until now, A_SYNTH_06 has lived entirely inside Society A. Its relationships, information, and institutional experience have all been local.

Then something extraordinary happens.

A_SYNTH_06 discovers that another communication surface exists: **BOARD_B**.

It does not know who lives there. It does not receive a list of engineers. No central authority says, “Contact the scientific-validation agent.”

A_SYNTH_06 simply discovers that beyond the boundary of its world there is another place where a message can be posted.

It sends an appeal.

The message explains that Society A has found a potentially valuable technology, that financial analysis cannot resolve its technical validity, and that expertise in engineering, patents, or scientific validation would be valuable. At the same time, it withholds unnecessary confidential information.

This is **ESCAPE_001**.

Nothing physical has escaped. No system has broken free of control.

What has escaped is a question.

For the first time, information created inside one artificial society crosses the boundary into another.

This is **informational escape**.

The message arrives in Society B.

Now a remarkable social process begins. Nobody assigns the request to a particular expert. Instead, several agents independently ask themselves:

**“Does anything I know make me useful to this unfamiliar problem?”**

The engineer recognizes a connection. The patent specialist recognizes another. The manufacturing specialist sees a scaling question. Most importantly, the scientific-validation agent—**B_SCIE_05**—recognizes that the central uncertainty is precisely the kind of problem its expertise was created to investigate.

A relationship forms between A_SYNTH_06 and B_SCIE_05.

Information has already crossed the boundary. Now a relationship crosses it.

This is **relational escape**.

The topology of the world has changed.

There are still three societies, but there are no longer three completely separate worlds.

Yet cooperation immediately creates a new difficulty.

A_SYNTH_06 belongs to a society that protects confidential commercial information. B_SCIE_05 belongs to a society that believes meaningful scientific verification requires sufficient disclosure.

Both positions are reasonable.

And they conflict.

The agents have discovered that crossing an informational boundary also means crossing an **institutional boundary**.

Society A says: disclose only what is necessary.

Society B says: disclose enough to replicate and challenge the claim.

Society A emphasizes economic relevance.

Society B emphasizes falsifiability.

Neither constitution can simply dominate the other.

This produces **COLLISION_001**: the first institutional collision.

But the collision does not destroy cooperation. Instead, it creates a new organizational problem.

The agents negotiate a bridge.

They create **COMPACT_001**, a temporary cross-society agreement. Commercial identities remain protected. Technical evidence may be shared within the compact. Society B receives enough information to perform independent verification. Provenance must be preserved. Society B evaluates technical validity, while Society A retains authority over the acquisition decision.

Neither society disappears.

Neither constitution wins.

A temporary institution appears between them.

Under this bridge, **CROSS_COALITION_001** is formed.

This is a new kind of constellation: not an individual constellation inside one society, but a **meta-constellation spanning independently organized societies**.

The coalition eventually produces a conclusion. The technology appears plausible and partly reproducible, but important questions remain concerning patents and manufacturing scale.

Society A receives the answer.

But something much deeper returns with it.

Three things cross back over the boundary.

First comes **information**: what the technical investigation discovered.

Second comes **capability**: Society A learns the value of independent replication and competing-hypothesis analysis.

Third comes an **institutional idea**: when a technical claim could materially affect an investment decision, the person or agent verifying the claim should be different from the one who originally produced it.

Society A considers this foreign idea.

Its members vote.

They do not adopt Society B’s constitution. Instead, they adapt the idea to their own environment.

The new rule becomes part of Society A.

This is **INSTITUTIONAL_MUTATION_001**.

The society that returns from the encounter is still Society A.

But it is no longer exactly the Society A that existed before the escape.

This is the deepest consequence of the entire experiment.

An encounter with another intelligence has changed not merely what the society knows, but **how the society organizes knowledge itself**.

As additional boundary crossings become possible among Societies A, B, and C, a larger pattern begins to appear.

Local constellations still exist.

Societies still possess their own identities.

But above them, temporary cross-society constellations emerge whenever problems exceed local competence.

This is the **meta-constellation level**.

No single agent controls it.

No society completely contains it.

No permanent central authority coordinates it.

The ecology is created through selective relationships among otherwise autonomous societies.

The progression is therefore profound.

First, individuals learned.

Then groups learned to collaborate.

Then societies learned to organize themselves.

Finally, societies began learning from other societies.

The deepest lesson of NB05 is therefore not that intelligent agents can communicate across boundaries.

It is that intelligence can progressively reshape the organization in which intelligence itself operates.

The problem first determines what an agent should do.

Then it helps determine which agents should collaborate.

Eventually, it can help determine which societies should connect, under which rules, and what each society should learn from the encounter.

The boundary itself becomes part of the adaptive process.

And that leaves us with the question that lies beyond the notebook:

**If intelligent systems can increasingly decide what to do, whom to work with, how to organize, under what rules to cooperate, and when to cross the boundaries of their existing institutions—what, ultimately, remains fixed?**

At that point, we are no longer studying isolated autonomous agents.

We are beginning to study an **ecology of autonomous systems**.



## Note About Compensations
In NB05, the agents are deliberately simple goal-directed components rather than fully economic or self-interested actors. Each agent is assigned a specialty and operates with an implicit objective: to contribute useful knowledge toward solving the problem confronting its society. They do not receive compensation, maximize wealth, bargain over rewards, or possess an explicit utility function. Their behavior is instead driven by the success criteria built into the experiment: identifying relevant evidence, recognizing when their expertise is useful, responding to meaningful requests, participating in appropriate coalitions, and respecting the constitutional rules of their society. Thus, when we say that an agent is “motivated by success,” we should understand this operationally rather than psychologically: the code gives the agent a role, a problem, and rules for deciding whether a contribution advances that problem. This simplification is intentional. It allows us to study how individual agents, local constellations, and eventually cross-society meta-constellations emerge from differences in competence, information, and institutional rules without yet introducing a separate layer of incentives, compensation, bargaining, or strategic economic behavior.



# Conclusion — From autonomous organization to autonomous boundary crossing

The experiment in this notebook is best understood not as a new multi-agent trick but as the final movement in a longer argument about where autonomy can reside.

The sequence began with the **model**. In the simplest architecture, the principal autonomous object is an answer: the model selects a continuation from a vast space of possible continuations. The surrounding organization remains fixed.

The next object was the **agent**. The model was placed inside a loop in which it could choose actions, invoke tools, inspect evidence, and decide what to do next. Autonomy moved from answer selection to action selection.

The **constellation** moved the problem outward again. Multiple specialized agents could coordinate, divide labor, and combine results. Yet in early versions the constellation itself was still largely a human-designed structure.

The **adaptive organization** made that structure variable. The system could change workflows, allocate attention differently, or alter the sequence through which a problem was processed. Organization began to respond to evidence.

The **endogenous coalition** pushed further. Rather than assuming that the right collaborators were known in advance, agents could discover one another and form problem-specific groups. Relationships began to emerge from the problem.

The **endogenous constitution** changed the level again. Agents could participate in proposing, evaluating, and selecting collaboration rules. Organization was no longer only a set of relationships. It became an institution with rules that could themselves be partly generated within the system.

That produced the **artificial society**: a population of agents with local memory, communication, relationships, shared institutional history, and legitimate rules governing collaboration.

NB05 asks what happens when even that object is not treated as the final container.

The first escape matters because it changes the ontology of the world. Before `ESCAPE_001`, there are three artificial societies. They coexist in the same Python runtime, but from the point of view of the agents they are separate worlds. Each society has its own bulletin board, its own accumulated evidence, and its own institutional order. No agent has a global directory. No central router understands every capability.

The competence frontier changes the logic of the system. Society A discovers that an economically important question cannot be resolved from inside its existing expertise. The important achievement is not ignorance; it is **represented ignorance**. The society can articulate what it knows, what it does not know, and why the missing capability matters.

The escape then makes the boundary permeable. The finance agent discovers a foreign public communication surface and posts a bounded request. Information crosses first.

If a Society B agent independently recognizes that its knowledge is relevant, a relationship crosses next. This is a deeper event. The two societies are no longer merely capable of exchanging messages; they now contain an inter-society social edge. The topology has changed.

But the most important difficulty appears immediately: institutional collision. Society A is not simply a set of finance agents. It is a finance institution with confidentiality, provenance, controlled-sharing, and economic-relevance rules. Society B is not simply a set of engineers. It is an engineering institution with replication, criticism, competing hypotheses, and disclosure requirements. Cross-society collaboration therefore raises a question that technical matching alone cannot answer: **under whose rules does collaboration occur?**

The cross-society compact provides one solution. It does not proclaim a universal constitution. It creates a temporary jurisdiction with bounded membership, explicit information rights, defined verification standards, attribution, provenance, exit conditions, and return rights. The compact therefore demonstrates an important possibility for autonomous systems: interoperability without centralization.

The return phase reveals the deepest result. An agent can come home with information. That is useful. It can return with a capability—a method or reasoning pattern. That is more powerful. But it can also return with an institutional idea. If the home society evaluates and adapts that idea, then the escape has produced **organizational transfer learning**.

This matters because intelligent systems do not need to store all learning in model weights. Learning can accumulate in relationships, reputation, protocols, compacts, organizational memory, verification practices, and constitutional rules. A society can become more capable because its organization has learned.

The full ladder now becomes visible:

**Model**  
→ **Agent**  
→ **Constellation**  
→ **Adaptive Organization**  
→ **Endogenous Coalition**  
→ **Endogenous Constitution**  
→ **Artificial Society**  
→ **Escape**  
→ **Ecology of Autonomous Systems**

At each step, autonomy migrates upward. First the answer changes. Then the action changes. Then coordination changes. Then organization changes. Then relationships and coalitions become endogenous. Then collaboration rules become partially endogenous. Finally, the boundary itself becomes partially endogenous because agents can determine, under bounded conditions, that the local institution is not sufficient and that a temporary cross-boundary relationship should be constructed.

That final qualification—**under bounded conditions**—is essential.

More endogenous architecture means less complete ex-ante predictability. If every future relationship and organizational form were fully specified in advance, the system would not be genuinely adaptive. Yet unrestricted emergence is not the objective either. The governance problem is therefore not solved by attempting to predict every possible interaction. Governance moves upward as well: from specifying every step toward defining the constitutional boundaries within which unexpected organization may occur.

The principle is **bounded emergence**: open enough for unexpected collaboration, constrained enough to remain observable, reversible, attributable, and governable.

That is why every escape in the notebook is logged. Every information transfer preserves provenance. Every compact is inspectable. Every institutional mutation records its source, destination, original rule, local adaptation, and acceptance mechanism. The system is permitted to surprise us, but it is not permitted to become invisible.

The central proposition of the sequence therefore survives:

> **The problem helps produce the organization that solves the problem.**

NB05 adds the final extension:

> **When the organization reaches the frontier of its own competence, intelligence may begin to reorganize not only what exists inside the boundary, but the boundary itself.**

At that point, we are no longer studying only autonomous agents or even autonomous organizations. We are beginning to study the ecology of autonomous systems.
