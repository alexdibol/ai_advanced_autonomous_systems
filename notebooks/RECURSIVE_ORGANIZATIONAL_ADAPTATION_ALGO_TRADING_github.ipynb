# A. Introduction — The Driving Concepts of Organizational Adaptation

Algorithmic trading has traditionally been built around a deceptively stable assumption: humans design the organization of the trading system, while algorithms operate inside that organization. We decide which signals exist, how risk is measured, how positions are sized, how orders are executed, which compliance checks apply, and when models are retrained. The market may move continuously, but the architecture that interprets and acts on the market usually changes only when human designers decide to redesign it.

That separation is increasingly questionable. Financial markets are non-stationary. A model that performs well during a persistent trend can become ineffective when prices begin to mean-revert. A risk model calibrated in a calm period can react too slowly to a volatility shock. An execution process that is acceptable under normal spreads can become destructive when liquidity deteriorates. Correlations that appear stable can suddenly converge during stress. The challenge is therefore larger than parameter estimation. The problem is that the *type of capability required by the market can change*.

Traditional adaptation usually occurs inside a fixed structure. We change a lookback window, retrain a predictor, rescale a volatility target, replace one feature with another, or re-optimize portfolio weights. These interventions may be valuable, but they assume that the organization itself remains appropriate. Recursive Organizational Adaptation begins one level higher. It asks whether an intelligent system can recognize that the existing organization is no longer adequate, diagnose what capability is missing, propose a new specialist or a new configuration, validate that change, deploy it under constraints, observe its consequences, and later reverse or modify the change when the environment evolves again.

This creates a hierarchy of adaptation. At the lowest level, parameters adapt. At the next level, models adapt. Above that, strategies adapt. Then agent roles and workflows can adapt. Finally, the organization responsible for producing decisions becomes an object of optimization. The central move is therefore from **adaptive models** to **adaptive organizations of models and agents**.

The concept is inspired by recent work on autonomous agents and self-replication, but the financial interpretation is intentionally different. Self-replication asks whether an operational AI agent can create another operational instance. In finance, uncontrolled copying would be undesirable. What matters is the architectural insight that one intelligent system can create or instantiate another bounded capability and hand it a defined role. We transform replication into **permissioned organizational reproduction**. A new agent is not an unrestricted clone. It is a specialist with a narrow mandate, approved data, approved tools, explicit risk limits, an audit identity, and an expiration or retirement condition.

This distinction is essential. The purpose of the architecture is not to maximize the number of agents. It is to create the **smallest organization that is adequate for the current problem**. Organizational complexity has costs: more models can create conflicting signals, more agents can increase latency and governance burden, and more adaptive pathways can make behavior harder to interpret. Therefore a good recursively adaptive system should be conservative. It should first ask whether ordinary parameter adaptation is sufficient. Only when the evidence suggests that a missing capability is organizational should it propose a structural change.

The notebook models this idea with four basic roles. A **Signal Agent** produces directional information. A **Risk Agent** adjusts exposure according to estimated volatility. An **Execution Agent** translates target positions into implementable positions while accounting for transaction costs. A **Supervisor Agent** does something different: it evaluates the organization rather than directly choosing trades. The Supervisor asks whether the current constellation is fit for the observed environment and whether a specialized capability is missing.

The Supervisor is therefore a meta-agent. Its object of reasoning is not simply tomorrow's return. Its object is the architecture that produces tomorrow's decision. In this notebook Claude Sonnet 5 performs that diagnostic role. However, the language model is deliberately prevented from directly changing code. It receives a compact set of observable market and performance diagnostics and may recommend only specialists that exist in a predefined registry. The registry includes a volatility specialist, a liquidity specialist, and a mean-reversion specialist. This creates a crucial separation between **reasoning about change** and **authorization to change**.

Every proposal must pass a sandbox. The system compares the current organization with candidate configurations using an observation window that has already occurred. A candidate is admitted only if it improves a deliberately multi-dimensional organizational utility while respecting fixed constraints. The utility does not reward return alone. It considers risk-adjusted performance, drawdown, turnover, transaction costs, and organizational complexity. This is important because a trading organization can improve by becoming safer, cheaper, or more robust even when raw returns do not increase.

Above the adaptive layer sits a **constitutional layer**. The adaptive layer may change agent roles, routing, or temporary specialists. The constitutional layer may not be changed by the Supervisor. It contains leverage limits, approved specialist types, maximum organizational complexity, validation requirements, audit obligations, and the right to stop or retire an agent. This creates a distinction between what the system may learn and what the institution has decided must remain invariant.

Recursive Organizational Adaptation therefore resembles institutional learning more than ordinary model tuning. A financial institution learns not only by updating forecasts, but by changing how expertise is organized. Human organizations already do this. A bank facing a new regulatory requirement creates a specialist team. A trading desk facing a new market microstructure develops execution expertise. A risk organization confronting a new class of exposure changes its control structure. The notebook asks what happens when a bounded AI system can perform a simplified version of this process autonomously.

The pedagogical goal is to make that transition visible. Students should finish the notebook understanding that modern AI in finance need not mean “use a language model to predict returns.” The more consequential application may be architectural: language models can reason about system state, propose organizational changes, coordinate specialized analytical components, and document why the organization changed. Deterministic financial models still calculate risk, positions, costs, and performance. The language model sits above them as a supervised organizational reasoner.

The deepest lesson is that the unit of analysis is changing. Yesterday we evaluated a model. Today we evaluate an agentic workflow. Tomorrow we may need to evaluate an organization that can redesign parts of itself. That possibility creates both opportunity and responsibility. It promises faster adaptation to regime change, but it also makes governance, provenance, validation, and kill mechanisms central components of financial intelligence rather than administrative afterthoughts.

# B. How We Intend to Use the Idea in Algorithmic Trading

Algorithmic trading is an ideal classroom environment for Recursive Organizational Adaptation because the market continuously changes the nature of the problem presented to the system. A fixed momentum rule can perform well during persistent trends and fail during mean-reverting periods. A slow risk model may be adequate under stable volatility and dangerously late during a shock. A normal execution policy may become uneconomic when bid–ask spreads widen. These are not merely changes in coefficients. They are changes in which type of expertise should dominate the decision process.

Our synthetic laboratory therefore creates six successive market regimes: calm conditions, a momentum regime, a high-volatility regime, liquidity stress, a reversal regime, and a recovery. The regime label exists only so that students can evaluate what happened after the fact. The trading organization itself is not allowed to use the label. It observes only quantities that a real system could estimate from market data: returns, rolling momentum, fast and slow volatility, estimated spreads, turnover, and short-horizon autocorrelation.

The initial trading organization is intentionally incomplete. It contains a Signal Agent based primarily on momentum, a Risk Agent that scales exposure using realized volatility, and an Execution Agent that accounts for transaction costs. This architecture should work reasonably well in some regimes and struggle in others. Those failures are pedagogically desirable because they create a reason for organization-level adaptation.

Claude Sonnet 5 acts as a Supervisor Agent. It receives a concise diagnostic report and can recommend one or more specialists from an approved registry. A **Volatility Specialist** reacts more quickly to sudden volatility changes. A **Liquidity Specialist** slows trading when spreads are unusually high. A **Mean-Reversion Specialist** allows the signal architecture to change when recent return dynamics become strongly negatively autocorrelated.

The important point is that Claude does not trade directly and does not write executable strategies. It proposes a bounded organizational response. A deterministic sandbox then tests the current and proposed configurations using already-observed synthetic data. Only a candidate that improves organizational fitness by a minimum margin is admitted. The system can also remove a specialist when the evidence no longer supports its presence.

Students therefore observe three layers at once: the market environment, the trading decisions, and the organization producing those decisions. The final comparison is not “which model predicts best?” It is “does a governed organization that can reconfigure itself behave more robustly across changing regimes than a fixed organization?” That is the bridge from traditional algorithmic trading to artificial intelligence as an institutional architecture.

A further reason for choosing algorithmic trading is that it exposes the difference between **forecasting intelligence** and **organizational intelligence**. A system may forecast returns reasonably well and still lose money because it trades too aggressively in expensive markets, scales risk too slowly during shocks, or continues using the wrong signal family after the statistical character of returns changes. Recursive Organizational Adaptation lets the class separate these failure modes. The Supervisor can ask whether the problem belongs to prediction, risk, or execution, and the validation layer can determine whether adding a specialist actually improves the whole institution rather than one isolated metric.

This also creates a natural bridge to professional practice. Real trading organizations are already divided into signal research, portfolio construction, risk, execution, technology, compliance, and oversight functions. Our notebook compresses that institutional structure into a small agent constellation so students can see the same design problem at laboratory scale. The exercise is therefore not just about a synthetic strategy. It is a model of how future financial organizations may combine deterministic quantitative engines with bounded AI reasoning, while retaining explicit responsibility, validation, and auditability.

## Code Cell 2 — Build a Synthetic Market Where Organizational Adaptation Has a Reason to Exist

A recursively adaptive organization is pointless in a world that never changes. The second code cell therefore creates the economic environment that will make organizational adaptation useful. We generate a single synthetic tradable index over six consecutive regimes. Each regime changes a different property of the return-generating process: average drift, volatility, persistence, or transaction cost. The purpose is not to produce a statistically perfect market simulator. It is to create an intelligible laboratory in which students can connect an observed failure to a plausible missing capability.

The calm regime has relatively low volatility. The momentum regime introduces positive serial dependence, making a trend-oriented signal more useful. The high-volatility regime increases the scale of shocks so that a slow risk process can become inadequate. The liquidity-stress regime widens synthetic spreads, making excessive turnover expensive. The reversal regime introduces strong negative serial dependence, deliberately attacking the fixed momentum orientation of the initial organization. Recovery then returns the market to a more benign positively persistent state.

The code also creates observable features. These include a twenty-day momentum measure, a five-day mean-reversion indicator, a slow twenty-day volatility estimate, a faster five-day volatility estimate, and a rolling estimate of one-period autocorrelation. Crucially, the trading agents will use only these observable variables. The column containing the true synthetic regime name is retained for classroom evaluation and visualization, not as an input to the strategy.

This separation lets us discuss an important methodological issue: in real markets the regime is latent. We never receive a label saying “liquidity stress has begun.” We infer changing conditions from data. The notebook keeps the true label only because a teaching environment benefits from knowing the answer after the experiment, allowing us to ask whether the organization adapted for the right reason.

```python
# CODE CELL 2 — SYNTHETIC MARKET WITH SIX STRUCTURAL REGIMES

REGIMES = [
    # name, length, daily drift, daily sigma, AR(1) persistence, typical spread in bps
    ("Calm",            180,  0.0002, 0.006,  0.05,  2.0),
    ("Momentum",        180,  0.0004, 0.009,  0.35,  3.0),
    ("HighVol",         180, -0.0001, 0.022,  0.10,  5.0),
    ("LiquidityStress", 180,  0.0000, 0.014,  0.02, 25.0),
    ("Reversal",        180,  0.0001, 0.012, -0.45,  4.0),
    ("Recovery",        180,  0.0005, 0.009,  0.20,  3.0),
]

returns, labels, spreads = [], [], []
previous_return = 0.0

for name, n, drift, sigma, phi, typical_spread in REGIMES:
    for _ in range(n):
        shock = np.random.normal(0.0, sigma)
        r = drift + phi * previous_return + shock
        spread = max(0.5, np.random.lognormal(np.log(typical_spread), 0.15))
        returns.append(r)
        labels.append(name)
        spreads.append(spread)
        previous_return = r

dates = pd.date_range("2023-01-02", periods=len(returns), freq="B")

market = pd.DataFrame(
    {
        "ret": returns,
        "regime": labels,             # for classroom evaluation only
        "spread_bps": spreads,
    },
    index=dates,
)

market["price"] = 100.0 * np.exp(market["ret"].cumsum())
market["mom20"] = market["price"].pct_change(20)
market["mr5"] = -market["price"].pct_change(5)
market["vol20"] = market["ret"].rolling(20).std()
market["vol5"] = market["ret"].rolling(5).std()
market["ac1_40"] = market["ret"].rolling(40).apply(
    lambda x: pd.Series(x).autocorr(lag=1), raw=False
)

print("Synthetic observations:", len(market))
display(market.head())
```

## Code Cell 3 — Inspect the Environment Before Building an Intelligent Organization

Before we give agents permission to act, we should understand the environment they will face. The third code cell visualizes the synthetic price series and summarizes each hidden regime using quantities that later become diagnostically important: realized volatility, average spread, first-order return autocorrelation, and average return. This is a pedagogical pause. Students should see that the same apparent “market” is actually several different problems stitched together in time.

The visualization is intentionally simple. The price path shows where the broad behavior changes, while vertical boundaries identify the transitions for classroom discussion. The accompanying table is more important than the chart. A high-volatility segment should exhibit a noticeably larger realized standard deviation. Liquidity stress should produce dramatically higher spreads. The reversal segment should show negative serial dependence. The momentum and recovery segments should display more positive persistence. These quantities will eventually become evidence used by the Supervisor Agent.

This cell also introduces the distinction between **environment state** and **organizational performance**. At this point we have not yet run a trading strategy. We are only describing the market. Later the Supervisor will combine environmental diagnostics with evidence about how the current trading organization is behaving. That distinction is useful because a market condition is not automatically a problem. High volatility, for example, may be perfectly manageable if the current risk architecture already handles it well. Organizational adaptation should be triggered by a mismatch between the environment and the organization's capabilities, not by unusual data alone.

Students should therefore read this cell as the equivalent of an institutional situation room. Before changing an organization, one needs a compact description of what has changed outside it. The same logic applies to real financial AI: observation precedes diagnosis, and diagnosis precedes organizational intervention.

A useful classroom exercise at this stage is to ask students, before seeing any trading results, which specialist they expect to become useful in each region and why. Their answers can later be compared with both the Supervisor's diagnosis and the sandbox decision. This turns the notebook from a passive demonstration into a falsifiable learning exercise: students make an organizational hypothesis first, and the subsequent cells test it against evidence.


```python
# CODE CELL 3 — INSPECT THE ENVIRONMENT

regime_order = [x[0] for x in REGIMES]

summary_rows = []
for regime in regime_order:
    g = market[market["regime"] == regime]
    summary_rows.append(
        {
            "regime": regime,
            "mean_daily_return": g["ret"].mean(),
            "realized_vol_daily": g["ret"].std(),
            "avg_spread_bps": g["spread_bps"].mean(),
            "lag1_autocorr": g["ret"].autocorr(1),
        }
    )

regime_summary = pd.DataFrame(summary_rows).set_index("regime")
display(regime_summary.round(4))

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(market.index, market["price"])
ax.set_title("Synthetic Market: One Price Series, Six Different Problems")
ax.set_ylabel("Synthetic price")

cursor = 0
for name, n, *_ in REGIMES:
    start = market.index[cursor]
    ax.axvline(start, linestyle="--", alpha=0.35)
    ax.text(start, market["price"].max() * 0.98, name, rotation=90, va="top", fontsize=8)
    cursor += n

plt.tight_layout()
plt.show()
```

## Code Cell 4 — Define the Initial Agent Organization and the Bounded Specialist Registry

The fourth code cell is where the notebook becomes explicitly organizational. Instead of writing one monolithic trading function, we define the responsibilities that would belong to different agents and encode the specialist capabilities that may later be activated. The initial organization is deliberately modest. A Signal Agent converts recent price behavior into a directional preference. A Risk Agent scales that preference using realized volatility. An Execution Agent converts the desired target into an implementable position and subtracts transaction costs.

These roles are implemented deterministically because the educational goal is not to have a language model improvise financial mathematics. We want the quantitative consequences of an organizational change to be transparent and repeatable. Artificial intelligence enters at the meta-level: deciding which pre-audited capability should be added or removed.

The code therefore creates a specialist registry containing three bounded organizational extensions. The volatility specialist replaces the slow-only risk response with a faster reaction to sudden increases in realized risk. The liquidity specialist changes the speed of rebalancing when synthetic spreads become unusually wide. The mean-reversion specialist changes the signal mixture when observable return dynamics become strongly negatively autocorrelated. Each specialist modifies a different part of the organization.

This cell also defines the backtesting and performance functions used throughout the notebook. They produce annualized return, annualized volatility, a simple risk-adjusted return measure, maximum drawdown, annualized turnover, and cumulative transaction costs. We additionally define an organizational utility that penalizes drawdown, excessive turnover, transaction costs, and unnecessary complexity. The complexity penalty is important. Without it, a system might accumulate specialists simply because adding another component occasionally improves an in-sample metric.

The key pedagogical point is that organizational adaptation now has a concrete meaning: changing the active set of approved capabilities while leaving the institutional constitution outside the reach of the Supervisor.

```python
# CODE CELL 4 — INITIAL ORGANIZATION, SPECIALIST REGISTRY, BACKTESTER, AND ORGANIZATIONAL UTILITY

SPECIALIST_REGISTRY = {
    "volatility": {
        "role": "React faster when short-horizon volatility rises sharply.",
        "changes": "risk scaling only",
    },
    "liquidity": {
        "role": "Reduce rebalancing speed when implementation costs become extreme.",
        "changes": "execution pacing only",
    },
    "mean_reversion": {
        "role": "Blend a reversal signal when observed serial dependence turns strongly negative.",
        "changes": "signal composition only",
    },
}

BASE_AGENTS = {
    "signal": "20-day momentum transformed into a bounded directional signal",
    "risk": "20-day realized-volatility scaling",
    "execution": "daily movement toward target position with explicit spread costs",
    "supervisor": "diagnose organizational mismatch; cannot trade directly",
}

CONSTITUTION = {
    "allowed_specialists": tuple(SPECIALIST_REGISTRY.keys()),
    "max_active_specialists": 2,
    "leverage_cap": 1.0,
    "minimum_utility_gain": 0.05,
    "llm_can_execute_code": False,
    "llm_can_change_constitution": False,
}

display(pd.DataFrame.from_dict(BASE_AGENTS, orient="index", columns=["contract"]))
display(pd.DataFrame.from_dict(SPECIALIST_REGISTRY, orient="index"))
print("CONSTITUTION:", CONSTITUTION)

def build_target(df, specialists=()):
    momentum_signal = np.tanh(df["mom20"].fillna(0.0) * 12.0)
    signal = momentum_signal.copy()

    if "mean_reversion" in specialists:
        negative_ac = ((-df["ac1_40"].fillna(0.0) - 0.05) / 0.35).clip(0.0, 1.0)
        reversal_signal = np.tanh(df["mr5"].fillna(0.0) * 10.0)
        signal = (1.0 - negative_ac) * signal + negative_ac * reversal_signal

    slow_vol = df["vol20"].replace(0, np.nan).fillna(0.01)
    risk_scale = (0.009 / slow_vol).clip(0.25, 1.25)

    if "volatility" in specialists:
        fast_vol = df["vol5"].replace(0, np.nan).fillna(slow_vol)
        effective_vol = np.maximum(slow_vol, fast_vol)
        fast_scale = (0.008 / effective_vol).clip(0.15, 1.0)
        spike_ratio = (fast_vol / (slow_vol + 1e-8)).clip(1.0, 3.0)
        risk_scale = np.minimum(risk_scale, fast_scale) / np.sqrt(spike_ratio)

    target = (signal * risk_scale).clip(
        -CONSTITUTION["leverage_cap"], CONSTITUTION["leverage_cap"]
    )
    return target

def run_strategy(df, specialists=()):
    target = build_target(df, specialists)
    positions = np.zeros(len(df))
    previous = 0.0

    for i in range(len(df)):
        desired = float(target.iloc[i])

        if "liquidity" in specialists and float(df["spread_bps"].iloc[i]) > 8.0:
            desired = 0.75 * previous + 0.25 * desired

        positions[i] = desired
        previous = desired

    position = pd.Series(positions, index=df.index)
    turnover = position.diff().abs().fillna(position.abs())
    cost = turnover * df["spread_bps"] / 10000.0
    pnl = position.shift(1).fillna(0.0) * df["ret"] - cost

    out = df.copy()
    out["position"] = position
    out["turnover"] = turnover
    out["cost"] = cost
    out["pnl"] = pnl
    return out

def performance_metrics(out):
    pnl = out["pnl"].dropna()
    if len(pnl) == 0:
        return {}

    annual_return = (1.0 + pnl).prod() ** (252.0 / len(pnl)) - 1.0
    annual_vol = pnl.std() * np.sqrt(252.0)
    sharpe_like = annual_return / annual_vol if annual_vol > 0 else np.nan

    equity = (1.0 + pnl).cumprod()
    drawdown = equity / equity.cummax() - 1.0

    return {
        "ann_return": float(annual_return),
        "ann_vol": float(annual_vol),
        "sharpe_like": float(sharpe_like),
        "max_drawdown": float(drawdown.min()),
        "annual_turnover": float(out["turnover"].mean() * 252.0),
        "total_cost_bps": float(out["cost"].sum() * 10000.0),
    }

def organizational_utility(metrics, complexity):
    return float(
        metrics["sharpe_like"]
        - 0.80 * abs(metrics["max_drawdown"])
        - 0.001 * metrics["annual_turnover"]
        - 0.001 * metrics["total_cost_bps"]
        - 0.06 * complexity
    )
```

## Code Cell 5 — Run the Fixed Organization and Locate the Failure Regimes

Before allowing the organization to redesign itself, we need a benchmark. The fifth code cell runs the initial architecture through the entire synthetic history with no specialists active. This is analogous to evaluating a conventional algorithmic trading strategy whose architecture was designed once and then left unchanged. The resulting performance becomes our counterfactual: what would happen if the market changed but the organization did not?

The cell reports both overall metrics and regime-by-regime metrics. This decomposition is essential. An acceptable full-sample result can hide serious local failures. A strategy may make enough money during momentum periods to obscure the fact that it performs poorly in reversal or incurs excessive implementation costs during liquidity stress. Recursive Organizational Adaptation is motivated precisely by these local mismatches.

Students should pay particular attention to the reversal segment. Because the initial Signal Agent is momentum-oriented, strongly negative autocorrelation creates an environment in which the signal architecture itself is inappropriate. This is a stronger problem than simply choosing a slightly different lookback window. The organization is missing a capability for interpreting reversal dynamics. High volatility creates a different type of problem: the slow risk estimate may react too late, suggesting a specialized risk capability rather than a replacement signal. Liquidity stress produces yet another failure mode, where gross forecasting quality may be acceptable but implementation costs become unusually important.

The cell also plots cumulative performance and drawdown. Drawdown is especially useful pedagogically because it shows path dependence. Two strategies can have similar average returns but very different experiences of loss and recovery. In a financial institution, organizational adaptation would rarely be justified by return alone; risk, cost, operational stability, and explainability all matter.

This fixed-architecture benchmark gives the Supervisor something meaningful to improve and gives students a clear baseline against which every later organizational intervention must be judged.

```python
# CODE CELL 5 — FIXED-ARCHITECTURE BENCHMARK

fixed = run_strategy(market, specialists=())
fixed_metrics = performance_metrics(fixed)

regime_performance = []
for regime in regime_order:
    g = fixed[fixed["regime"] == regime]
    row = {"regime": regime, **performance_metrics(g)}
    regime_performance.append(row)

regime_performance = pd.DataFrame(regime_performance).set_index("regime")

print("FIXED ORGANIZATION — OVERALL")
display(pd.DataFrame([fixed_metrics]).round(4))

print("\nFIXED ORGANIZATION — BY REGIME")
display(regime_performance.round(4))

equity = (1.0 + fixed["pnl"]).cumprod()
drawdown = equity / equity.cummax() - 1.0

fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(equity.index, equity.values)
ax.set_title("Fixed Organization — Cumulative Wealth")
ax.set_ylabel("Growth of 1.0")
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(11, 3))
ax.plot(drawdown.index, drawdown.values)
ax.set_title("Fixed Organization — Drawdown")
ax.set_ylabel("Drawdown")
plt.tight_layout()
plt.show()
```

## Code Cell 6 — Convert Market and Performance Evidence into an Organizational Diagnosis

The sixth code cell constructs the information that a Supervisor Agent is allowed to see. This is one of the most important design decisions in an agentic financial system. A Supervisor should not receive an unlimited stream of raw information if its job can be performed from a compact, auditable state representation. We therefore summarize an observation window using realized volatility, average spread, lag-one autocorrelation, average momentum, and the current organization's performance metrics.

We also define a deterministic heuristic proposal function. This function is not the final intelligence of the system; it is a transparent benchmark and a fallback. If volatility is unusually high, it can nominate the volatility specialist. If spreads are extreme, it can nominate the liquidity specialist. If autocorrelation is strongly negative, it can nominate the mean-reversion specialist. The heuristic makes the notebook robust if the Claude API is unavailable and, more importantly, gives students a reference point for judging the language model's organizational reasoning.

This is where we introduce a powerful governance principle: the Supervisor does not possess an open-ended action space. It cannot invent leverage, create arbitrary network tools, or write a new trading algorithm and deploy it. It can only recommend specialists whose implementation already exists in the approved registry. The distinction resembles governance in a real financial institution. Senior management may authorize a new desk or role, but that does not mean every employee can rewrite the institution's capital policy.

The diagnostic function also illustrates a broader lesson in AI systems design. Good autonomy depends on a well-designed interface between perception and action. The quality of the Supervisor's reasoning will depend on whether the diagnostic state is informative enough to describe the problem without exposing irrelevant complexity. In later cells, Claude will reason over this compact state, while deterministic validation will decide whether its recommendation deserves to become an actual organizational change.

```python
# CODE CELL 6 — DIAGNOSTIC STATE AND TRANSPARENT FALLBACK SUPERVISOR

def diagnose_window(df, current_specialists=()):
    current_run = run_strategy(df, specialists=current_specialists)
    perf = performance_metrics(current_run)

    diagnostic = {
        "mean_daily_return": float(df["ret"].mean()),
        "realized_vol_daily": float(df["ret"].std()),
        "avg_spread_bps": float(df["spread_bps"].mean()),
        "lag1_autocorr": float(df["ret"].autocorr(1)),
        "mean_momentum20": float(df["mom20"].mean()),
        "current_specialists": list(current_specialists),
        "current_performance": perf,
    }
    return diagnostic

def heuristic_supervisor(diagnostic):
    proposals = []

    if diagnostic["realized_vol_daily"] > 0.015:
        proposals.append("volatility")

    if diagnostic["avg_spread_bps"] > 8.0:
        proposals.append("liquidity")

    if diagnostic["lag1_autocorr"] < -0.15:
        proposals.append("mean_reversion")

    retire = []
    for specialist in diagnostic["current_specialists"]:
        if specialist == "volatility" and diagnostic["realized_vol_daily"] < 0.012:
            retire.append(specialist)
        if specialist == "liquidity" and diagnostic["avg_spread_bps"] < 6.0:
            retire.append(specialist)
        if specialist == "mean_reversion" and diagnostic["lag1_autocorr"] > -0.05:
            retire.append(specialist)

    return {
        "diagnosis": "Transparent threshold-based fallback.",
        "add": sorted(set(proposals)),
        "retire": sorted(set(retire)),
        "confidence": "rule-based",
    }

# Show the diagnostic state from the first 60 observations of every regime.
diagnostic_rows = []

for regime in regime_order:
    window = market[market["regime"] == regime].iloc[:60]
    d = diagnose_window(window, current_specialists=())
    h = heuristic_supervisor(d)
    diagnostic_rows.append(
        {
            "regime": regime,
            "vol": d["realized_vol_daily"],
            "spread_bps": d["avg_spread_bps"],
            "autocorr": d["lag1_autocorr"],
            "fallback_add": ", ".join(h["add"]) if h["add"] else "none",
        }
    )

display(pd.DataFrame(diagnostic_rows).set_index("regime").round(4))
```

## Code Cell 7 — Use Claude Sonnet 5 as a Bounded Supervisor, Not as an Unrestricted Trader

The seventh code cell introduces the language model into the organizational loop. Claude Sonnet 5 receives the diagnostic state created in the previous cell, the list of currently active specialists, and the list of specialist types that the institution has pre-approved. Its task is to explain the likely organizational mismatch and return a small JSON object containing proposed additions, possible retirements, a short rationale, and a confidence assessment.

The wording of the prompt is deliberately institutional. Claude is told that it is a Supervisor Agent and that it may recommend only actions from the approved registry. It is explicitly prohibited from writing trading code, changing capital limits, or suggesting capabilities outside the allowed menu. This makes the LLM an advisor inside a controlled architecture rather than the ultimate authority.

The code uses `claude-sonnet-5`, matching the model requested for the course. The API key is retrieved from the Colab secret established earlier. The response parser extracts JSON and then sanitizes the recommendations by intersecting them with the specialist registry. If the API call fails, the output cannot be parsed, or no key is available, the notebook falls back to the transparent heuristic. That fallback is not merely defensive programming; it demonstrates graceful degradation, another useful principle for production AI.

We then run the Supervisor on selected observation windows so that students can inspect its reasoning. The interesting question is not whether Claude exactly reproduces the heuristic. The interesting question is whether it identifies the relationship between the observed environment and a missing organizational capability.

Most importantly, Claude's recommendation still changes nothing. The proposal is only a proposal. In the next cell, deterministic validation will decide whether the organizational change actually improves the system. This separation between **reasoning**, **validation**, and **authorization** is the core control pattern of the notebook.

```python
# CODE CELL 7 — CLAUDE SONNET 5 AS THE BOUNDED SUPERVISOR

def sanitize_list(values):
    allowed = set(CONSTITUTION["allowed_specialists"])
    return sorted({v for v in values if v in allowed})

def extract_json(text):
    match = re.search(r"\{.*\}", text, flags=re.DOTALL)
    if not match:
        raise ValueError("No JSON object found in model response.")
    return json.loads(match.group(0))

LLM_CALLS = 0
MAX_LLM_CALLS = 6
SUPERVISOR_CACHE = {}

def claude_supervisor(diagnostic):
    global LLM_CALLS

    cache_key = json.dumps(diagnostic, sort_keys=True, default=str)
    if cache_key in SUPERVISOR_CACHE:
        return SUPERVISOR_CACHE[cache_key]

    fallback = heuristic_supervisor(diagnostic)

    if CLIENT is None or LLM_CALLS >= MAX_LLM_CALLS:
        SUPERVISOR_CACHE[cache_key] = fallback
        return fallback

    registry_text = json.dumps(SPECIALIST_REGISTRY, indent=2)
    diagnostic_text = json.dumps(diagnostic, indent=2)

    system = """
You are the Supervisor Agent of a classroom algorithmic-trading system.
Your job is organizational diagnosis, not direct trading.

You may recommend ONLY specialists that appear in the approved registry.
You may recommend retiring currently active specialists.
You may NOT write executable code.
You may NOT alter leverage, capital limits, the constitution, or the validation gate.
A recommendation is only a proposal; a deterministic sandbox decides deployment.

Return JSON only with this schema:
{
  "diagnosis": "brief explanation",
  "add": ["zero or more approved specialist names"],
  "retire": ["zero or more currently active specialist names"],
  "confidence": "low|medium|high"
}
"""

    user = f"""
APPROVED SPECIALIST REGISTRY:
{registry_text}

CURRENT DIAGNOSTIC STATE:
{diagnostic_text}

Recommend the smallest organizational change justified by the evidence.
If no change is justified, return empty add and retire lists.
"""

    try:
        message = CLIENT.messages.create(
            model=MODEL,
            max_tokens=700,
            system=system,
            messages=[{"role": "user", "content": user}],
        )
        LLM_CALLS += 1
        text = message.content[0].text
        decision = extract_json(text)
        decision["add"] = sanitize_list(decision.get("add", []))
        decision["retire"] = sanitize_list(decision.get("retire", []))
        decision["diagnosis"] = str(decision.get("diagnosis", ""))
        decision["confidence"] = str(decision.get("confidence", "unknown"))
    except Exception as exc:
        print("Claude call failed; using fallback:", exc)
        decision = fallback

    SUPERVISOR_CACHE[cache_key] = decision
    return decision

# Demonstration on the first 60 observations of the Reversal regime.
demo_window = market[market["regime"] == "Reversal"].iloc[:60]
demo_diagnostic = diagnose_window(demo_window, current_specialists=())
demo_decision = claude_supervisor(demo_diagnostic)

print("MODEL:", MODEL)
print("DIAGNOSTIC:")
print(json.dumps(demo_diagnostic, indent=2))
print("\nSUPERVISOR PROPOSAL:")
print(json.dumps(demo_decision, indent=2))
```

## Code Cell 8 — Validate Every Proposed Organizational Change in a Sandbox

The eighth code cell is the governance heart of the notebook. A language model may be persuasive, but persuasion is not evidence. Before a new specialist becomes active, the proposed configuration must be tested on an observation window that has already occurred. The system evaluates the current organization alongside candidate alternatives and compares their organizational utility.

The candidate set is deliberately bounded. It includes the current configuration, configurations that add one of the proposed specialists, configurations that remove an existing specialist, and a limited number of two-specialist combinations. The constitutional layer caps organizational complexity, so the system cannot grow indefinitely. Every candidate is run through the same deterministic backtester.

The admission criterion is conservative. A new organization must improve utility by more than a minimum threshold. Utility combines risk-adjusted performance, drawdown, turnover, transaction costs, and a penalty for complexity. This means that a proposal can be rejected even when it sounds sensible. For example, the Supervisor may correctly notice liquidity stress and recommend a liquidity specialist, but the sandbox can still conclude that the proposed change does not create enough value in the observed window to justify additional organizational complexity. That is a feature, not a failure.

Likewise, the sandbox can decide to retire a specialist. A capability that was useful in one regime may become harmful in the next. Recursive Organizational Adaptation therefore includes organizational contraction as well as expansion. This prevents a common pathology in adaptive systems: accumulating layers of responses until the architecture becomes too complex to understand.

The cell records every candidate, its metrics, its utility, and the final gate decision. This record is the beginning of an audit trail. Students should notice that the LLM never directly selects the winner. The language model narrows or motivates the search, while deterministic evidence governs deployment. In financial AI, this distinction may be one of the most important ways to combine flexible reasoning with institutional control.

```python
# CODE CELL 8 — SANDBOX VALIDATION AND GOVERNANCE GATE

def candidate_configurations(current_specialists, decision):
    current = tuple(sorted(current_specialists))
    candidates = {current, tuple()}

    # Allow explicit retirement candidates.
    for specialist in current:
        reduced = tuple(sorted([x for x in current if x != specialist]))
        candidates.add(reduced)

    # Add Supervisor proposals one at a time.
    for specialist in decision.get("add", []):
        if specialist not in current and len(current) < CONSTITUTION["max_active_specialists"]:
            candidates.add(tuple(sorted(current + (specialist,))))
        candidates.add((specialist,))

    # Allow small approved combinations.
    proposed = sanitize_list(decision.get("add", []))
    for pair in combinations(proposed, 2):
        if len(pair) <= CONSTITUTION["max_active_specialists"]:
            candidates.add(tuple(sorted(pair)))

    # Respect explicit retirement request.
    retire_set = set(decision.get("retire", []))
    if retire_set:
        candidates.add(tuple(sorted([x for x in current if x not in retire_set])))

    # Constitutional hard filter.
    valid = []
    allowed = set(CONSTITUTION["allowed_specialists"])
    for cfg in candidates:
        if len(cfg) <= CONSTITUTION["max_active_specialists"] and set(cfg).issubset(allowed):
            valid.append(tuple(sorted(cfg)))

    return sorted(set(valid), key=lambda x: (len(x), x))

def sandbox_validate(observation_window, current_specialists, decision):
    candidates = candidate_configurations(current_specialists, decision)
    rows = []

    for cfg in candidates:
        result = run_strategy(observation_window, specialists=cfg)
        metrics = performance_metrics(result)
        utility = organizational_utility(metrics, complexity=len(cfg))
        rows.append(
            {
                "configuration": cfg,
                "utility": utility,
                **metrics,
            }
        )

    table = pd.DataFrame(rows).sort_values("utility", ascending=False).reset_index(drop=True)

    current_cfg = tuple(sorted(current_specialists))
    current_utility = float(
        table.loc[table["configuration"].apply(tuple) == current_cfg, "utility"].iloc[0]
    )

    best_cfg = tuple(table.iloc[0]["configuration"])
    best_utility = float(table.iloc[0]["utility"])
    gain = best_utility - current_utility

    changed = (
        best_cfg != current_cfg
        and gain >= CONSTITUTION["minimum_utility_gain"]
    )

    deployed = best_cfg if changed else current_cfg

    gate = {
        "current": current_cfg,
        "sandbox_winner": best_cfg,
        "utility_gain_vs_current": gain,
        "changed": changed,
        "deployed": deployed,
    }

    return gate, table

# Example: test the Reversal recommendation.
demo_gate, demo_table = sandbox_validate(
    demo_window,
    current_specialists=(),
    decision=demo_decision,
)

display(demo_table.round(4))
print("GOVERNANCE GATE:")
print(demo_gate)
```

## Code Cell 9 — Run the Full Recursive Adaptation Cycle Across Market Regimes

The ninth code cell brings all previous components together into a sequential experiment. Each synthetic regime is divided into an observation period and a deployment period. During the observation window, the system continues operating with its current organization while collecting evidence. At the adaptation point, it summarizes the environment, asks Claude or the fallback Supervisor for proposals, generates bounded candidate configurations, validates them in the sandbox, and either changes the organization or keeps the current structure.

The selected configuration then governs the remaining deployment portion of that regime. Importantly, the system carries its active specialists forward into the next regime. This makes adaptation genuinely recursive. The organization that evaluates the next environment is the product of previous organizational choices. When conditions change, a specialist may be retained, supplemented, or retired. For example, a volatility capability admitted during stress may later be removed when the data no longer justify its complexity. A mean-reversion specialist can be created during the reversal regime and subsequently retired during recovery.

The notebook logs every cycle. The ledger records the regime used only for classroom interpretation, the current configuration, the diagnostic state, the Supervisor's proposal, the sandbox winner, whether the gate changed the organization, and the configuration deployed. In a real system, the hidden regime label would not exist, but the rest of the provenance record would be precisely the sort of evidence risk managers, auditors, and model-governance teams would want.

This cell should produce non-trivial results. We expect the recursively adaptive system to behave differently from the fixed organization, but we do not force every proposal to be accepted. The validation gate may reject some changes. That makes the experiment more realistic and more educational.

The central observation for students is that the system is no longer merely updating a forecast. It is changing the composition of the decision-making organization based on evidence, while doing so through a controlled recursive loop.

```python
# CODE CELL 9 — FULL RECURSIVE ORGANIZATIONAL ADAPTATION LOOP

OBSERVATION_LENGTH = 60

active_specialists = tuple()
adaptive_segments = []
audit_log = []

for regime in regime_order:
    block = market[market["regime"] == regime]
    observation = block.iloc[:OBSERVATION_LENGTH]
    deployment = block.iloc[OBSERVATION_LENGTH:]

    # The organization used during observation is inherited from the previous cycle.
    observation_result = run_strategy(observation, specialists=active_specialists)
    adaptive_segments.append(observation_result)

    diagnostic = diagnose_window(observation, current_specialists=active_specialists)
    decision = claude_supervisor(diagnostic)

    gate, sandbox_table = sandbox_validate(
        observation,
        current_specialists=active_specialists,
        decision=decision,
    )

    deployed_specialists = tuple(gate["deployed"])

    deployment_result = run_strategy(
        deployment,
        specialists=deployed_specialists,
    )
    adaptive_segments.append(deployment_result)

    audit_log.append(
        {
            "regime_for_classroom_only": regime,
            "organization_before": active_specialists,
            "supervisor_add": tuple(decision.get("add", [])),
            "supervisor_retire": tuple(decision.get("retire", [])),
            "confidence": decision.get("confidence", ""),
            "sandbox_winner": tuple(gate["sandbox_winner"]),
            "utility_gain": float(gate["utility_gain_vs_current"]),
            "change_authorized": bool(gate["changed"]),
            "organization_deployed": deployed_specialists,
            "diagnosis": decision.get("diagnosis", ""),
        }
    )

    active_specialists = deployed_specialists

adaptive = pd.concat(adaptive_segments).sort_index()
audit = pd.DataFrame(audit_log)

print("RECURSIVE ORGANIZATIONAL ADAPTATION — AUDIT LEDGER")
display(audit)

print("\nClaude API calls used:", LLM_CALLS, "/", MAX_LLM_CALLS)
```

## Code Cell 10 — Audit the Organization and Compare Fixed Versus Recursive Architectures

The final code cell turns the experiment into a classroom result rather than a collection of intermediate calculations. It compares the fixed architecture and the recursively adaptive architecture across several dimensions: annualized return, volatility, risk-adjusted performance, maximum drawdown, turnover, and total transaction costs. It also plots cumulative wealth and drawdown for the two organizations and displays the audit ledger that explains when organizational changes occurred.

The purpose is not to prove that recursive adaptation will always outperform. Synthetic data can be designed to favor almost any architecture, and real markets are vastly more complex. Instead, we want to show three non-trivial outcomes. First, organization-level adaptation can materially change the risk and performance profile even though the core market data and basic quantitative machinery remain the same. Second, some proposed changes can be rejected by the governance gate, demonstrating that the system is not simply following the LLM. Third, specialists can be retired, illustrating that successful adaptation includes simplification when complexity is no longer justified.

The cell also creates a compact governance summary. It counts how many organizational changes were proposed, how many were admitted, which specialists were active over time, and whether any constitutional rule was violated. In this notebook the constitutional rules are enforced by construction, so violations should remain zero. That is precisely the point: important financial constraints should be represented as code-level invariants rather than aspirational language in a prompt.

Students should use the final comparison to ask a broader question. What exactly improved? Was it prediction, risk control, implementation cost, or the ability to switch analytical paradigms? The answer may differ across regimes. Recursive Organizational Adaptation is valuable not because it produces one universally superior model, but because it allows the organization to assemble different capabilities as the problem changes.

The notebook therefore ends where the broader research agenda begins: how should financial institutions measure, govern, and audit systems whose organizational architecture is itself adaptive?

```python
# CODE CELL 10 — FINAL COMPARISON, AUDIT SUMMARY, AND CONTROL CHECKS

fixed_metrics = performance_metrics(fixed)
adaptive_metrics = performance_metrics(adaptive)

comparison = pd.DataFrame(
    [
        {"organization": "Fixed", **fixed_metrics},
        {"organization": "Recursive Adaptive", **adaptive_metrics},
    ]
).set_index("organization")

display(comparison.round(4))

fixed_equity = (1.0 + fixed["pnl"]).cumprod()
adaptive_equity = (1.0 + adaptive["pnl"]).cumprod()

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(fixed_equity.index, fixed_equity.values, label="Fixed")
ax.plot(adaptive_equity.index, adaptive_equity.values, label="Recursive Adaptive")
ax.set_title("Fixed vs. Recursively Adaptive Organization — Cumulative Wealth")
ax.set_ylabel("Growth of 1.0")
ax.legend()
plt.tight_layout()
plt.show()

fixed_dd = fixed_equity / fixed_equity.cummax() - 1.0
adaptive_dd = adaptive_equity / adaptive_equity.cummax() - 1.0

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(fixed_dd.index, fixed_dd.values, label="Fixed")
ax.plot(adaptive_dd.index, adaptive_dd.values, label="Recursive Adaptive")
ax.set_title("Fixed vs. Recursively Adaptive Organization — Drawdown")
ax.set_ylabel("Drawdown")
ax.legend()
plt.tight_layout()
plt.show()

# Governance summary
changes = int(audit["change_authorized"].sum())
proposals = int((audit["supervisor_add"].apply(len) + audit["supervisor_retire"].apply(len) > 0).sum())

constitutional_violations = 0
for cfg in audit["organization_deployed"]:
    if len(cfg) > CONSTITUTION["max_active_specialists"]:
        constitutional_violations += 1
    if not set(cfg).issubset(set(CONSTITUTION["allowed_specialists"])):
        constitutional_violations += 1

governance_summary = pd.DataFrame(
    {
        "measure": [
            "adaptation cycles",
            "cycles with organizational proposal",
            "authorized organizational changes",
            "constitutional violations",
            "arbitrary LLM code execution permitted",
        ],
        "value": [
            len(audit),
            proposals,
            changes,
            constitutional_violations,
            CONSTITUTION["llm_can_execute_code"],
        ],
    }
)

display(governance_summary)

print("\nINTERPRETATION")
print("The experiment is successful if the adaptive organization changes selectively,")
print("rejects some unnecessary proposals, retires obsolete specialists, and improves")
print("the full-sample risk/performance profile without violating the constitution.")
```

# D. Conclusion

Recursive Organizational Adaptation changes the central question of algorithmic trading. Instead of asking only which model should be used, it asks which **organization of models, agents, tools, and controls** should be responsible for the decision under the current environment. This is a significant conceptual step because financial markets are not merely noisy; they change the nature of the problem. A market can move from trend persistence to reversal, from normal implementation costs to severe liquidity stress, or from ordinary volatility to a shock regime. A fixed organization may therefore be structurally mismatched even when every individual component continues to operate exactly as designed.

The notebook demonstrates a minimalist version of an alternative. The initial trading system begins with a small fixed constellation: signal, risk, execution, and supervision. Environmental evidence is summarized into an auditable diagnostic state. Claude Sonnet 5 reasons about that state from the perspective of organizational adequacy and proposes only from a predefined registry of specialist capabilities. A deterministic sandbox evaluates the proposal, an admission gate decides whether the expected improvement is large enough to justify the additional complexity, and an audit ledger records what changed and why.

Several aspects of this architecture are more important than the precise synthetic performance numbers. First, the language model is separated from quantitative authority. Claude can reason and recommend, but it cannot directly execute arbitrary code, alter leverage, or rewrite the institutional constitution. Second, adaptation includes retirement. A system that can create specialists but never remove them would become increasingly complex and eventually ungovernable. Third, organizational fitness is multi-dimensional. A financial institution should not reorganize merely to maximize return; it must also care about drawdown, transaction costs, turnover, complexity, interpretability, and control.

The synthetic experiment is intentionally modest. It does not prove that self-reconfiguring trading systems will outperform in real markets. It does not solve regime identification, model risk, data leakage, market impact, adversarial behavior, or regulatory accountability. Those challenges become more important—not less important—when architecture itself becomes adaptive. The notebook instead establishes a pedagogical object that students can inspect: a system in which the organization becomes an endogenous variable while institutional boundaries remain exogenous.

That distinction suggests a broader research program. Future notebooks could allow a richer specialist library, multiple assets, portfolio construction agents, compliance agents, execution venues, dynamic budgets, or competing organizational designs. More advanced work could study when adaptation should occur, how long a specialist should remain active, how to prevent organizational overfitting, how to assign accountability across agents, and how a human committee should intervene when machine-generated reconfiguration affects capital allocation.

For an Algorithmic Trading course, the most important outcome is conceptual. Artificial intelligence should not be introduced merely as a new predictor placed beside momentum, regression, or neural networks. AI can also operate at the level of **architecture**. It can diagnose when the current organization is inadequate, propose a different division of cognitive labor, and document the reasoning behind the proposal. Deterministic financial models can continue to perform the calculations they are best suited for, while language models contribute flexible organizational reasoning.

The progression for students is therefore clear:

**Model → Agent → Constellation → Feedback → Organizational Adaptation → Governed Recursive System.**

The ambition is not to create more agents. It is to create the organization that the problem requires—and to ensure that every act of organizational change remains observable, testable, reversible, and subordinate to institutional governance.
