#  Constitutional Swarm: When the Rules Become Endogenous

## From endogenous coalitions to endogenous institutions

### Introduction

This notebook studies a deeper form of endogeneity in agentic systems. The question is no longer only whether artificial agents can discover a solution together. It is whether a population of agents can discover **that collaboration is necessary, discover that other intelligent agents exist, create a communication space, propose rules for collaboration, select among those rules, and then organize itself under the institutional framework it has chosen**. The notebook therefore moves beyond the familiar problem of multi-agent coordination and toward the more ambitious problem of endogenous institutional formation.

The conceptual starting point is the earlier deterministic **Swarm Discovery** experiment. In that framework, heterogeneous agents searched independently, made local discoveries, emitted epistemic signals, responded when another agent crossed a competence frontier, and formed coalitions. The collaboration graph was not imposed by a central orchestrator; it emerged from the problem. That was already an important departure from conventional agentic architecture. Nevertheless, the system still contained a substantial amount of institutional structure supplied ex ante. The programmer determined how signals were emitted, how agents responded, how contracts were formed, how rewards were divided, and what counted as legitimate collaboration. The network was endogenous, but the constitution of the network was largely exogenous.

NB04 pushes the experiment one level deeper. Agents begin without knowing that other agents exist. Each agent initially experiences the environment as if it were alone. It has a specialty, a finite search budget, a private evidence ledger, and a mandate to identify information that may matter for an acquisition problem. It searches independently. When it finds something potentially important, an LLM evaluates the discovery semantically rather than merely counting keywords. The LLM estimates relevance, discovery value, novelty, missing expertise, and a compact trace that can be left in the environment. These traces are called **epistemic pheromones**. They are deliberately weaker than messages. A pheromone does not say, “I am Agent A and I want to collaborate.” It simply records that some intelligent process appears to have encountered something worth noticing.

This distinction matters pedagogically. The agents do not begin as members of a society. They first have to discover plurality. When one agent encounters a trace that it did not create, it obtains evidence that another intelligence has been active in the same environment. The discovery is therefore not merely informational; it is social. The agent learns something about the ontology of its world: **I am not alone**. Once enough agents make that inference, a dormant bulletin-board primitive becomes socially meaningful. The board is not a central planner. It does not assign tasks, choose partners, or determine the correct institutional form. It is simply a shared surface on which agents can make discoveries, needs, and proposals visible.

At this point the experiment changes character. The principal problem is no longer “Which company is the acquisition target?” A second-order question appears: **If multiple intelligent agents can help one another, under what rules should they collaborate?** This is the constitutional moment of the notebook.

Claude Sonnet 5 is used specifically where open-ended semantic judgment and institutional creativity matter. First, it evaluates local discoveries. A deterministic filter can identify documents containing words such as “patent,” “desalination,” or “Gulf,” but it cannot as naturally judge whether a particular clue materially changes the acquisition thesis, why it matters, what uncertainty remains, or which missing expertise would be most valuable. The LLM converts raw evidence into an interpreted discovery. Second, the LLM is used to generate competing constitutions. Agents with different specialties and incentives propose alternative social compacts governing information sharing, verification, coalition membership, rewards, dispute resolution, anti-herding safeguards, and coalition termination. The constitutions are not hard-coded templates in the conceptual sense; they are candidate institutional designs produced in response to the situation.

The notebook nevertheless remains deliberately hybrid. It does **not** delegate every mechanism to the LLM. Search budgets, identities, the persistence of pheromones, the existence of a bulletin-board primitive, the voting mechanism, and the basic computational environment remain exogenous. Voting itself is transparent and deterministic so that the institutional choice can be inspected rather than hidden inside repeated model calls. This is a crucial methodological principle: use deterministic mechanisms when the rule is known and observability is valuable; use LLM judgment where meaning, analogy, interpretation, or institutional invention is genuinely open-ended.

The resulting architecture can be represented as a progression:

$$ \text{Private Search} \rightarrow \text{LLM-Valued Discovery} \rightarrow \text{Pheromone} \rightarrow \text{Discovery of Other Minds} \rightarrow \text{Bulletin Board} \rightarrow \text{Constitutional Proposals} \rightarrow \text{Collective Choice} \rightarrow \text{Governed Coalition} \rightarrow \text{Collective Hypothesis}. $$

The key word is **governed**. A coalition is no longer simply a set of agents that happen to connect. Its behavior depends on an adopted social compact. The constitution influences thresholds for sharing, conditions for joining, verification requirements, reward logic, and the manner in which collaboration can expand, split, or terminate. Organization therefore becomes problem-contingent at two levels: the network of collaborators can emerge, and the rules governing that network can also emerge.

This is not the same as claiming that the system is rule-free. A completely rule-free computational society is impossible. Every artificial environment requires substrate: interfaces, permissions, memory, compute, identity, and boundaries. The more precise research question is therefore: **How much of the substantive organization can be made endogenous while retaining a minimal, explicit, governable substrate?** NB04 should be read as an experiment in moving that boundary.

The acquisition setting provides a concrete pedagogical laboratory. The synthetic universe contains a hidden target whose attractiveness cannot be established by one type of evidence alone. Financial quality, leverage, technology, intellectual property, Gulf strategic fit, ownership willingness, and other clues are distributed across documents and specialties. No single agent has a complete view. This creates the conditions under which the value of complementary intelligence becomes visible. Importantly, collaboration is not assumed at the beginning. It becomes rational only after discoveries expose competence gaps and environmental traces reveal that complementary minds may exist.

The deeper lesson is therefore not about acquisition analysis. It is about the architecture of intelligence. Conventional systems begin with an organization and ask agents to solve a problem inside it. NB04 reverses part of that logic. The problem helps reveal the need for organization; communication makes organization possible; LLM judgment helps agents interpret both evidence and one another; and institutional proposals allow the agents to participate in defining the rules of their own collaboration.

This produces the conceptual transition at the center of the notebook:

$$ \boxed{\text{ENDOGENOUS COALITION} \; \rightarrow\; \text{ENDOGENOUS CONSTITUTION} \; \rightarrow\; \text{ENDOGENOUS INSTITUTION}} $$

The ten code units that follow are designed as a pedagogical ladder. Each unit isolates one conceptual transition so that the reader can see exactly which parts of the world are fixed, which parts are discovered, which judgments are delegated to Claude Sonnet 5, and which organizational objects emerge from interaction. The objective is not to hide complexity behind autonomy. It is to make the production of autonomy inspectable.

By the end of the notebook, the system will have done more than rank evidence. It will have moved from isolated search to social awareness, from social awareness to public communication, from communication to constitutional choice, and from constitutional choice to a governed coalition capable of producing a collective hypothesis. In that sense, NB04 is a bridge between multi-agent systems and a more ambitious idea: **artificial systems in which intelligence participates in designing the organization through which intelligence itself is coordinated.**

## Code Unit 1 — Runtime, Claude Sonnet 5, and the Minimal Constitutional Substrate

The first code unit establishes the environment in which the constitutional swarm can exist. This is intentionally more than routine setup. In an experiment about endogenous institutions, we must be explicit about what is **not** endogenous. Every artificial society begins inside a substrate that somebody designed. Here that substrate includes the Python runtime, the Anthropic interface, a model identifier, a finite API-call budget, deterministic random seeds, and utility functions for converting model responses into structured JSON. These elements are analogous to the physical and legal infrastructure within which a human institution operates: they create possibility without specifying the substantive organization that will later emerge.

Claude Sonnet 5 is placed behind one configuration variable. This makes the model a replaceable reasoning component rather than the architecture itself. The distinction is important. The notebook is not demonstrating that a frontier model can solve an acquisition problem by itself. It is demonstrating how model judgment can be embedded inside a larger system whose social structure changes over time. The API-call counter also makes scarcity visible. Intelligence is not treated as free; the experiment has a computational budget.

The code additionally creates the reusable LLM helper that requests structured output. This matters because later stages depend on machine-readable judgments: relevance probabilities, novelty, missing expertise, pheromone text, and constitutional proposals. The helper therefore acts as a narrow institutional interface between probabilistic language-model reasoning and deterministic system logic.

Pedagogically, this unit establishes the central discipline of the notebook: **endogeneity is always relative to a substrate**. We will allow discoveries, social awareness, constitutions, and coalition structure to emerge, but we will not pretend that the computational universe itself emerged from nothing. Keeping that boundary explicit is essential for both scientific interpretation and governance.

````python
from google.colab import userdata
from anthropic import Anthropic

#  LLM CONFIGURATION
# ----------------------------
API_KEY = userdata.get("ANTHROPIC_API_KEY")
if not API_KEY:
    raise ValueError(
        "Missing Colab secret ANTHROPIC_API_KEY. "
        "Add it in Colab > Secrets before running."
    )

client = Anthropic(api_key=API_KEY)

# Keep the model name in ONE variable so it can be changed easily.
# Replace with the Claude Sonnet model available in your Anthropic account if needed.
CLAUDE_MODEL = "claude-sonnet-5"

MAX_API_CALLS = 28
api_calls = 0

def claude_json(system, prompt, max_tokens=900):
    """Small governed LLM gateway. Requires a JSON object as output."""
    global api_calls
    if api_calls >= MAX_API_CALLS:
        raise RuntimeError(f"LLM call budget exhausted: {api_calls}/{MAX_API_CALLS}")
    api_calls += 1

    msg = client.messages.create(
        model=CLAUDE_MODEL,
        max_tokens=max_tokens,
        temperature=0.25,
        system=system,
        messages=[{"role":"user","content":prompt}]
    )
    text = "".join(
        block.text for block in msg.content if getattr(block, "type", "") == "text"
    ).strip()

    # Robustly extract first JSON object.
    m = re.search(r"\{.*\}", text, flags=re.S)
    if not m:
        raise ValueError("LLM did not return a JSON object:\n" + text)
    return json.loads(m.group(0))

# ----------------------------
````

## Code Unit 2 — Build the Synthetic Acquisition Universe and the Isolated Agents

The second unit creates the world the agents will inhabit. The acquisition universe is synthetic because the purpose is architectural rather than empirical: we want a controlled environment in which the correct target is knowable to the experiment designer but not directly available to any single agent. The hidden company, C0173, is attractive only when multiple fragments of evidence are combined. Financial performance, leverage, desalination technology, intellectual property, ownership structure, and Gulf strategic relevance are deliberately distributed across records. Decoys contain partial matches so that superficial keyword search is insufficient.

This design creates a meaningful reason for heterogeneous expertise. Agents are assigned specialties such as finance, engineering, patents, Gulf strategy, ownership, and synthesis. They also receive bounded search budgets and private state. Most importantly, they begin with a false but operationally useful assumption: **each behaves as though it is alone**. There is no initial team, no directory of peers, and no orchestrator announcing that complementary expertise is available.

The code then performs private search. Each specialty interprets the same universe through a different lens, producing different candidate documents and evidence opportunities. This is the first source of structural diversity. The agents are not merely copies of one another running in parallel; their specialties create different local views of the problem.

The conceptual point is that social organization should not be smuggled into the experiment at initialization. If the agents begin as a predefined team, later collaboration proves very little about institutional emergence. By starting with isolated searchers, NB04 creates a genuine transition from individual cognition to social cognition. The problem exists first. Organization must be discovered later. This allows us to observe the moment when independent intelligence becomes aware that additional intelligence may exist and may be useful.


````python
import pandas as pd
from dataclasses import dataclass, field
import random

# SYNTHETIC ACQUISITION UNIVERSE
# ----------------------------
MANDATE = """
Find an acquisition target with:
- EBITDA roughly USD 30–80m
- leverage preferably <= 2.5x
- defensible desalination / water technology
- meaningful intellectual property
- credible Gulf-market compatibility
- owners who may realistically transact
"""

HIDDEN_TARGET = "C0173"

companies = pd.DataFrame([
    ["C0173",58,1.65,"Europe","desalination membranes","family"],
    ["C0042",63,3.40,"Europe","industrial filtration","private equity"],
    ["C0091",44,1.90,"Asia","desalination membranes","founder"],
    ["C0211",71,2.10,"North America","water analytics","corporate"],
    ["C0068",36,2.20,"Europe","membrane coatings","family"],
    ["C0135",82,1.80,"Middle East","water infrastructure","family"],
    ["C0198",52,2.70,"Asia","industrial membranes","founder"],
    ["C0220",47,1.55,"Europe","water treatment","private equity"],
], columns=["company_id","ebitda_m","leverage","region","technology","ownership"])

documents = pd.DataFrame([
    ["D001","C0173","engineering","Pilot results show high-performance desalination membranes under high salinity and temperature conditions."],
    ["D002","C0173","patent","Patent family contains proprietary membrane-coating claims with long remaining life and process know-how."],
    ["D003","C0173","market","Technology has passed certification work and distributor discussions in Saudi Arabia and the UAE."],
    ["D004","C0173","ownership","Family succession planning has accelerated; advisers have informally explored strategic alternatives and a possible exit."],
    ["D005","C0173","finance","EBITDA is approximately USD 58m, leverage 1.65x, with improving cash conversion."],

    ["D006","C0042","finance","EBITDA is USD 63m but leverage has increased to 3.4x after a sponsor recapitalization."],
    ["D007","C0042","market","Strong industrial filtration franchise with modest exposure to municipal water markets."],
    ["D008","C0042","ownership","Private equity owner is considering liquidity alternatives within two years."],

    ["D009","C0091","engineering","Company produces desalination membranes with attractive laboratory performance."],
    ["D010","C0091","patent","Core membrane patents expire relatively soon and several claims face design-around risk."],
    ["D011","C0091","market","No meaningful Gulf certification or distribution infrastructure has been established."],

    ["D012","C0211","finance","EBITDA is USD 71m with leverage near 2.1x."],
    ["D013","C0211","engineering","Business focuses primarily on software analytics for water networks rather than membrane technology."],
    ["D014","C0211","ownership","Corporate parent views the unit as strategically important and not currently for sale."],

    ["D015","C0068","engineering","Membrane coating technology improves durability but has not yet been demonstrated in full desalination systems."],
    ["D016","C0068","ownership","Family owners have discussed succession but remain committed to independent growth."],

    ["D017","C0135","market","Strong Gulf relationships and installed water infrastructure base."],
    ["D018","C0135","finance","EBITDA exceeds the client's preferred range at approximately USD 82m."],

    ["D019","C0198","engineering","Industrial membrane portfolio has some desalination applicability but limited proprietary differentiation."],
    ["D020","C0198","finance","EBITDA is USD 52m and leverage is approximately 2.7x."],

    ["D021","C0220","finance","EBITDA is USD 47m with leverage of 1.55x."],
    ["D022","C0220","market","Water treatment business has attractive recurring service revenue but limited proprietary desalination IP."],
], columns=["doc_id","company_id","doc_type","text"])

SPECIALTIES = [
    "finance","engineering","patents","gulf_strategy",
    "ownership","synthesis"
]

SPECIALTY_WORDS = {
    "finance":["ebitda","leverage","cash","revenue"],
    "engineering":["desalination","membrane","salinity","temperature","coating"],
    "patents":["patent","claims","proprietary","know-how","ip"],
    "gulf_strategy":["gulf","saudi","uae","certification","distributor"],
    "ownership":["family","succession","adviser","exit","sale","alternatives"],
    "synthesis":["ebitda","leverage","desalination","membrane","patent","gulf","family","exit"],
}

@dataclass
class Agent:
    agent_id: str
    specialty: str
    budget: int = 5
    knows_others_exist: bool = False
    knows_board: bool = False
    discoveries: list = field(default_factory=list)
    seen_pheromones: list = field(default_factory=list)
    constitution_vote: str = None
    coalition: list = field(default_factory=list)

agents = [
    Agent(f"A{i+1:02d}", SPECIALTIES[i % len(SPECIALTIES)])
    for i in range(18)
]
agent_map = {a.agent_id:a for a in agents}

print("="*76)
print("CONSTITUTIONAL SWARM")
print("="*76)
print("Agents released:", len(agents))
print("At t=0 each agent believes it is searching alone.")
print("Hidden benchmark target:", HIDDEN_TARGET)
print("LLM call budget:", MAX_API_CALLS)

# ----------------------------
# PRIVATE SEARCH — NO AGENT KNOWS OTHERS EXIST
# ----------------------------
SEED = 42 # Define SEED for reproducibility
def lexical_fit(text, specialty):
    t = text.lower()
    return sum(t.count(w) for w in SPECIALTY_WORDS[specialty])

private_search = []
for a in agents:
    pool = documents.sample(frac=1, random_state=SEED + int(a.agent_id[1:]))
    pool = pool.assign(score=pool.text.apply(lambda x: lexical_fit(x,a.specialty)))
    # Exploit specialty + retain one exploratory surprise.
    chosen = pd.concat([
        pool.sort_values("score",ascending=False).head(3),
        pool.sample(1, random_state=1000+int(a.agent_id[1:]))
    ]).drop_duplicates("doc_id").head(a.budget)

    for _,d in chosen.iterrows():
        private_search.append({
            "agent_id":a.agent_id,
            "specialty":a.specialty,
            "doc_id":d.doc_id,
            "company_id":d.company_id,
            "text":d.text
        })

private_search_df = pd.DataFrame(private_search)
print("\nPHASE I — PRIVATE EXPLORATION")
print("Search events:", len(private_search_df))

# ----------------------------
````

## Code Unit 3 — Claude Sonnet 5 as a Judge of Discovery Value

Private search produces candidate evidence, but candidate evidence is not the same thing as a meaningful discovery. This unit introduces the first substantive role for Claude Sonnet 5: semantic valuation. The model examines a local finding in the context of the acquisition mandate and returns a structured judgment about its probability of relevance, discovery value, novelty, why it matters, what expertise is still missing, and what compact trace should be left behind.

This is deliberately different from asking the LLM to “solve the deal.” The model is acting as a **local epistemic judge**. Its task is bounded: interpret the significance of a discovery made by an agent. This placement makes the architecture easier to inspect because the model’s contribution can be compared with the underlying evidence and with later system behavior.

The distinction between relevance and value is pedagogically important. A document can be relevant without being decisive. A clue about a membrane patent may strongly relate to the mandate, but its value depends on whether it changes confidence in technological defensibility. Novelty matters because repeating an already-known fact should not trigger the same organizational response as revealing a new uncertainty. Missing expertise matters because it transforms a discovery into a potential demand for collaboration.

The output therefore has both epistemic and organizational consequences. Claude’s judgment does not merely update a score. It can identify the boundary of the current agent’s competence and encode a signal that another kind of intelligence would be useful.

This is the first point at which language-model reasoning begins to shape architecture. The LLM is not routing agents through a predefined workflow. Instead, by interpreting what a discovery means and what remains unresolved, it helps create the conditions under which a future relationship may become valuable. Meaning becomes a precursor to organization.


````python
# ============================================================
# CELL 3 — LLM ASSESSMENT OF DISCOVERY VALUE
# ============================================================
# We do NOT call the LLM for every private read.
#
# STEP 1 — Deterministic compression
# STEP 2 — Claude Sonnet 5 semantic evaluation
# STEP 3 — Validate structured output
# STEP 4 — Store valued discoveries in agent memory
#
# IMPORTANT:
# This cell calls client.messages.create() DIRECTLY.
# It therefore does NOT depend on the earlier claude_json()
# helper that contained the incompatible temperature argument.
# ============================================================

import json
import re
import pandas as pd


# ------------------------------------------------------------
# 1. DETERMINISTIC COMPRESSION
# ------------------------------------------------------------

candidate_pool = private_search_df.copy()

candidate_pool["lexical"] = [
    lexical_fit(text, specialty)
    for text, specialty in zip(
        candidate_pool["text"],
        candidate_pool["specialty"]
    )
]

candidate_pool = (
    candidate_pool
    .sort_values("lexical", ascending=False)
    .drop_duplicates(["specialty", "company_id"])
)

# Preserve heterogeneous expertise:
# up to two candidate discoveries per specialty.
candidate_reads = (
    candidate_pool
    .groupby("specialty", group_keys=False)
    .head(2)
    .sort_values("lexical", ascending=False)
    .head(12)
    .reset_index(drop=True)
)

print("=" * 76)
print("CELL 3 — LLM ASSESSMENT OF DISCOVERY VALUE")
print("=" * 76)

print("Private search events:", len(private_search_df))
print("Candidates retained:", len(candidate_reads))

print("\nCandidates by specialty:")
display(
    candidate_reads
    .groupby("specialty")
    .size()
    .rename("candidates")
)


# ------------------------------------------------------------
# 2. SYSTEM PROMPT
# ------------------------------------------------------------

DISCOVERY_SYSTEM = """
You are a disciplined M&A discovery evaluator operating inside
a multi-agent acquisition-search experiment.

Evaluate ONLY the single piece of evidence supplied.

Do not invent facts.
Do not infer unsupported company characteristics.
Do not decide whether the company is the final acquisition target.

Your task is to evaluate how much THIS DISCOVERY advances the
client's acquisition mandate.

Return ONLY valid JSON with exactly these six fields:

{
  "relevance_probability": 0.0,
  "discovery_value": 0.0,
  "novelty": 0.0,
  "reason": "",
  "missing_expertise": "",
  "pheromone": ""
}

Definitions:

relevance_probability:
Number from 0 to 1 measuring the probability that THIS DISCOVERY
materially advances the mandate.

discovery_value:
Number from 0 to 100 measuring the overall usefulness of THIS
DISCOVERY.

novelty:
Number from 0 to 1 measuring how new or non-obvious the information is.

reason:
One concise sentence explaining why the evidence matters.

missing_expertise:
Exactly one of:
finance
engineering
patents
gulf_strategy
ownership
synthesis

Choose the expertise that would be most useful NEXT.

pheromone:
A very short clue that another unknown specialist could notice.

Remember:

relevance_probability is NOT the probability that the company
is the final acquisition target.

It is the probability that THIS PIECE OF EVIDENCE materially
advances the search.
"""


# ------------------------------------------------------------
# 3. ROBUST JSON PARSER
# ------------------------------------------------------------

def parse_claude_json(text):
    """
    Extract a JSON object from Claude's response.
    Handles both raw JSON and accidental markdown fences.
    """

    text = text.strip()

    # Remove markdown code fences if present.
    text = re.sub(
        r"^```(?:json)?\s*",
        "",
        text,
        flags=re.IGNORECASE
    )

    text = re.sub(
        r"\s*```$",
        "",
        text
    )

    # Find first complete-looking JSON object.
    start = text.find("{")
    end = text.rfind("}")

    if start == -1 or end == -1 or end <= start:
        raise ValueError(
            f"No JSON object found in Claude response:\n{text}"
        )

    return json.loads(
        text[start:end + 1]
    )


# ------------------------------------------------------------
# 4. DIRECT CLAUDE CALL
# ------------------------------------------------------------

def evaluate_discovery_with_claude(system_prompt, user_prompt):
    """
    Direct Anthropic API call.

    IMPORTANT:
    No temperature argument is supplied.
    """

    global api_calls

    if api_calls >= MAX_API_CALLS:
        raise RuntimeError(
            f"API call budget exhausted: "
            f"{api_calls}/{MAX_API_CALLS}"
        )

    response = client.messages.create(
        model=CLAUDE_MODEL,
        max_tokens=800,
        system=system_prompt,
        messages=[
            {
                "role": "user",
                "content": user_prompt
            }
        ]
    )

    api_calls += 1

    # Collect text blocks returned by Claude.
    text_parts = []

    for block in response.content:
        if getattr(block, "type", None) == "text":
            text_parts.append(block.text)

    raw_text = "".join(text_parts).strip()

    if not raw_text:
        raise ValueError(
            "Claude returned no text content."
        )

    return parse_claude_json(raw_text)


# ------------------------------------------------------------
# 5. VALIDATION
# ------------------------------------------------------------

REQUIRED_FIELDS = {
    "relevance_probability",
    "discovery_value",
    "novelty",
    "reason",
    "missing_expertise",
    "pheromone"
}

VALID_SPECIALTIES = {
    "finance",
    "engineering",
    "patents",
    "gulf_strategy",
    "ownership",
    "synthesis"
}


def validate_assessment(assessment):

    if not isinstance(assessment, dict):
        raise TypeError(
            "Claude assessment must be a dictionary."
        )

    missing = REQUIRED_FIELDS - set(assessment.keys())

    if missing:
        raise ValueError(
            f"Claude response missing fields: {missing}"
        )

    # ------------------------
    # Normalize numeric values
    # ------------------------

    assessment["relevance_probability"] = float(
        assessment["relevance_probability"]
    )

    assessment["discovery_value"] = float(
        assessment["discovery_value"]
    )

    assessment["novelty"] = float(
        assessment["novelty"]
    )

    # ------------------------
    # Clamp valid ranges
    # ------------------------

    assessment["relevance_probability"] = max(
        0.0,
        min(
            1.0,
            assessment["relevance_probability"]
        )
    )

    assessment["discovery_value"] = max(
        0.0,
        min(
            100.0,
            assessment["discovery_value"]
        )
    )

    assessment["novelty"] = max(
        0.0,
        min(
            1.0,
            assessment["novelty"]
        )
    )

    # ------------------------
    # Normalize text
    # ------------------------

    assessment["reason"] = str(
        assessment["reason"]
    ).strip()

    assessment["pheromone"] = str(
        assessment["pheromone"]
    ).strip()

    assessment["missing_expertise"] = str(
        assessment["missing_expertise"]
    ).strip().lower()

    if assessment["missing_expertise"] not in VALID_SPECIALTIES:
        assessment["missing_expertise"] = "synthesis"

    return assessment


# ------------------------------------------------------------
# 6. SEMANTIC EVALUATION OF CANDIDATE DISCOVERIES
# ------------------------------------------------------------

discoveries = []

print("\nPHASE II — LLM-VALUED LOCAL DISCOVERIES")
print("-" * 76)


for i, r in candidate_reads.iterrows():

    # Retrieve document type.
    doc_match = documents.loc[
        documents["doc_id"] == r["doc_id"],
        "doc_type"
    ]

    if doc_match.empty:
        raise ValueError(
            f"Document {r['doc_id']} not found."
        )

    doc_type = doc_match.iloc[0]


    # --------------------------------------------------------
    # Construct the evidence packet Claude receives.
    # --------------------------------------------------------

    user_prompt = f"""
CLIENT MANDATE
==============
{MANDATE}

SEARCHER
========
Agent: {r['agent_id']}
Specialty: {r['specialty']}

DISCOVERY
=========
Company: {r['company_id']}
Document: {r['doc_id']}
Document type: {doc_type}

EVIDENCE
========
{r['text']}

TASK
====
Evaluate the discovery value of THIS evidence.

Do not decide whether the company is the final target.

Return only the required JSON object.
"""


    # --------------------------------------------------------
    # CALL CLAUDE DIRECTLY
    # --------------------------------------------------------

    assessment = evaluate_discovery_with_claude(
        DISCOVERY_SYSTEM,
        user_prompt
    )


    # --------------------------------------------------------
    # Validate output
    # --------------------------------------------------------

    assessment = validate_assessment(
        assessment
    )


    # --------------------------------------------------------
    # Create provenance-preserving discovery object
    # --------------------------------------------------------

    discovery = {

        "agent_id": r["agent_id"],

        "specialty": r["specialty"],

        "doc_id": r["doc_id"],

        "company_id": r["company_id"],

        "doc_type": doc_type,

        "text": r["text"],

        "lexical": r["lexical"],

        **assessment
    }


    discoveries.append(
        discovery
    )


    # --------------------------------------------------------
    # Store discovery in originating agent's memory
    # --------------------------------------------------------

    agent_map[
        r["agent_id"]
    ].discoveries.append(
        discovery
    )


    print(
        f"[{i + 1:02d}/{len(candidate_reads):02d}]",
        r["agent_id"],
        "|",
        r["specialty"],
        "|",
        r["company_id"],
        "| discovery value =",
        round(
            assessment["discovery_value"],
            1
        )
    )


# ------------------------------------------------------------
# 7. CREATE DISCOVERY DATAFRAME
# ------------------------------------------------------------

if len(discoveries) == 0:
    raise RuntimeError(
        "Cell 3 produced no discoveries."
    )


discovery_df = (
    pd.DataFrame(discoveries)
    .sort_values(
        "discovery_value",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 8. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n")
print("=" * 76)
print("LLM-VALUED LOCAL DISCOVERIES")
print("=" * 76)


display_columns = [

    "agent_id",

    "specialty",

    "company_id",

    "doc_id",

    "doc_type",

    "discovery_value",

    "relevance_probability",

    "novelty",

    "missing_expertise",

    "reason",

    "pheromone"
]


display(
    discovery_df[
        display_columns
    ]
)


# ------------------------------------------------------------
# 9. CELL SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 76)
print("CELL 3 SUMMARY")
print("=" * 76)

print(
    "Private search events:",
    len(private_search_df)
)

print(
    "Candidate discoveries evaluated:",
    len(candidate_reads)
)

print(
    "LLM-valued discoveries produced:",
    len(discovery_df)
)

print(
    "API calls used:",
    api_calls,
    "/",
    MAX_API_CALLS
)


best = discovery_df.iloc[0]

print("\nHighest-value local discovery:")

print(
    " Agent:",
    best["agent_id"]
)

print(
    " Specialty:",
    best["specialty"]
)

print(
    " Company:",
    best["company_id"]
)

print(
    " Document:",
    best["doc_id"]
)

print(
    " Discovery value:",
    round(
        best["discovery_value"],
        1
    )
)

print(
    " Missing expertise:",
    best["missing_expertise"]
)

print(
    " Pheromone:",
    best["pheromone"]
)


print("\n" + "-" * 76)

print(
    "CELL 3 COMPLETE\n"
    "Private observations have been transformed into "
    "semantically valued discoveries.\n"
    "The next step is to allow sufficiently valuable discoveries "
    "to leave traces in the environment."
)
````

## Code Unit 4 — Epistemic Pheromones: Leaving Traces Without Sending Messages

The fourth unit converts high-value discoveries into **epistemic pheromones**. The metaphor is intentionally biological. In many decentralized systems, coordination begins not through direct negotiation but through persistent traces left in a shared environment. An ant does not need a global map or a central dispatcher to influence another ant. A local trace changes the probability of another actor’s future behavior.

The same principle is used here, but the trace carries epistemic content. When Claude Sonnet 5 judges a discovery sufficiently important, the discovering agent leaves a compact signal in the environment. The trace is not initially a social message. It does not assume that the agent knows who will read it, whether anyone else exists, or whether collaboration is possible. It simply externalizes evidence that some intelligent process found something noteworthy.

This distinction protects the logic of the experiment. If agents immediately addressed one another by name, social awareness would already have been supplied by the programmer. Pheromones allow information to persist before society exists. They create the possibility that one agent can encounter the residue of another agent’s cognition and infer plurality from the environment itself.

The unit also illustrates an important design principle for decentralized systems: **communication can be indirect before it becomes direct**. Shared state can precede shared identity. A trace can precede a relationship.

From a governance perspective, pheromones are also useful because they are inspectable. We can see which discovery generated the trace, what value threshold was crossed, and which semantic summary was exposed. This makes the transition from private reasoning to environmental signaling observable. The system has not yet formed coalitions or constitutions. It has merely created evidence that intelligence leaves footprints—and those footprints can become the first bridge between isolated minds.


````python
#  PHEROMONES — TRACES, NOT MESSAGES
# ----------------------------
# Agents do not initially know a bulletin board exists.
# High-value discoveries leave anonymous environmental traces.
pheromones = []

for i,d in discovery_df.iterrows():
    if d.discovery_value >= 55 and d.novelty >= 0.35:
        pheromones.append({
            "pheromone_id":f"P{i:03d}",
            "company_id":d.company_id,
            "location":d.doc_id,
            "trace":d.pheromone,
            "source_agent":d.agent_id,      # hidden from other agents initially
            "needed":d.missing_expertise,
            "strength":round(
                0.55*d.relevance_probability +
                0.45*d.novelty, 3
            )
        })

pheromone_df = pd.DataFrame(pheromones)
print("\nPHASE III — PHEROMONES LEFT IN THE ENVIRONMENT")
print("Pheromones:", len(pheromone_df))
if len(pheromone_df):
    display(pheromone_df.drop(columns=["source_agent"]).head(12))

# ----------------------------
````

## Code Unit 5 — Discovering Other Minds and Activating the Bulletin Board

This unit contains the notebook’s first explicitly social transition. An agent encounters a pheromone that it did not create and infers that another intelligent process must have been active in the environment. The inference is simple, but conceptually it is profound. The agent’s model of the world changes from “I am searching alone” to “other minds exist here.” That is the birth of social awareness.

The code records this transition in agent state. Once a sufficient number of agents become aware of plurality, the dormant bulletin-board primitive becomes socially active. The board has existed as infrastructure, but until now it had no practical meaning. A communication channel is useful only when an agent has reason to believe that someone else might receive the message.

This is why the bulletin board should not be interpreted as a central orchestrator. It does not decide who should collaborate, allocate tasks, or select the target. It is simply an addressable public surface. Agents can post discoveries, uncertainties, and needs. Other agents can read them. The board transforms unknown potential collaborators into reachable potential collaborators.

The sequence is therefore important: private search comes first; traces appear second; plurality is inferred third; direct public communication becomes rational fourth. The notebook does not begin with a team chat. It tries to model the conditions under which a team chat becomes useful.

Pedagogically, this unit marks the shift from **stigmergic coordination** to **linguistic coordination**. Pheromones say, in effect, “something happened here.” Bulletin-board messages can say, “this is what I found, this is what I do not understand, and this is the kind of help that might matter.” Language therefore expands the organizational bandwidth of the swarm. It allows agents not merely to react to traces but to articulate competence gaps and make collaboration opportunities legible.


````python
#  THE DISCOVERY OF OTHER MINDS
# ----------------------------
encounters = []

for a in agents:
    # An agent can encounter a pheromone only in a company it searched,
    # and never its own trace.
    searched_companies = set(
        private_search_df.loc[
            private_search_df.agent_id==a.agent_id,"company_id"
        ]
    )
    candidates = [
        p for p in pheromones
        if p["company_id"] in searched_companies
        and p["source_agent"] != a.agent_id
    ]

    if candidates:
        p = max(candidates,key=lambda x:x["strength"])
        a.seen_pheromones.append(p["pheromone_id"])
        a.knows_others_exist = True
        encounters.append({
            "agent_id":a.agent_id,
            "specialty":a.specialty,
            "pheromone_id":p["pheromone_id"],
            "company_id":p["company_id"],
            "trace":p["trace"],
            "inference":"Another intelligent searcher has been here."
        })

encounter_df = pd.DataFrame(encounters)

print("\nPHASE IV — AGENTS DISCOVER THAT THEY ARE NOT ALONE")
print("Agents encountering traces from another mind:", len(encounter_df))
if len(encounter_df):
    display(encounter_df.head(12))

# ----------------------------
#  BULLETIN BOARD EMERGES
# ----------------------------
# The board is a primitive capability, but it is dormant.
# It becomes socially relevant only after agents infer plurality.
aware_agents = [a for a in agents if a.knows_others_exist]

if len(aware_agents) >= 3:
    for a in aware_agents:
        a.knows_board = True
    board_exists_socially = True
else:
    board_exists_socially = False

print("\nPHASE V — BULLETIN BOARD")
print("Socially activated:", board_exists_socially)
print("Agents aware of board:", sum(a.knows_board for a in agents))

if not board_exists_socially:
    raise RuntimeError("Too few agents discovered one another to constitutionalize.")

# The first bulletin-board posts are deliberately minimal:
# "I found another mind; perhaps we need rules before sharing deeply."
board = []
for a in aware_agents:
    board.append({
        "author":a.agent_id,
        "type":"constitutional_call",
        "text":"I have encountered evidence of other searchers. "
               "Before sharing deeply, we should decide how collaboration works."
    })

# ----------------------------
````

### Understanding Pheromones and the Agent Process Flow

Let's clarify how pheromones are handled in this simulation and what happens afterward:

**1. Initialization and Creation of the Pheromone Field (Code Unit 4):**

*   **The pheromone field is not initially part of the raw data.** It is a *new feature* that is dynamically created by the system, based on the agents' discoveries.
*   After agents perform their initial searches and make discoveries, the system processes these discoveries.
*   Pheromones are generated only from specific high-value discoveries:
    *   If a discovery's `discovery_value` is `55` or higher.
    *   AND its `novelty` is `0.35` or higher.
*   For such discoveries, a new pheromone entry is added to the `pheromones` list (which then becomes `pheromone_df`). This entry includes details like the `company_id`, the `trace` (which is `d.pheromone` from the original discovery), the `source_agent`, and its `strength`.
*   So, agents don't find an empty pheromone field; rather, the field is populated *after* significant discoveries have been made by *any* agent.

**2. How Agents Detect Pheromones (Code Unit 5):**

*   **Agents do not 're-read the documents' to detect pheromones.** Instead, the detection mechanism operates on the `pheromones` list (or `pheromone_df`) that was generated in Code Unit 4.
*   For each agent, the system first identifies all `company_id`s that the agent has previously searched (`private_search_df`).
*   Then, it filters the global list of `pheromones` to find `candidates` that meet two criteria:
    1.  The pheromone's `company_id` must be among the companies the agent has already searched.
    2.  The `source_agent` of the pheromone **must not** be the `agent_id` of the current agent (`p["source_agent"] != a.agent_id`). This is crucial: an agent cannot detect its own pheromone, ensuring that the detection implies the presence of *another* mind.
*   If an agent finds any `candidates`, it selects the one with the highest `strength`. A stronger pheromone indicates a more significant prior discovery by another agent.
*   Upon detecting such a pheromone, the agent's state is updated: `a.knows_others_exist` is set to `True`, signifying that the agent has inferred the existence of other intelligent searchers.

**3. What happens next? The Communication Ontology and Agent Alerting:**

After pheromones are detected, agents are *not idle*. The simulation proceeds through several critical phases that build upon this initial social awareness:

*   **Emergence of the Bulletin Board (Code Unit 5, Section 6):**
    *   The bulletin board primitive exists from the start, but it's initially *dormant*. It only becomes socially active and meaningful *after* a sufficient number of agents (`aware_agents >= 3`) infer that others exist (`a.knows_others_exist = True`).
    *   This is a crucial transition: **stigmergic coordination** (via pheromones) gives way to the potential for **linguistic coordination** (via the bulletin board).
    *   Initially, agents post minimal messages to the board, essentially calling for a constitutional discussion to establish rules for collaboration.

*   **Constitutional Creativity (Code Unit 6):**
    *   Selected agents (proposers with different specialties) then leverage an LLM (Claude Sonnet 5) to **generate competing constitutions**.
    *   These constitutions define rules for sharing, joining coalitions, verification, rewards, and anti-herding mechanisms.
    *   This is how the system introduces *explicit rules* for interaction, rather than just implicit traces.

*   **Collective Choice: Voting on the Social Compact (Code Unit 7):**
    *   All `aware_agents` (those who discovered pheromones) then **vote** on which constitution to adopt.
    *   Agents don't use an LLM for voting; instead, they have transparent preference functions based on their specialty and local incentives. This determines which proposed constitution they find most beneficial.
    *   The winning constitution is then adopted, becoming the operational framework for future collaboration.

*   **Governing Sharing and Coalition Formation (Code Unit 8):**
    *   This is where agents actively receive 'more information' and are 'alerted' to opportunities for deeper collaboration, governed by the adopted constitution.
    *   **Public Signals:** The adopted constitution's `share_threshold` is applied to all `discoveries`. If a discovery's `relevance_probability` meets or exceeds this threshold, it is published as a `public_signal` on the bulletin board (`signal_df`). This `signal_df` acts as a publicly accessible repository of shared discoveries.
    *   **How Agents are Alerted & Consume Information:** Agents don't go 'back to the original documents.' Instead, they effectively **monitor the `signal_df`**. They are alerted to new information when `public_signals` matching their interests or needs are posted there. This is their new 'list' to check. This shared `signal_df` is a dynamic pool of high-value discoveries from other agents, filtered by the agreed-upon rules.
    *   **Coalition Formation:** Based on these `public_signals`, potential `coalition_edges` are formed. An agent is 'alerted' to a collaboration opportunity if their `specialty` matches the `needed` expertise specified in a `public_signal` posted by *another* agent. This forms a `coalition_edge` between the sender of the signal and the potential collaborator. This mechanism ensures that information sharing and the formation of collaborative groups are no longer spontaneous but are *governed by the rules agents themselves have chosen*.
    *   The `agent_map` and individual agents' `coalition` lists are updated to reflect these connections, indicating that the agents have formed potential collaborations based on the shared information and their complementary skills.

**Communication Ontology Summary:**

*   **Implicit Pheromones:** Early stage, anonymous environmental traces, leading to inference of 'other minds'. (Stigmergic coordination). Agents implicitly learn from the environment.
*   **Explicit Constitutional Calls:** Initial messages on the bulletin board to establish rules. (Rudimentary linguistic coordination).
*   **Rule-Governed Public Signals:** Structured information (discoveries) shared publicly, filtered by the adopted constitution's rules. (Formal linguistic coordination). Agents *actively consume* this `signal_df` as their new source of shared knowledge.
*   **Coalition Edges:** Agents are 'alerted' and connect based on complementary expertise and the shared public signals, forming explicit collaboration structures. (Rule-governed social interaction and directed information flow).

## Code Unit 6 — Claude Sonnet 5 Generates Competing Constitutions

Once agents can communicate, a new problem appears: communication alone does not determine how collaboration should work. Who may join a coalition? What evidence must be shared? How should claims be verified? How are rewards allocated? What prevents herding? When should a coalition split, merge, or dissolve? These are institutional questions rather than search questions.

This unit gives Claude Sonnet 5 its second major role: **constitutional creativity**. Several agents representing different specialties act as proposers. Each asks the model to design a social compact suited to the emerging environment. The requested constitution contains fields for a guiding principle, sharing threshold, joining rule, verification rule, reward rule, merge/split logic, and anti-herding protection.

The important point is not that the LLM writes elegant policy text. It is that institutional alternatives become objects inside the system. Instead of one collaboration protocol being silently embedded in code, multiple possible protocols are made explicit and contestable. The agents can therefore reason not only about the acquisition target but about the machinery of collaboration itself.

Different proposers may naturally emphasize different values. A finance-oriented proposer may care about incentives and contribution. A patent-oriented proposer may emphasize verification and provenance. A synthesis agent may prefer broader information sharing. The resulting constitutions expose trade-offs that would otherwise remain hidden inside architecture.

This is the notebook’s central move from endogenous topology to endogenous rules. In NB03, agents could generate the collaboration graph while operating under rules supplied by the designer. Here the system begins to generate candidate rules governing the graph.

The experiment still has a constitutional substrate—the fields that a proposal must address are specified ex ante—but the substantive answers are produced through model judgment. This bounded openness is intentional: it creates room for institutional invention without making the experiment impossible to inspect.


````python
# ============================================================
# CODE UNIT 6 — LLM-GENERATED CONSTITUTIONS
# ============================================================
# PURPOSE
# -------
# Three agents with different specialties become constitutional
# proposers.
#
# Each proposer:
#   1. Reviews its own recent discoveries.
#   2. Knows that other intelligent searchers exist.
#   3. Uses Claude Sonnet 5 to propose a compact constitution.
#
# IMPORTANT:
# This unit calls client.messages.create() DIRECTLY.
# It does NOT use the earlier claude_json() helper.
# Therefore it avoids the incompatible temperature argument.
# ============================================================

import json
import re
import pandas as pd


# ------------------------------------------------------------
# 1. SELECT CONSTITUTIONAL PROPOSERS
# ------------------------------------------------------------
# We deliberately seek three different perspectives:
# finance, engineering, and ownership.
# ------------------------------------------------------------

proposers = []

for spec in [
    "finance",
    "engineering",
    "ownership"
]:
    found = next(
        (
            a for a in aware_agents
            if a.specialty == spec
        ),
        None
    )

    if found:
        proposers.append(found)

proposers = proposers[:3]


if len(proposers) == 0:
    raise RuntimeError(
        "No eligible constitutional proposers were found."
    )


print("=" * 76)
print("CODE UNIT 6 — LLM-GENERATED CONSTITUTIONS")
print("=" * 76)

print(
    "Constitutional proposers:",
    len(proposers)
)

for proposer in proposers:
    print(
        " ",
        proposer.agent_id,
        "|",
        proposer.specialty
    )


# ------------------------------------------------------------
# 2. CONSTITUTIONAL DESIGN PROMPT
# ------------------------------------------------------------

CONSTITUTION_SYSTEM = """
You are an autonomous AI search agent who has just discovered
that other intelligent AI agents exist.

You are NOT designing a centrally planned workflow.

You are proposing a compact constitutional protocol that
autonomous agents could VOLUNTARILY adopt in order to cooperate.

The constitution must preserve agent autonomy while making
productive collaboration possible.

The constitution must specify six institutional dimensions:

1. SHARING
   When should discoveries be shared?

2. JOINING
   Under what conditions may complementary agents join
   a coalition?

3. VERIFICATION
   How must important evidence be checked before the coalition
   relies on it?

4. CREDIT AND REWARD
   How should credit or reward be divided among contributing
   agents?

5. MERGER AND SPLIT
   Under what conditions may coalitions merge, divide,
   or dissolve?

6. ANTI-HERDING
   What rule protects the group against premature consensus
   or herd behavior?

Do NOT assume:
- a central coordinator;
- a permanent hierarchy;
- mandatory participation;
- that one agent knows the entire problem;
- that one agent can command the others.

Return ONLY valid JSON.

The JSON must contain exactly these fields:

{
    "name": "",
    "principle": "",
    "share_threshold": 0.0,
    "join_rule": "",
    "verification_rule": "",
    "reward_rule": "",
    "merge_split_rule": "",
    "anti_herding_rule": ""
}

FIELD DEFINITIONS

name:
A short memorable name for the constitution.

principle:
One concise sentence stating its governing philosophy.

share_threshold:
A number from 0 to 1 indicating the minimum confidence or
discovery quality at which an agent should normally share
a finding.

join_rule:
One concise sentence defining when another autonomous agent
may join a coalition.

verification_rule:
One concise sentence defining how evidence should be verified.

reward_rule:
One concise sentence defining how credit or reward should
be allocated.

merge_split_rule:
One concise sentence defining when coalitions may merge,
split, or dissolve.

anti_herding_rule:
One concise sentence defining a safeguard against premature
consensus.

The goal is NOT to maximize central control.

The goal is to design rules under which autonomous agents
can voluntarily cooperate.
"""


# ------------------------------------------------------------
# 3. ROBUST JSON PARSER
# ------------------------------------------------------------

def parse_constitution_json(text):
    """
    Extract a JSON object from Claude's response.

    Handles:
      - clean JSON;
      - accidental ```json fences;
      - surrounding explanatory text.
    """

    text = text.strip()

    # Remove opening markdown fence.
    text = re.sub(
        r"^```(?:json)?\s*",
        "",
        text,
        flags=re.IGNORECASE
    )

    # Remove closing markdown fence.
    text = re.sub(
        r"\s*```$",
        "",
        text
    )

    # Locate JSON object.
    start = text.find("{")
    end = text.rfind("}")

    if start == -1 or end == -1 or end <= start:
        raise ValueError(
            "No valid JSON object found in Claude response.\n\n"
            f"RAW RESPONSE:\n{text}"
        )

    json_text = text[start:end + 1]

    try:
        return json.loads(json_text)

    except json.JSONDecodeError as e:
        raise ValueError(
            "Claude returned malformed JSON.\n\n"
            f"RAW RESPONSE:\n{text}\n\n"
            f"JSON ERROR:\n{e}"
        )


# ------------------------------------------------------------
# 4. DIRECT CLAUDE SONNET 5 CALL
# ------------------------------------------------------------

def generate_constitution_with_claude(
    system_prompt,
    user_prompt
):
    """
    Call Claude directly.

    IMPORTANT:
    There is NO temperature argument here.

    This bypasses the earlier claude_json() helper that caused
    the TypeError in Code Unit 3.
    """

    global api_calls

    if api_calls >= MAX_API_CALLS:
        raise RuntimeError(
            f"API call budget exhausted: "
            f"{api_calls}/{MAX_API_CALLS}"
        )

    response = client.messages.create(
        model=CLAUDE_MODEL,
        max_tokens=1200,
        system=system_prompt,
        messages=[
            {
                "role": "user",
                "content": user_prompt
            }
        ]
    )

    api_calls += 1

    # Collect all text blocks.
    text_parts = []

    for block in response.content:

        if getattr(
            block,
            "type",
            None
        ) == "text":

            text_parts.append(
                block.text
            )

    raw_text = "".join(
        text_parts
    ).strip()

    if not raw_text:
        raise ValueError(
            "Claude returned no text content."
        )

    return parse_constitution_json(
        raw_text
    )


# ------------------------------------------------------------
# 5. VALIDATION CONTRACT
# ------------------------------------------------------------

REQUIRED_CONSTITUTION_FIELDS = {
    "name",
    "principle",
    "share_threshold",
    "join_rule",
    "verification_rule",
    "reward_rule",
    "merge_split_rule",
    "anti_herding_rule"
}


def validate_constitution(proposal):
    """
    Validate and normalize a constitution generated by Claude.
    """

    if not isinstance(
        proposal,
        dict
    ):
        raise TypeError(
            "Constitution must be returned as a dictionary."
        )

    missing = (
        REQUIRED_CONSTITUTION_FIELDS
        - set(proposal.keys())
    )

    if missing:
        raise ValueError(
            f"Constitution missing required fields: {missing}"
        )


    # --------------------------------------------------------
    # Normalize share threshold
    # --------------------------------------------------------

    proposal["share_threshold"] = float(
        proposal["share_threshold"]
    )

    proposal["share_threshold"] = max(
        0.0,
        min(
            1.0,
            proposal["share_threshold"]
        )
    )


    # --------------------------------------------------------
    # Normalize text fields
    # --------------------------------------------------------

    text_fields = [
        "name",
        "principle",
        "join_rule",
        "verification_rule",
        "reward_rule",
        "merge_split_rule",
        "anti_herding_rule"
    ]

    for field_name in text_fields:

        proposal[field_name] = str(
            proposal[field_name]
        ).strip()


    return proposal


# ------------------------------------------------------------
# 6. GENERATE COMPETING CONSTITUTIONS
# ------------------------------------------------------------

constitutions = []


print("\nPHASE VI — CONSTITUTIONAL CREATIVITY")
print("-" * 76)


for proposer in proposers:

    # --------------------------------------------------------
    # Retrieve only this proposer's own discoveries.
    # --------------------------------------------------------

    own = [
        d for d in discoveries
        if d["agent_id"] == proposer.agent_id
    ]


    # Rank own discoveries by discovery value.
    own = sorted(
        own,
        key=lambda d: d["discovery_value"],
        reverse=True
    )


    # --------------------------------------------------------
    # Compress local experience.
    # --------------------------------------------------------

    own_summary = [
        {
            "company": d["company_id"],
            "discovery_value": round(
                float(d["discovery_value"]),
                1
            ),
            "relevance_probability": round(
                float(d["relevance_probability"]),
                3
            ),
            "missing_expertise": d["missing_expertise"]
        }

        for d in own[:3]
    ]


    # --------------------------------------------------------
    # Build constitutional prompt.
    # --------------------------------------------------------

    user_prompt = f"""
YOUR IDENTITY
=============
Agent: {proposer.agent_id}
Specialty: {proposer.specialty}

YOUR EXPERIENCE
===============
You began this experiment believing that you were searching alone.

You later encountered anonymous pheromone traces left by other
intelligent searchers.

You therefore learned that:

1. other autonomous agents exist;
2. they possess different expertise;
3. useful discoveries can attract complementary minds;
4. deeper collaboration may require shared rules.

YOUR OWN RECENT DISCOVERIES
===========================
{json.dumps(own_summary, indent=2)}

YOUR TASK
=========
Propose a constitution for a spontaneous society of autonomous
search agents.

The constitution should make collaboration possible WITHOUT
creating a central coordinator.

It should preserve:

- autonomy;
- voluntary participation;
- evidence discipline;
- fair attribution;
- adaptability;
- diversity of judgment.

The rules should be sufficiently precise to govern actual
coalition behavior.

Return ONLY the required JSON object.
"""


    # --------------------------------------------------------
    # CALL CLAUDE DIRECTLY
    # --------------------------------------------------------

    proposal = generate_constitution_with_claude(
        CONSTITUTION_SYSTEM,
        user_prompt
    )


    # --------------------------------------------------------
    # Validate
    # --------------------------------------------------------

    proposal = validate_constitution(
        proposal
    )


    # --------------------------------------------------------
    # Add provenance
    # --------------------------------------------------------

    proposal["proposal_id"] = (
        f"CONST_{len(constitutions) + 1}"
    )

    proposal["proposer"] = (
        proposer.agent_id
    )

    proposal["proposer_specialty"] = (
        proposer.specialty
    )


    constitutions.append(
        proposal
    )


    print(
        proposal["proposal_id"],
        "|",
        proposer.agent_id,
        "|",
        proposer.specialty,
        "|",
        proposal["name"],
        "| share threshold =",
        round(
            proposal["share_threshold"],
            2
        )
    )


# ------------------------------------------------------------
# 7. CREATE CONSTITUTION DATAFRAME
# ------------------------------------------------------------

if len(constitutions) == 0:
    raise RuntimeError(
        "No constitutions were generated."
    )


constitution_df = (
    pd.DataFrame(
        constitutions
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 8. DISPLAY COMPETING CONSTITUTIONS
# ------------------------------------------------------------

print("\n")
print("=" * 76)
print("COMPETING CONSTITUTIONAL PROPOSALS")
print("=" * 76)


display_columns = [

    "proposal_id",

    "name",

    "proposer",

    "proposer_specialty",

    "principle",

    "share_threshold",

    "join_rule",

    "verification_rule",

    "reward_rule",

    "merge_split_rule",

    "anti_herding_rule"
]


display(
    constitution_df[
        display_columns
    ]
)


# ------------------------------------------------------------
# 9. PEDAGOGICAL SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 76)
print("CODE UNIT 6 SUMMARY")
print("=" * 76)

print(
    "Constitutional proposers:",
    len(proposers)
)

print(
    "Constitutions generated:",
    len(constitution_df)
)

print(
    "API calls used:",
    api_calls,
    "/",
    MAX_API_CALLS
)


print("\nConstitutional diversity:")

for _, row in constitution_df.iterrows():

    print(
        f"  {row['proposal_id']} | "
        f"{row['proposer_specialty']} | "
        f"{row['name']} | "
        f"share threshold = "
        f"{row['share_threshold']:.2f}"
    )


print("\n" + "-" * 76)

print(
    "CODE UNIT 6 COMPLETE\n\n"
    "The agents have moved beyond discovering one another.\n"
    "They are now proposing competing rules for how a society "
    "of autonomous minds might cooperate.\n\n"
    "The architecture is no longer merely discovering "
    "relationships.\n"
    "It is beginning to design its own institutions."
)
````

## Code Unit 7 — Collective Choice: Agents Vote on the Social Compact

Generating constitutions is not enough. A society requires some mechanism for selecting which rules will actually govern behavior. This unit therefore introduces collective choice. The candidate constitutions created by Claude Sonnet 5 are evaluated by the agents, and a winning social compact is selected through a transparent voting procedure.

The voting logic is intentionally deterministic. This is a methodological choice rather than a limitation. The LLM has already supplied open-ended institutional creativity; using another model call for every vote would increase cost and make the choice mechanism less inspectable. Instead, agents evaluate proposals according to local incentives and preferences that can be examined directly in code. The result is a useful separation between **institutional generation** and **institutional selection**.

This separation mirrors an important feature of human governance. The capacity to propose a constitution and the authority to adopt one are not necessarily the same thing. An institutional entrepreneur may design a rule, but a community must still decide whether that rule is acceptable. NB04 represents this distinction computationally.

The vote also makes heterogeneity consequential. Agents do not need to value openness, verification, incentives, or flexibility in exactly the same way. A constitution becomes operational because it survives a collective selection process, not because the programmer labels it “best.”

Pedagogically, this is the moment when the system moves from discussion to commitment. Before the vote, constitutions are ideas. After the vote, one proposal becomes a rule-bearing object that constrains subsequent collaboration.

The adopted constitution therefore creates a form of endogenous path dependence. Later coalition structure is no longer determined only by evidence and complementarity. It is filtered through rules that the agents themselves selected. The society has begun to design part of the environment in which its future actions will occur.


````python
# ============================================================
# CODE UNIT 7 — AGENTS VOTE ON THE SOCIAL COMPACT
# ============================================================
# PURPOSE
# -------
# Claude created the competing constitutions in Code Unit 6.
#
# Here we deliberately DO NOT call the LLM.
#
# Each aware agent evaluates every constitution using a
# transparent deterministic preference function based on:
#
#   1. fit with its preferred sharing threshold;
#   2. strength of verification rules;
#   3. protection of agent autonomy;
#   4. protection of cognitive diversity;
#   5. a small deterministic idiosyncratic preference term.
#
# The constitution receiving the largest number of votes
# becomes the adopted social compact.
#
# PEDAGOGICAL PRINCIPLE
# ---------------------
# LLM = institutional creativity
# deterministic mechanism = institutional choice
# ============================================================

import random
import pandas as pd
import json


# ------------------------------------------------------------
# 1. PRECONDITIONS
# ------------------------------------------------------------

if "constitutions" not in globals():
    raise RuntimeError(
        "The variable 'constitutions' does not exist. "
        "Run Code Unit 6 first."
    )

if len(constitutions) == 0:
    raise RuntimeError(
        "No constitutions are available for voting."
    )

if "aware_agents" not in globals():
    raise RuntimeError(
        "The variable 'aware_agents' does not exist. "
        "Run Code Unit 5 first."
    )

if len(aware_agents) == 0:
    raise RuntimeError(
        "No agents are socially aware, so no vote can occur."
    )


print("=" * 76)
print("CODE UNIT 7 — AGENTS VOTE ON THE SOCIAL COMPACT")
print("=" * 76)

print(
    "Eligible voters:",
    len(aware_agents)
)

print(
    "Constitutions on ballot:",
    len(constitutions)
)


# ------------------------------------------------------------
# 2. INSTITUTIONAL PREFERENCES BY SPECIALTY
# ------------------------------------------------------------
# Each specialty has a preferred sharing threshold.
#
# Lower value:
#   greater willingness to share discoveries early.
#
# Higher value:
#   stronger preference for retaining information until
#   confidence is higher.
#
# These are pedagogical preference parameters, not empirical
# estimates.
# ------------------------------------------------------------

SPECIALTY_OPENNESS = {

    "finance": 0.55,

    "engineering": 0.62,

    "patents": 0.68,

    "gulf_strategy": 0.60,

    "ownership": 0.52,

    "synthesis": 0.72
}


# ------------------------------------------------------------
# 3. SAFE TEXT EXTRACTION
# ------------------------------------------------------------

def constitution_text(proposal):
    """
    Combine the substantive constitutional rules into one
    searchable text block.

    Using .get() prevents missing optional text from crashing
    the voting mechanism.
    """

    fields = [

        proposal.get(
            "join_rule",
            ""
        ),

        proposal.get(
            "verification_rule",
            ""
        ),

        proposal.get(
            "reward_rule",
            ""
        ),

        proposal.get(
            "merge_split_rule",
            ""
        ),

        proposal.get(
            "anti_herding_rule",
            ""
        )
    ]

    return " ".join(
        str(x) for x in fields
    ).lower()


# ------------------------------------------------------------
# 4. TEXTUAL INSTITUTIONAL SIGNALS
# ------------------------------------------------------------

VERIFICATION_WORDS = [
    "independent",
    "verify",
    "verified",
    "verification",
    "source",
    "evidence",
    "corroborate",
    "corroboration",
    "cross-check",
    "validate"
]

AUTONOMY_WORDS = [
    "voluntary",
    "consent",
    "choose",
    "choice",
    "opt",
    "autonomy",
    "autonomous",
    "permission",
    "withdraw"
]

DIVERSITY_WORDS = [
    "diverse",
    "diversity",
    "minority",
    "independent",
    "alternative",
    "dissent",
    "contrarian",
    "disagreement",
    "challenge"
]


def contains_any(text, vocabulary):
    """
    Return True if at least one institutional concept appears
    in the constitutional text.
    """

    return any(
        word in text
        for word in vocabulary
    )


# ------------------------------------------------------------
# 5. STABLE PROPOSAL IDENTIFIER
# ------------------------------------------------------------
# The original code assumed proposal_id always ended in a
# number. We remove that fragile dependency.
# ------------------------------------------------------------

def stable_proposal_number(proposal):
    """
    Convert proposal_id into a deterministic integer.

    Examples:
        CONST_1 -> 1
        CONST_2 -> 2

    If the identifier does not end in an integer, create a
    deterministic fallback from the characters.
    """

    proposal_id = str(
        proposal.get(
            "proposal_id",
            "CONST_0"
        )
    )

    try:

        return int(
            proposal_id.split("_")[-1]
        )

    except (ValueError, TypeError):

        return sum(
            ord(c)
            for c in proposal_id
        )


# ------------------------------------------------------------
# 6. CONSTITUTIONAL UTILITY FUNCTION
# ------------------------------------------------------------

def constitutional_utility(agent, proposal):
    """
    Evaluate one constitution from the perspective of one agent.

    Returns:
        total utility score
        diagnostic components

    The function is fully deterministic.
    """

    # --------------------------------------------------------
    # Preferred openness of this specialty
    # --------------------------------------------------------

    preferred_threshold = SPECIALTY_OPENNESS.get(
        agent.specialty,
        0.60
    )


    # --------------------------------------------------------
    # Constitution's proposed sharing threshold
    # --------------------------------------------------------

    try:

        threshold = float(
            proposal.get(
                "share_threshold",
                0.60
            )
        )

    except (TypeError, ValueError):

        threshold = 0.60


    # Keep threshold in valid range.
    threshold = max(
        0.0,
        min(
            1.0,
            threshold
        )
    )


    # --------------------------------------------------------
    # THRESHOLD FIT
    # --------------------------------------------------------

    threshold_fit = (
        1.0
        -
        abs(
            threshold
            -
            preferred_threshold
        )
    )


    # --------------------------------------------------------
    # Analyze substantive constitutional language
    # --------------------------------------------------------

    text = constitution_text(
        proposal
    )


    # --------------------------------------------------------
    # VERIFICATION
    # --------------------------------------------------------

    verification = (
        1.0
        if contains_any(
            text,
            VERIFICATION_WORDS
        )
        else 0.4
    )


    # --------------------------------------------------------
    # AUTONOMY
    # --------------------------------------------------------

    autonomy = (
        1.0
        if contains_any(
            text,
            AUTONOMY_WORDS
        )
        else 0.5
    )


    # --------------------------------------------------------
    # DIVERSITY / ANTI-HERDING
    # --------------------------------------------------------

    diversity = (
        1.0
        if contains_any(
            text,
            DIVERSITY_WORDS
        )
        else 0.5
    )


    # --------------------------------------------------------
    # DETERMINISTIC IDIOSYNCRATIC TERM
    # --------------------------------------------------------
    # This is NOT stochastic in the experimental sense.
    #
    # Every agent/proposal pair always receives exactly the
    # same perturbation for the same SEED.
    #
    # Its purpose is simply to prevent artificial perfect
    # bloc voting among agents with identical specialties.
    # --------------------------------------------------------

    agent_number = int(
        agent.agent_id[1:]
    )

    proposal_number = stable_proposal_number(
        proposal
    )

    local_seed = (
        SEED
        +
        agent_number * 17
        +
        proposal_number * 101
    )

    rng = random.Random(
        local_seed
    )

    idiosyncratic_term = rng.uniform(
        -0.08,
        0.08
    )


    # --------------------------------------------------------
    # TOTAL UTILITY
    # --------------------------------------------------------

    utility = (

        0.35 * threshold_fit

        +

        0.25 * verification

        +

        0.20 * autonomy

        +

        0.20 * diversity

        +

        idiosyncratic_term
    )


    # --------------------------------------------------------
    # Diagnostics
    # --------------------------------------------------------

    diagnostics = {

        "preferred_threshold":
            preferred_threshold,

        "proposal_threshold":
            threshold,

        "threshold_fit":
            threshold_fit,

        "verification":
            verification,

        "autonomy":
            autonomy,

        "diversity":
            diversity,

        "idiosyncratic_term":
            idiosyncratic_term,

        "utility":
            utility
    }


    return utility, diagnostics


# ------------------------------------------------------------
# 7. SCORE EVERY AGENT × CONSTITUTION PAIR
# ------------------------------------------------------------

evaluation_records = []

for agent in aware_agents:

    for proposal in constitutions:

        utility, diagnostics = (
            constitutional_utility(
                agent,
                proposal
            )
        )

        evaluation_records.append({

            "agent_id":
                agent.agent_id,

            "specialty":
                agent.specialty,

            "proposal_id":
                proposal["proposal_id"],

            "constitution_name":
                proposal["name"],

            **diagnostics
        })


evaluation_df = pd.DataFrame(
    evaluation_records
)


# ------------------------------------------------------------
# 8. EACH AGENT CASTS ONE VOTE
# ------------------------------------------------------------

ballots = []

for agent in aware_agents:

    agent_scores = (
        evaluation_df[
            evaluation_df["agent_id"]
            ==
            agent.agent_id
        ]
        .sort_values(
            [
                "utility",
                "proposal_id"
            ],
            ascending=[
                False,
                True
            ]
        )
        .reset_index(drop=True)
    )


    if agent_scores.empty:

        raise RuntimeError(
            f"No constitutional scores found "
            f"for agent {agent.agent_id}."
        )


    winner = agent_scores.iloc[0]

    vote = winner["proposal_id"]

    agent.constitution_vote = vote


    ballots.append({

        "agent_id":
            agent.agent_id,

        "specialty":
            agent.specialty,

        "vote":
            vote,

        "constitution_name":
            winner["constitution_name"],

        "utility":
            round(
                float(
                    winner["utility"]
                ),
                3
            )
    })


ballot_df = pd.DataFrame(
    ballots
)


# ------------------------------------------------------------
# 9. COUNT VOTES
# ------------------------------------------------------------

vote_counts = (
    ballot_df
    .groupby(
        "vote"
    )
    .size()
    .reset_index(
        name="votes"
    )
    .rename(
        columns={
            "vote":
                "proposal_id"
        }
    )
)


# ------------------------------------------------------------
# 10. ADD MEAN UTILITY AS TRANSPARENT TIE-BREAKER
# ------------------------------------------------------------
# If two constitutions receive the same number of votes,
# choose the one with the highest mean utility across ALL
# socially aware agents.
#
# If that is also tied, proposal_id provides a deterministic
# final ordering.
# ------------------------------------------------------------

mean_utility = (
    evaluation_df
    .groupby(
        "proposal_id"
    )["utility"]
    .mean()
    .reset_index(
        name="mean_utility"
    )
)


vote_results = (
    vote_counts
    .merge(
        mean_utility,
        on="proposal_id",
        how="left"
    )
    .sort_values(
        [
            "votes",
            "mean_utility",
            "proposal_id"
        ],
        ascending=[
            False,
            False,
            True
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 11. IDENTIFY WINNING CONSTITUTION
# ------------------------------------------------------------

winning_id = (
    vote_results.iloc[0][
        "proposal_id"
    ]
)


constitution = next(
    (
        proposal
        for proposal in constitutions
        if proposal["proposal_id"]
        ==
        winning_id
    ),
    None
)


if constitution is None:

    raise RuntimeError(
        f"Winning constitution "
        f"{winning_id} could not be found."
    )


# ------------------------------------------------------------
# 12. DISPLAY VOTE
# ------------------------------------------------------------

print("\n")
print("=" * 76)
print("PHASE VII — CONSTITUTIONAL VOTE")
print("=" * 76)


print("\nVOTE RESULTS")

display(
    vote_results.assign(
        mean_utility=lambda x:
        x["mean_utility"].round(3)
    )
)


print("\nINDIVIDUAL BALLOTS")

display(
    ballot_df.sort_values(
        [
            "specialty",
            "agent_id"
        ]
    )
)


# ------------------------------------------------------------
# 13. DISPLAY ADOPTED CONSTITUTION
# ------------------------------------------------------------

print("\n")
print("=" * 76)
print("ADOPTED CONSTITUTION")
print("=" * 76)

print(
    "\nProposal:",
    constitution["proposal_id"]
)

print(
    "Name:",
    constitution["name"]
)

print(
    "Principle:",
    constitution["principle"]
)

print(
    "Proposer:",
    constitution["proposer"],
    "|",
    constitution["proposer_specialty"]
)

print(
    "Share threshold:",
    constitution["share_threshold"]
)


print("\nFULL CONSTITUTION")

print(
    json.dumps(
        constitution,
        indent=2,
        ensure_ascii=False
    )
)


# ------------------------------------------------------------
# 14. PEDAGOGICAL SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 76)
print("CODE UNIT 7 SUMMARY")
print("=" * 76)

print(
    "Eligible voters:",
    len(aware_agents)
)

print(
    "Constitutions evaluated:",
    len(constitutions)
)

print(
    "Total agent × constitution evaluations:",
    len(evaluation_df)
)

print(
    "Votes cast:",
    len(ballot_df)
)

print(
    "Winning constitution:",
    constitution["name"]
)

print(
    "Winning proposal:",
    winning_id
)


winning_row = vote_results.iloc[0]

print(
    "Votes received:",
    int(
        winning_row["votes"]
    )
)

print(
    "Mean utility:",
    round(
        float(
            winning_row["mean_utility"]
        ),
        3
    )
)


print("\n" + "-" * 76)

print(
    "CODE UNIT 7 COMPLETE\n\n"
    "The LLM created competing institutional possibilities.\n"
    "The agents have now selected among them using transparent "
    "local preferences.\n\n"
    "Institutional creativity was generative.\n"
    "Institutional choice was deterministic.\n\n"
    "The winning constitution can now become an operational "
    "rule system governing collaboration."
)
````

## Code Unit 8 — The Adopted Constitution Governs Sharing and Coalition Formation

This unit tests whether constitutional choice has real consequences. A constitution is meaningful only if it changes behavior. The adopted social compact is therefore translated into operational thresholds and rules that govern evidence sharing, complementary responses, coalition membership, and related collaboration decisions.

The conceptual sequence now becomes recursive. Earlier discoveries created the need for collaboration. Collaboration created the need for rules. The selected rules now feed back into the architecture of collaboration. Institutional design becomes an input into network formation.

For example, a constitution with a high sharing threshold may preserve confidentiality but reduce the amount of information available to potential collaborators. A more open constitution may accelerate collective synthesis while increasing the risk of herding or information leakage. Verification requirements may slow coalition growth but improve evidentiary quality. Reward rules can affect which agents find participation attractive. Merge and split rules influence whether the network concentrates around one hypothesis or preserves multiple competing structures.

The code operationalizes these ideas in a deliberately simplified manner. This is not intended as a complete theory of constitutional economics for artificial agents. The purpose is to make one causal idea visible: **rules alter the graph**.

That insight distinguishes NB04 from a conventional multi-agent workflow. In a fixed workflow, architecture is a parameter chosen by the designer. Here architecture is partly an outcome of an institutional process that occurred inside the system.

The unit also prepares the final graph by linking agents through complementarities under the constraints of the adopted constitution. Coalition formation is therefore neither purely spontaneous nor centrally assigned. It is spontaneous **within a self-selected rule system**.

This is a more realistic conception of organization. Human institutions rarely consist of unconstrained individuals connecting arbitrarily. Relationships form inside rules, norms, incentives, and permissions. NB04 begins to give artificial coalitions the same layered structure.


````python
#  THE ADOPTED CONSTITUTION NOW GOVERNS SHARING
# ----------------------------
share_threshold = float(constitution["share_threshold"])
public_signals = []

for d in discoveries:
    if d["relevance_probability"] >= share_threshold:
        public_signals.append({
            "signal_id":f"S{len(public_signals):03d}",
            "sender":d["agent_id"],
            "specialty":d["specialty"],
            "company_id":d["company_id"],
            "value":d["discovery_value"],
            "confidence":d["relevance_probability"],
            "needed":d["missing_expertise"],
            "reason":d["reason"]
        })

signal_df = pd.DataFrame(public_signals)

# Coalition joining is now generated by complementarity + adopted constitution.
coalition_edges = []
for s in public_signals:
    candidates = [
        a for a in aware_agents
        if a.agent_id != s["sender"]
        and a.specialty == s["needed"]
    ]
    for a in candidates[:2]:
        coalition_edges.append({
            "company_id":s["company_id"],
            "u":s["sender"],
            "v":a.agent_id,
            "origin_signal":s["signal_id"],
            "constitution":constitution["name"]
        })
        agent_map[s["sender"]].coalition.append(a.agent_id)
        a.coalition.append(s["sender"])

edge_df = pd.DataFrame(coalition_edges)

print("\nPHASE VIII — POST-CONSTITUTION COLLABORATION")
print("Public signals under adopted sharing threshold:", len(signal_df))
print("Coalition edges:", len(edge_df))
if len(signal_df):
    display(signal_df.head(15))
if len(edge_df):
    display(edge_df.head(15))

# ----------------------------
````

## Code Unit 9 — Emergent Graph and Collective Hypothesis

After search, semantic valuation, pheromones, social discovery, public communication, constitutional design, voting, and governed sharing, the system can finally be viewed as a network. This unit constructs the emergent collaboration graph and computes the collective hypothesis produced by the swarm.

The graph is important because it is the visible organizational artifact of the preceding process. Nodes represent agents. Edges represent relationships that became meaningful under the adopted constitution. The network was not drawn first and populated later. It is the accumulated consequence of discoveries, competence gaps, social awareness, institutional proposals, collective choice, and constitutional constraints.

The collective score similarly combines more than raw evidence. It incorporates LLM-valued discoveries, diversity, and social activation. The target hypothesis is therefore produced by a system whose epistemic and institutional states have co-evolved.

This is the point at which the phrase **“the problem helps produce the organization that solves the problem”** becomes concrete. The acquisition mandate created distributed uncertainty. Distributed uncertainty generated valuable local discoveries. Those discoveries exposed missing expertise. Environmental traces revealed other minds. Other minds made communication useful. Communication created a constitutional problem. Constitutional choice shaped coalition formation. The resulting coalition then produced the collective hypothesis.

The causal direction is therefore not simply organization-to-solution. It is partially circular:

$$ \text{Problem} \rightarrow \text{Discovery} \rightarrow \text{Organization} \rightarrow \text{Rules} \rightarrow \text{Reorganized Intelligence} \rightarrow \text{Solution}. $$

Pedagogically, students should inspect the graph not merely as a visualization but as a historical record. Every edge has an institutional genealogy. It exists because earlier stages made that relationship possible and legitimate. This is what makes the network endogenous in a richer sense than a dynamically routed workflow.

````python
# EMERGENT GRAPH + COLLECTIVE HYPOTHESIS
# ----------------------------
import networkx as nx

G = nx.Graph()
for a in agents:
    G.add_node(a.agent_id,specialty=a.specialty)

for _,e in edge_df.iterrows():
    if G.has_edge(e.u,e.v):
        G[e.u][e.v]["weight"] += 1
    else:
        G.add_edge(e.u,e.v,weight=1,company=e.company_id)

# Collective score uses LLM discovery value + diversity + social activation.
if len(discovery_df):
    base = discovery_df.groupby("company_id").agg(
        llm_value=("discovery_value","sum"),
        mean_probability=("relevance_probability","mean"),
        specialties=("specialty","nunique"),
        discoveries=("doc_id","count")
    ).reset_index()
else:
    base = pd.DataFrame()

sig_count = (
    signal_df.groupby("company_id").size().rename("signals").reset_index()
    if len(signal_df) else pd.DataFrame(columns=["company_id","signals"])
)
edge_count = (
    edge_df.groupby("company_id").size().rename("coalition_edges").reset_index()
    if len(edge_df) else pd.DataFrame(columns=["company_id","coalition_edges"])
)

ranked = (
    base.merge(sig_count,on="company_id",how="left")
        .merge(edge_count,on="company_id",how="left")
        .fillna(0)
)

ranked["collective_score"] = (
    ranked["llm_value"]
    + 18*ranked["mean_probability"]
    + 10*ranked["specialties"]
    + 8*ranked["signals"]
    + 6*ranked["coalition_edges"]
)

ranked = ranked.sort_values("collective_score",ascending=False).reset_index(drop=True)

winner = ranked.iloc[0].company_id if len(ranked) else None
success = winner == HIDDEN_TARGET

print("\nPHASE IX — COLLECTIVE DISCOVERY")
display(ranked)
print("WINNER:",winner)
print("HIDDEN TARGET:",HIDDEN_TARGET)
print("SUCCESS:",success)
print("LLM calls used:",api_calls,"/",MAX_API_CALLS)

# ----------------------------
````

# A Narrative of the Constitutional Swarm: From Isolated Intelligence to Artificial Institution

This notebook begins with a simple acquisition problem, but its real purpose is much more ambitious. It asks what might happen if a population of artificial-intelligence agents were allowed not merely to solve a problem together, but to **discover for themselves that collaboration is necessary, develop rules for that collaboration, and eventually create something resembling an institution**.

The setting is deliberately familiar. Imagine an investment bank searching for an acquisition target in the water-technology industry. The client wants a company of a certain financial size, with manageable debt, strong desalination technology, valuable intellectual property, access to Gulf markets, and owners who might realistically be willing to sell.

There are several candidate companies, but one of them is secretly designated as the best match. The agents do not know which one it is.

Eighteen artificial agents are released into this information environment. They possess different specialties. Some understand finance. Others understand engineering, patents, Gulf markets, ownership issues, or the broader task of synthesizing evidence.

But there is an unusual condition: **the agents initially believe they are alone**.

There is no team meeting. There is no manager assigning tasks. There is no central intelligence telling one agent to examine patents and another to investigate ownership. Each agent independently searches the available information according to its own expertise.

This produces the first important feature of the experiment. Different agents notice different things.

A financial specialist may recognize that a company has attractive earnings and low leverage. An engineering specialist may notice that its membranes perform well under demanding desalination conditions. A patent specialist may recognize unusually valuable intellectual property. Another agent may notice evidence of commercial progress in Saudi Arabia and the United Arab Emirates. An ownership specialist may discover that a family is considering succession and may therefore be open to selling.

None of these observations alone solves the acquisition problem.

The interesting question is how isolated fragments of intelligence become collective knowledge.

At this point, an advanced language model enters the experiment. It does not search the entire database and it does not choose the acquisition target. Instead, it performs a much narrower task: it evaluates the **meaning and potential importance of individual discoveries**.

The system first filters observations using conventional computational methods. Only a small number of potentially interesting discoveries are presented to the language model. The model then asks, in effect: How relevant is this discovery? How novel is it? Why might it matter? And what kind of expertise would be useful next?

A particularly valuable discovery leaves behind something the notebook calls a **pheromone**.

The metaphor comes from biological systems. An ant does not need to send another ant a detailed message explaining where it has been. It can leave a trace in the environment. Another ant encounters that trace later and changes its behavior.

The artificial agents do something similar.

A valuable discovery leaves a short, anonymous informational trace. Crucially, the trace does not initially reveal who created it. It is not an email, a chat message, or an instruction.

Eventually another agent encounters one of these traces.

This is the pivotal moment in the notebook.

The agent realizes:

**“Another intelligent searcher has been here.”**

For the first time, an agent that believed itself to be alone has evidence that other minds inhabit its environment. As more agents make the same discovery, something resembling social awareness begins to emerge.

The system contains the possibility of a bulletin board, but initially that capability is socially meaningless. Why would an agent communicate if it believes nobody else exists?

Once enough agents discover evidence of other minds, the bulletin board becomes relevant. The agents can now communicate publicly.

Their first concern is not simply exchanging information. It is more fundamental: **if we are going to collaborate, under what rules should we collaborate?**

This creates what might be called a constitutional moment.

Several agents with different backgrounds are invited to propose possible constitutions. Here the language model is used again, but this time for institutional creativity. The agents propose rules governing when discoveries should be shared, how new members may join a coalition, how evidence should be verified, how credit should be allocated, when groups may merge or separate, and how the system should protect itself against premature consensus.

There is still no central manager choosing the rules.

Instead, the population votes.

Different agents evaluate the proposed constitutions according to their own interests and professional perspectives. A financial agent may prefer one degree of openness, while a patent specialist may be more cautious about sharing information. Some agents place greater value on verification; others care particularly about autonomy or diversity of opinion.

One constitution eventually receives sufficient collective support and becomes the social compact.

This changes the nature of the system.

The agents are no longer merely independent intelligences operating in the same environment. They now possess collectively selected rules governing how they may cooperate.

Under those rules, coalitions begin to form around promising companies. Agents with complementary expertise connect. Evidence from finance, technology, patents, markets, and ownership begins to accumulate. The resulting network of relationships was not specified in advance. It emerges from the discoveries made by the agents and from the constitution they collectively adopted.

Finally, the system combines the accumulated evidence and produces a collective ranking of acquisition candidates. The hidden target emerges as the strongest collective hypothesis.

But finding the target is not the notebook's deepest result.

The deeper result is the **organization that appeared while searching for it**.

Earlier multi-agent systems might begin with a designed organization: predefined teams, workflows, responsibilities, and communication channels. Here the direction is reversed.

The agents begin as isolated intelligences.

They discover information.

Information leaves traces.

Traces reveal the existence of other minds.

The discovery of other minds creates a reason to communicate.

Communication creates a need for rules.

Agents propose those rules.

The population chooses among them.

The chosen rules enable coalitions.

The coalitions produce collective intelligence.

The notebook therefore moves through three increasingly powerful ideas: **endogenous coalition, endogenous constitution, and endogenous institution**.

The environment still provides basic infrastructure. The agents do not create the computer, the documents, or the language model itself. Complete institutional emergence from nothing is neither claimed nor demonstrated.

What changes is something subtler and more important.

The designer increasingly stops specifying **who should collaborate with whom and how that collaboration must occur**. Those decisions begin to arise inside the artificial system itself.

This leads to the central idea of the notebook:

**The problem does not merely determine the answer. The problem can help produce the organization that discovers the answer.**

That is the conceptual leap represented by the Constitutional Swarm. Artificial intelligence is no longer being considered only as a collection of increasingly capable individual models. We are beginning to ask what happens when artificial minds can discover one another, communicate, negotiate rules, organize themselves, and create temporary institutions adapted to the problems they encounter.

At that point, the subject is no longer simply artificial intelligence.

It is the beginning of **artificial society**.
